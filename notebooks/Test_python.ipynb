## hydrol_diag_soil

```python
import os
import time
import functools
import numpy as np
from scipy.io import FortranFile
# from global_module import Global_module

def read_dummy(avan, evapot, frac_snow_nobio, irrigation, ks, mcfc, mcr, mcs, mcw, njsc, nvan, precip_rain, reinfiltration, returnflow, soiltile, tot_melt, totfrac_nobio, us, veget_max, vevapnu):
    print(f'--- inside the read dummy routine for hydrol_diag_soil ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_diag_soil/dummy.bin'
    ffile = FortranFile(path, 'r')
    for var in [avan, evapot, frac_snow_nobio, irrigation, ks, mcfc, mcr, mcs, mcw, njsc, nvan, precip_rain, reinfiltration, returnflow, soiltile, tot_melt, totfrac_nobio, us, veget_max, vevapnu]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                var[:] = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif var.dtype == np.int32:
                var[:] = ffile.read_ints(np.int32).reshape(arr_shape, order='F')

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        return result
    return wrapper

@timer
def hydrol_diag_soil(gm, ks, nvan, avan, mcr, mcs, mcfc, mcw, veget_max, soiltile, njsc, runoff, drainage, evapot, vevapnu, returnflow, reinfiltration, irrigation, shumdiag, shumdiag_perma, k_litt, litterhumdiag, humrel, vegstress, drysoil_frac, tot_melt, us, precip_rain, totfrac_nobio, frac_snow_nobio):
    mask_vegtot = np.zeros((gm.kjpindex,), dtype=np.float64)
    for jst in range(0, gm.nstm, 1):
        for ji in range(0, gm.kjpindex, 1):
            gm.ae_ns[ji, jst] = gm.ae_ns[ji, jst] * gm.mask_soiltile[ji, jst]
            gm.dr_ns[ji, jst] = gm.dr_ns[ji, jst] * gm.mask_soiltile[ji, jst]
            gm.ru_ns[ji, jst] = gm.ru_ns[ji, jst] * gm.mask_soiltile[ji, jst]
            gm.tmc[ji, jst] = gm.tmc[ji, jst] * gm.mask_soiltile[ji, jst]
            for jv in range(0, gm.nvm, 1):
                gm.humrelv[ji, jv, jst] = gm.humrelv[ji, jv, jst] * gm.mask_soiltile[ji, jst]
                for jsl in range(0, gm.nslm, 1):
                    us[ji, jv, jst, jsl] = us[ji, jv, jst, jsl] * gm.mask_soiltile[ji, jst]
            for jsl in range(0, gm.nslm, 1):
                gm.mc[ji, jsl, jst] = gm.mc[ji, jsl, jst] * gm.mask_soiltile[ji, jst]
    runoff[:] = gm.zero
    drainage[:] = gm.zero
    gm.humtot[:] = gm.zero
    shumdiag[:, :] = gm.zero
    shumdiag_perma[:, :] = gm.zero
    k_litt[:] = gm.zero
    litterhumdiag[:] = gm.zero
    gm.tmc_litt_dry_mea[:] = gm.zero
    gm.tmc_litt_wet_mea[:] = gm.zero
    gm.tmc_litt_mea[:] = gm.zero
    humrel[:, :] = gm.zero
    vegstress[:, :] = gm.zero
    if gm.ok_freeze_cwrr:
        gm.profil_froz_hydro[:, :] = gm.zero
    for ji in range(0, gm.kjpindex, 1):
        mask_vegtot[ji] = 0
        if gm.vegtot[ji] > gm.min_sechiba:
            mask_vegtot[ji] = 1
    for ji in range(0, gm.kjpindex, 1):
        gm.ae_ns[ji, :] = mask_vegtot[ji] * gm.ae_ns[ji, :] * gm.frac_bare_ns[ji, :]
    for jst in range(0, gm.nstm, 1):
        for ji in range(0, gm.kjpindex, 1):
            drainage[ji] = mask_vegtot[ji] * (drainage[ji] + gm.vegtot[ji] * soiltile[ji, jst] * gm.dr_ns[ji, jst])
            runoff[ji] = mask_vegtot[ji] * (runoff[ji] + gm.vegtot[ji] * soiltile[ji, jst] * gm.ru_ns[ji, jst]) + (1 - mask_vegtot[ji]) * (tot_melt[ji] + irrigation[ji] + returnflow[ji] + reinfiltration[ji])
            gm.humtot[ji] = mask_vegtot[ji] * (gm.humtot[ji] + gm.vegtot[ji] * soiltile[ji, jst] * gm.tmc[ji, jst])
            if gm.ok_freeze_cwrr:
                gm.profil_froz_hydro[ji, :] = mask_vegtot[ji] * (gm.profil_froz_hydro[ji, :] + gm.vegtot[ji] * soiltile[ji, jst] * gm.profil_froz_hydro_ns[ji, :, jst])
    for ji in range(0, gm.kjpindex, 1):
        if gm.vegtot[ji] != 0.0:
            vevapnu[ji] = vevapnu[ji] + gm.subsinksoil[ji] * gm.vegtot[ji]
        else:
            vevapnu[ji] = vevapnu[ji] + gm.subsinksoil[ji]
        runoff[ji] = runoff[ji] + precip_rain[ji] * totfrac_nobio[ji] * (1 - frac_snow_nobio[ji, gm.iice - 1])
    for jst in range(0, gm.nstm, 1):
        for jv in range(0, gm.nvm, 1):
            for ji in range(0, gm.kjpindex, 1):
                if veget_max[ji, jv] > gm.min_sechiba:
                    vegstress[ji, jv] = vegstress[ji, jv] + gm.vegstressv[ji, jv, jst]
                    vegstress[ji, jv] = np.maximum(vegstress[ji, jv], gm.zero)
    for jst in range(0, gm.nstm, 1):
        for jv in range(0, gm.nvm, 1):
            for ji in range(0, gm.kjpindex, 1):
                humrel[ji, jv] = humrel[ji, jv] + gm.humrelv[ji, jv, jst]
                humrel[ji, jv] = np.maximum(humrel[ji, jv], gm.zero)
    for jst in range(0, gm.nstm, 1):
        for ji in range(0, gm.kjpindex, 1):
            if gm.tmc_litter[ji, jst] < gm.tmc_litter_res[ji, jst]:
                i = gm.imin
            else:
                tmc_litter_ratio = (gm.tmc_litter[ji, jst] - gm.tmc_litter_res[ji, jst]) / (gm.tmc_litter_sat[ji, jst] - gm.tmc_litter_res[ji, jst])
                i = np.maximum(np.minimum(int((gm.imax - gm.imin) * tmc_litter_ratio) + gm.imin, gm.imax - 1), gm.imin)
            k_tmp = np.maximum(gm.k_lin[i - 1, 0, ji] * ks[ji], gm.zero)
            k_litt[ji] = k_litt[ji] + gm.vegtot[ji] * soiltile[ji, jst] * np.sqrt(k_tmp)
        for ji in range(0, gm.kjpindex, 1):
            litterhumdiag[ji] = litterhumdiag[ji] + gm.soil_wet_litter[ji, jst] * soiltile[ji, jst]
            gm.tmc_litt_wet_mea[ji] = gm.tmc_litt_wet_mea[ji] + gm.tmc_litter_awet[ji, jst] * soiltile[ji, jst]
            gm.tmc_litt_dry_mea[ji] = gm.tmc_litt_dry_mea[ji] + gm.tmc_litter_adry[ji, jst] * soiltile[ji, jst]
            gm.tmc_litt_mea[ji] = gm.tmc_litt_mea[ji] + gm.tmc_litter[ji, jst] * soiltile[ji, jst]
    for ji in range(0, gm.kjpindex, 1):
        if gm.tmc_litt_wet_mea[ji] - gm.tmc_litt_dry_mea[ji] > gm.zero:
            drysoil_frac[ji] = gm.un + np.maximum(np.minimum((gm.tmc_litt_dry_mea[ji] - gm.tmc_litt_mea[ji]) / (gm.tmc_litt_wet_mea[ji] - gm.tmc_litt_dry_mea[ji]), gm.zero), -gm.un)
        else:
            drysoil_frac[ji] = gm.zero
    gm.soilmoist[:, :] = gm.zero
    for jst in range(0, gm.nstm, 1):
        for ji in range(0, gm.kjpindex, 1):
            gm.soilmoist[ji, 0] = gm.soilmoist[ji, 0] + soiltile[ji, jst] * gm.dz[1] * (gm.trois * gm.mc[ji, 0, jst] + gm.mc[ji, 1, jst]) / gm.huit
            for jsl in range(1, gm.nslm - 1, 1):
                gm.soilmoist[ji, jsl] = gm.soilmoist[ji, jsl] + soiltile[ji, jst] * (gm.dz[jsl] * (gm.trois * gm.mc[ji, jsl, jst] + gm.mc[ji, jsl - 1, jst]) / gm.huit + gm.dz[jsl + 1] * (gm.trois * gm.mc[ji, jsl, jst] + gm.mc[ji, jsl + 1, jst]) / gm.huit)
            gm.soilmoist[ji, gm.nslm - 1] = gm.soilmoist[ji, gm.nslm - 1] + soiltile[ji, jst] * gm.dz[gm.nslm - 1] * (gm.trois * gm.mc[ji, gm.nslm - 1, jst] + gm.mc[ji, gm.nslm - 2, jst]) / gm.huit
    for ji in range(0, gm.kjpindex, 1):
        gm.soilmoist[ji, :] = gm.soilmoist[ji, :] * gm.vegtot[ji]
    gm.soilmoist_s[:, :, :] = gm.zero
    for jst in range(0, gm.nstm, 1):
        for ji in range(0, gm.kjpindex, 1):
            gm.soilmoist_s[ji, 0, gm.nstm - 1] = gm.soilmoist_s[ji, 0, gm.nstm - 1] + soiltile[ji, jst] * gm.dz[1] * (gm.trois * gm.mc[ji, 0, jst] + gm.mc[ji, 1, jst]) / gm.huit
            for jsl in range(1, gm.nslm - 1, 1):
                gm.soilmoist_s[ji, jsl, gm.nstm - 1] = gm.soilmoist_s[ji, jsl, gm.nstm - 1] + soiltile[ji, jst] * (gm.dz[jsl] * (gm.trois * gm.mc[ji, jsl, jst] + gm.mc[ji, jsl - 1, jst]) / gm.huit + gm.dz[jsl + 1] * (gm.trois * gm.mc[ji, jsl, jst] + gm.mc[ji, jsl + 1, jst]) / gm.huit)
            gm.soilmoist_s[ji, gm.nslm - 1, gm.nstm - 1] = gm.soilmoist_s[ji, gm.nslm - 1, gm.nstm - 1] + soiltile[ji, jst] * gm.dz[gm.nslm - 1] * (gm.trois * gm.mc[ji, gm.nslm - 1, jst] + gm.mc[ji, gm.nslm - 2, jst]) / gm.huit
    for ji in range(0, gm.kjpindex, 1):
        gm.soilmoist_s[ji, :, :] = gm.soilmoist_s[ji, :, :] * gm.vegtot[ji]
    gm.soilmoist_liquid[:, :] = gm.zero
    for jst in range(0, gm.nstm, 1):
        for ji in range(0, gm.kjpindex, 1):
            gm.soilmoist_liquid[ji, 0] = gm.soilmoist_liquid[ji, 0] + soiltile[ji, jst] * gm.dz[1] * (gm.trois * gm.mcl[ji, 0, jst] + gm.mcl[ji, 1, jst]) / gm.huit
            for jsl in range(1, gm.nslm - 1, 1):
                gm.soilmoist_liquid[ji, jsl] = gm.soilmoist_liquid[ji, jsl] + soiltile[ji, jst] * (gm.dz[jsl] * (gm.trois * gm.mcl[ji, jsl, jst] + gm.mcl[ji, jsl - 1, jst]) / gm.huit + gm.dz[jsl + 1] * (gm.trois * gm.mcl[ji, jsl, jst] + gm.mcl[ji, jsl + 1, jst]) / gm.huit)
            gm.soilmoist_liquid[ji, gm.nslm - 1] = gm.soilmoist_liquid[ji, gm.nslm - 1] + soiltile[ji, jst] * gm.dz[gm.nslm - 1] * (gm.trois * gm.mcl[ji, gm.nslm - 1, jst] + gm.mcl[ji, gm.nslm - 2, jst]) / gm.huit
    for ji in range(0, gm.kjpindex, 1):
        gm.soilmoist_liquid[ji, :] = gm.soilmoist_liquid[ji, :] * gm.vegtot_old[ji]
    for jst in range(0, gm.nstm, 1):
        for jsl in range(0, gm.nslm, 1):
            for ji in range(0, gm.kjpindex, 1):
                shumdiag[ji, jsl] = shumdiag[ji, jsl] + gm.soil_wet_ns[ji, jsl, jst] * soiltile[ji, jst] * ((mcs[ji] - mcw[ji]) / (mcfc[ji] - mcw[ji]))
                shumdiag[ji, jsl] = np.maximum(np.minimum(shumdiag[ji, jsl], gm.un), gm.zero)
    for jsl in range(0, gm.nslm, 1):
        for ji in range(0, gm.kjpindex, 1):
            shumdiag_perma[ji, jsl] = gm.soilmoist[ji, jsl] / (gm.dh[jsl] * mcs[ji])
            shumdiag_perma[ji, jsl] = np.maximum(np.minimum(shumdiag_perma[ji, jsl], gm.un), gm.zero)

def test_hydrol_diag_soil(gm, vevapnu, runoff, drysoil_frac, drainage, k_litt, vegstress, litterhumdiag, us, shumdiag_perma, shumdiag, humrel):
    print('--- inside the test function for hydrol_diag_soil ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_diag_soil/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['ae_ns', 'dr_ns', 'drainage', 'drysoil_frac', 'humrel', 'humrelv', 'humtot', 'k_litt', 'litterhumdiag', 'mc', 'profil_froz_hydro', 'ru_ns', 'runoff', 'shumdiag', 'shumdiag_perma', 'soilmoist', 'soilmoist_liquid', 'soilmoist_s', 'tmc', 'tmc_litt_dry_mea', 'tmc_litt_mea', 'tmc_litt_wet_mea', 'us', 'vegstress', 'vevapnu']
    for (variable, value) in zip(modif_var, [gm.ae_ns, gm.dr_ns, drainage, drysoil_frac, humrel, gm.humrelv, gm.humtot, k_litt, litterhumdiag, gm.mc, gm.profil_froz_hydro, gm.ru_ns, runoff, shumdiag, shumdiag_perma, gm.soilmoist, gm.soilmoist_liquid, gm.soilmoist_s, gm.tmc, gm.tmc_litt_dry_mea, gm.tmc_litt_mea, gm.tmc_litt_wet_mea, us, vegstress, vevapnu]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gm = Global_module()
    gm.declaration_initialization()
    avan = np.zeros((gm.kjpindex,), dtype=np.float64)
    drainage = np.zeros((gm.kjpindex,), dtype=np.float64)
    drysoil_frac = np.zeros((gm.kjpindex,), dtype=np.float64)
    evapot = np.zeros((gm.kjpindex,), dtype=np.float64)
    frac_snow_nobio = np.zeros((gm.kjpindex, gm.nnobio), dtype=np.float64)
    humrel = np.zeros((gm.kjpindex, gm.nvm), dtype=np.float64)
    irrigation = np.zeros((gm.kjpindex,), dtype=np.float64)
    k_litt = np.zeros((gm.kjpindex,), dtype=np.float64)
    ks = np.zeros((gm.kjpindex,), dtype=np.float64)
    litterhumdiag = np.zeros((gm.kjpindex,), dtype=np.float64)
    mcfc = np.zeros((gm.kjpindex,), dtype=np.float64)
    mcr = np.zeros((gm.kjpindex,), dtype=np.float64)
    mcs = np.zeros((gm.kjpindex,), dtype=np.float64)
    mcw = np.zeros((gm.kjpindex,), dtype=np.float64)
    njsc = np.zeros((gm.kjpindex,), dtype=np.int32)
    nvan = np.zeros((gm.kjpindex,), dtype=np.float64)
    precip_rain = np.zeros((gm.kjpindex,), dtype=np.float64)
    reinfiltration = np.zeros((gm.kjpindex,), dtype=np.float64)
    returnflow = np.zeros((gm.kjpindex,), dtype=np.float64)
    runoff = np.zeros((gm.kjpindex,), dtype=np.float64)
    shumdiag = np.zeros((gm.kjpindex, gm.nslm), dtype=np.float64)
    shumdiag_perma = np.zeros((gm.kjpindex, gm.nslm), dtype=np.float64)
    soiltile = np.zeros((gm.kjpindex, gm.nstm), dtype=np.float64)
    tot_melt = np.zeros((gm.kjpindex,), dtype=np.float64)
    totfrac_nobio = np.zeros((gm.kjpindex,), dtype=np.float64)
    us = np.zeros((gm.kjpindex, gm.nvm, gm.nstm, gm.nslm), dtype=np.float64)
    veget_max = np.zeros((gm.kjpindex, gm.nvm), dtype=np.float64)
    vegstress = np.zeros((gm.kjpindex, gm.nvm), dtype=np.float64)
    vevapnu = np.zeros((gm.kjpindex,), dtype=np.float64)
    read_dummy(avan, evapot, frac_snow_nobio, irrigation, ks, mcfc, mcr, mcs, mcw, njsc, nvan, precip_rain, reinfiltration, returnflow, soiltile, tot_melt, totfrac_nobio, us, veget_max, vevapnu)
    hydrol_diag_soil(gm, ks, nvan, avan, mcr, mcs, mcfc, mcw, veget_max, soiltile, njsc, runoff, drainage, evapot, vevapnu, returnflow, reinfiltration, irrigation, shumdiag, shumdiag_perma, k_litt, litterhumdiag, humrel, vegstress, drysoil_frac, tot_melt, us, precip_rain, totfrac_nobio, frac_snow_nobio)
    test_hydrol_diag_soil(gm, vevapnu, runoff, drysoil_frac, drainage, k_litt, vegstress, litterhumdiag, us, shumdiag_perma, shumdiag, humrel)
if __name__ == '__main__':
    main()
```

## hydrol_soil_froz

```python
import os
import time
import functools
import numpy as np
from scipy.io import FortranFile
# from global_module import Global_module

def read_dummy(gm, avan, ins, mcr, mcs, njsc, nvan, stempdiag):
    print(f'--- inside the read dummy routine for hydrol_soil_froz ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_froz/dummy.bin'
    ffile = FortranFile(path, 'r')
    avan[:] = ffile.read_reals(np.float64).reshape((gm.kjpindex,), order='F')
    ins = ffile.read_ints(np.int32)[0]
    for var in [mcr, mcs, njsc, nvan, stempdiag]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                var[:] = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif var.dtype == np.int32:
                var[:] = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
    return ins

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        return result
    return wrapper

@timer
def hydrol_soil_froz(gm, nvan, avan, mcr, mcs, ins, njsc, stempdiag):
    froz_frac_moy = np.zeros((gm.kjpindex,), dtype=np.float64)
    smtot_moy = np.zeros((gm.kjpindex,), dtype=np.float64)
    mc_ns = np.zeros((gm.kjpindex, gm.nslm), dtype=np.float64)
    for jsl in range(0, gm.nslm, 1):
        for ji in range(0, gm.kjpindex, 1):
            m = 1.0 - 1.0 / nvan[ji]
            if not gm.ok_thermodynamical_freezing or gm.mc[ji, jsl, ins - 1] < mcr[ji] + gm.min_sechiba:
                if stempdiag[ji, jsl] >= gm.fr_center + gm.fr_dT / 2.0:
                    x = 1.0
                elif stempdiag[ji, jsl] >= gm.fr_center - gm.fr_dT / 2.0 and stempdiag[ji, jsl] < gm.fr_center + gm.fr_dT / 2.0:
                    x = (stempdiag[ji, jsl] - (gm.fr_center - gm.fr_dT / 2.0)) / gm.fr_dT
                else:
                    x = 0.0
            elif gm.ok_thermodynamical_freezing:
                if stempdiag[ji, jsl] >= gm.fr_center + gm.fr_dT / 2.0:
                    x = 1.0
                elif stempdiag[ji, jsl] >= gm.fr_center - gm.fr_dT / 2.0 and stempdiag[ji, jsl] < gm.fr_center + gm.fr_dT / 2.0:
                    x = np.minimum((mcs[ji] - mcr[ji]) * ((2.2 * 1000.0 * avan[ji] * (gm.fr_center + gm.fr_dT / 2.0 - stempdiag[ji, jsl]) * gm.lhf / gm.ZeroCelsius / 10.0) ** nvan[ji] + 1.0) ** (-m) / (gm.mc[ji, jsl, ins - 1] - mcr[ji]), 1.0)
                else:
                    x = 0.0
            gm.profil_froz_hydro_ns[ji, jsl, ins - 1] = 1.0 - x
            mc_ns[ji, jsl] = gm.mc[ji, jsl, ins - 1] / mcs[ji]
    froz_frac_moy[:] = gm.zero
    denom = gm.zero
    for jsl in range(0, gm.nslm, 1):
        froz_frac_moy[:] = froz_frac_moy[:] + gm.dh[jsl] * gm.profil_froz_hydro_ns[:, jsl, ins - 1]
        denom = denom + gm.dh[jsl]
    froz_frac_moy[:] = froz_frac_moy[:] / denom
    smtot_moy[:] = gm.zero
    denom = gm.zero
    for jsl in range(0, gm.nslm - 1, 1):
        smtot_moy[:] = smtot_moy[:] + gm.dh[jsl] * mc_ns[:, jsl]
        denom = denom + gm.dh[jsl]
    smtot_moy[:] = smtot_moy[:] / denom
    for jsl in range(0, gm.nslm, 1):
        gm.profil_froz_hydro_ns[:, jsl, ins - 1] = np.minimum(gm.profil_froz_hydro_ns[:, jsl, ins - 1] * froz_frac_moy[:] ** gm.froz_frac_corr * smtot_moy[:] ** gm.smtot_corr, gm.max_froz_hydro)

def test_hydrol_soil_froz(gm):
    print('--- inside the test function for hydrol_soil_froz ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_froz/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['profil_froz_hydro_ns']
    for (variable, value) in zip(modif_var, [gm.profil_froz_hydro_ns]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gm = global_module_soil_froz()
    gm.declaration_initialization()
    avan = np.zeros((gm.kjpindex,), dtype=np.float64)
    ins = np.int32(0)
    mcr = np.zeros((gm.kjpindex,), dtype=np.float64)
    mcs = np.zeros((gm.kjpindex,), dtype=np.float64)
    njsc = np.zeros((gm.kjpindex,), dtype=np.int32)
    nvan = np.zeros((gm.kjpindex,), dtype=np.float64)
    stempdiag = np.zeros((gm.kjpindex, gm.nslm), dtype=np.float64)
    ins = read_dummy(gm, avan, ins, mcr, mcs, njsc, nvan, stempdiag)
    hydrol_soil_froz(gm, nvan, avan, mcr, mcs, ins, njsc, stempdiag)
    test_hydrol_soil_froz(gm)
if __name__ == '__main__':
    main()
```

```python
import os
import time
import functools
import numpy as np
from scipy.io import FortranFile
# from global_module import Global_module

def read_dummy(ins, mcr, mcs, njsc):
    print(f'--- inside the read dummy routine for hydrol_soil_coef ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_coef/dummy.bin'
    ffile = FortranFile(path, 'r')
    ins = ffile.read_ints(np.int32)[0]
    for var in [mcr, mcs, njsc]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                var[:] = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif var.dtype == np.int32:
                var[:] = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
    return ins

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        return result
    return wrapper

@timer
def hydrol_soil_coef(gm, mcr, mcs, ins, njsc):
    if gm.ok_freeze_cwrr:
        for jsl in range(0, gm.nslm, 1):
            for ji in range(0, gm.kjpindex, 1):
                x = 1.0 - gm.profil_froz_hydro_ns[ji, jsl, ins - 1]
                mc_used = mcr[ji] + x * np.maximum(gm.mc[ji, jsl, ins - 1] - mcr[ji], gm.zero)
                i = np.maximum(gm.imin, np.minimum(gm.imax - 1, int(gm.imin + (gm.imax - gm.imin) * (mc_used - mcr[ji]) / (mcs[ji] - mcr[ji]))))
                gm.a[ji, jsl] = gm.a_lin[i - 1, jsl, ji] * gm.kfact_root[ji, jsl, ins - 1]
                gm.b[ji, jsl] = gm.b_lin[i - 1, jsl, ji] * gm.kfact_root[ji, jsl, ins - 1]
                gm.d[ji, jsl] = gm.d_lin[i - 1, jsl, ji] * gm.kfact_root[ji, jsl, ins - 1]
                gm.k[ji, jsl] = gm.kfact_root[ji, jsl, ins - 1] * np.maximum(gm.k_lin[gm.imin, jsl, ji], gm.a_lin[i - 1, jsl, ji] * mc_used + gm.b_lin[i - 1, jsl, ji])
    else:
        for jsl in range(0, gm.nslm, 1):
            for ji in range(0, gm.kjpindex, 1):
                mc_ratio = np.maximum(gm.mc[ji, jsl, ins - 1] - mcr[ji], gm.zero) / (mcs[ji] - mcr[ji])
                i = np.maximum(np.minimum(int((gm.imax - gm.imin) * mc_ratio) + gm.imin, gm.imax - 1), gm.imin)
                gm.a[ji, jsl] = gm.a_lin[i - 1, jsl, ji] * gm.kfact_root[ji, jsl, ins - 1]
                gm.b[ji, jsl] = gm.b_lin[i - 1, jsl, ji] * gm.kfact_root[ji, jsl, ins - 1]
                gm.d[ji, jsl] = gm.d_lin[i - 1, jsl, ji] * gm.kfact_root[ji, jsl, ins - 1]
                gm.k[ji, jsl] = gm.kfact_root[ji, jsl, ins - 1] * np.maximum(gm.k_lin[gm.imin, jsl, ji], gm.a_lin[i - 1, jsl, ji] * gm.mc[ji, jsl, ins - 1] + gm.b_lin[i - 1, jsl, ji])

def test_hydrol_soil_coef(gm):
    print('--- inside the test function for hydrol_soil_coef ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_coef/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['a', 'b', 'd', 'k']
    for (variable, value) in zip(modif_var, [gm.a, gm.b, gm.d, gm.k]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gm = global_module_soil_coef()
    gm.declaration_initialization()
    ins = np.int32(0)
    mcr = np.zeros((gm.kjpindex,), dtype=np.float64)
    mcs = np.zeros((gm.kjpindex,), dtype=np.float64)
    njsc = np.zeros((gm.kjpindex,), dtype=np.int32)
    ins = read_dummy(ins, mcr, mcs, njsc)
    hydrol_soil_coef(gm, mcr, mcs, ins, njsc)
    test_hydrol_soil_coef(gm)
if __name__ == '__main__':
    main()
```

## hydrol_soil_setup

```python
import os
import time
import functools
import numpy as np
from scipy.io import FortranFile
# from global_module import Global_module

def read_dummy(ins):
    print(f'--- inside the read dummy routine for hydrol_soil_setup ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_setup/dummy.bin'
    ffile = FortranFile(path, 'r')
    ins = ffile.read_ints(np.int32)[0]
    return ins

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        return result
    return wrapper

@timer
def hydrol_soil_setup(gm, ins):
    temp3 = gm.w_time * (gm.dt_sechiba / gm.one_day) / gm.deux
    temp4 = (gm.un - gm.w_time) * (gm.dt_sechiba / gm.one_day) / gm.deux
    for ji in range(0, gm.kjpindex, 1):
        gm.e[ji, 0] = gm.zero
        gm.f[ji, 0] = gm.trois * gm.dz[1] / gm.huit + temp3 * ((gm.d[ji, 0] + gm.d[ji, 1]) / gm.dz[1] + gm.a[ji, 0])
        gm.g1[ji, 0] = gm.dz[1] / gm.huit - temp3 * ((gm.d[ji, 0] + gm.d[ji, 1]) / gm.dz[1] - gm.a[ji, 1])
        gm.ep[ji, 0] = gm.zero
        gm.fp[ji, 0] = gm.trois * gm.dz[1] / gm.huit - temp4 * ((gm.d[ji, 0] + gm.d[ji, 1]) / gm.dz[1] + gm.a[ji, 0])
        gm.gp[ji, 0] = gm.dz[1] / gm.huit + temp4 * ((gm.d[ji, 0] + gm.d[ji, 1]) / gm.dz[1] - gm.a[ji, 1])
    for jsl in range(1, gm.nslm - 1, 1):
        for ji in range(0, gm.kjpindex, 1):
            gm.e[ji, jsl] = gm.dz[jsl] / gm.huit - temp3 * ((gm.d[ji, jsl] + gm.d[ji, jsl - 1]) / gm.dz[jsl] + gm.a[ji, jsl - 1])
            gm.f[ji, jsl] = gm.trois * (gm.dz[jsl] + gm.dz[jsl + 1]) / gm.huit + temp3 * ((gm.d[ji, jsl] + gm.d[ji, jsl - 1]) / gm.dz[jsl] + (gm.d[ji, jsl] + gm.d[ji, jsl + 1]) / gm.dz[jsl + 1])
            gm.g1[ji, jsl] = gm.dz[jsl + 1] / gm.huit - temp3 * ((gm.d[ji, jsl] + gm.d[ji, jsl + 1]) / gm.dz[jsl + 1] - gm.a[ji, jsl + 1])
            gm.ep[ji, jsl] = gm.dz[jsl] / gm.huit + temp4 * ((gm.d[ji, jsl] + gm.d[ji, jsl - 1]) / gm.dz[jsl] + gm.a[ji, jsl - 1])
            gm.fp[ji, jsl] = gm.trois * (gm.dz[jsl] + gm.dz[jsl + 1]) / gm.huit - temp4 * ((gm.d[ji, jsl] + gm.d[ji, jsl - 1]) / gm.dz[jsl] + (gm.d[ji, jsl] + gm.d[ji, jsl + 1]) / gm.dz[jsl + 1])
            gm.gp[ji, jsl] = gm.dz[jsl + 1] / gm.huit + temp4 * ((gm.d[ji, jsl] + gm.d[ji, jsl + 1]) / gm.dz[jsl + 1] - gm.a[ji, jsl + 1])
    for ji in range(0, gm.kjpindex, 1):
        gm.e[ji, gm.nslm - 1] = gm.dz[gm.nslm - 1] / gm.huit - temp3 * ((gm.d[ji, gm.nslm - 1] + gm.d[ji, gm.nslm - 2]) / gm.dz[gm.nslm - 1] + gm.a[ji, gm.nslm - 2])
        gm.f[ji, gm.nslm - 1] = gm.trois * gm.dz[gm.nslm - 1] / gm.huit + temp3 * ((gm.d[ji, gm.nslm - 1] + gm.d[ji, gm.nslm - 2]) / gm.dz[gm.nslm - 1] - gm.a[ji, gm.nslm - 1] * (gm.un - gm.deux * gm.free_drain_coef[ji, ins - 1]))
        gm.g1[ji, gm.nslm - 1] = gm.zero
        gm.ep[ji, gm.nslm - 1] = gm.dz[gm.nslm - 1] / gm.huit + temp4 * ((gm.d[ji, gm.nslm - 1] + gm.d[ji, gm.nslm - 2]) / gm.dz[gm.nslm - 1] + gm.a[ji, gm.nslm - 2])
        gm.fp[ji, gm.nslm - 1] = gm.trois * gm.dz[gm.nslm - 1] / gm.huit - temp4 * ((gm.d[ji, gm.nslm - 1] + gm.d[ji, gm.nslm - 2]) / gm.dz[gm.nslm - 1] - gm.a[ji, gm.nslm - 1] * (gm.un - gm.deux * gm.free_drain_coef[ji, ins - 1]))
        gm.gp[ji, gm.nslm - 1] = gm.zero

def test_hydrol_soil_setup(gm):
    print('--- inside the test function for hydrol_soil_setup ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_setup/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['e', 'ep', 'f', 'fp', 'g1', 'gp']
    for (variable, value) in zip(modif_var, [gm.e, gm.ep, gm.f, gm.fp, gm.g1, gm.gp]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gm = global_module_soil_setup()
    gm.declaration_initialization()
    ins = np.int32(0)
    ins = read_dummy(ins)
    hydrol_soil_setup(gm, ins)
    test_hydrol_soil_setup(gm)
if __name__ == '__main__':
    main()
```

```python
import os
import time
import functools
import numpy as np
from scipy.io import FortranFile
# from global_module import Global_module

def read_dummy(e_frac, evap_bare_lim, evap_bare_lim_ns, F_absorption, humrel, soiltile, tot_bare_soil, transpir, us, veget_max, vevapnu):
    print(f'--- inside the read dummy routine for hydrol_split_soil ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_split_soil/dummy.bin'
    ffile = FortranFile(path, 'r')
    for var in [e_frac, evap_bare_lim, evap_bare_lim_ns, F_absorption, humrel, soiltile, tot_bare_soil, transpir, us, veget_max, vevapnu]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                var[:] = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif var.dtype == np.int32:
                var[:] = ffile.read_ints(np.int32).reshape(arr_shape, order='F')

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        return result
    return wrapper

@timer
def hydrol_split_soil(gm, veget_max, soiltile, vevapnu, transpir, humrel, evap_bare_lim, evap_bare_lim_ns, tot_bare_soil, us, e_frac, F_absorption):
    tmp_check1 = np.zeros((gm.kjpindex,), dtype=np.float64)
    tmp_check2 = np.zeros((gm.kjpindex,), dtype=np.float64)
    tmp_check3 = np.zeros((gm.kjpindex, gm.nstm), dtype=np.float64)
    gm.precisol_ns[:, :] = gm.zero
    for jv in range(0, gm.nvm, 1):
        for ji in range(0, gm.kjpindex, 1):
            jst = gm.pref_soil_veg[jv] - 1
            if veget_max[ji, jv] > gm.min_sechiba and soiltile[ji, jst] * gm.vegtot[ji] > gm.min_sechiba:
                gm.precisol_ns[ji, jst] = gm.precisol_ns[ji, jst] + gm.precisol[ji, jv] / (soiltile[ji, jst] * gm.vegtot[ji])
    gm.ae_ns[:, :] = gm.zero
    for jst in range(0, gm.nstm, 1):
        for ji in range(0, gm.kjpindex, 1):
            if evap_bare_lim[ji] > gm.min_sechiba:
                gm.ae_ns[ji, jst] = vevapnu[ji] * evap_bare_lim_ns[ji, jst] / evap_bare_lim[ji]
    gm.tr_ns[:, :] = gm.zero
    for jv in range(0, gm.nvm, 1):
        jst = gm.pref_soil_veg[jv] - 1
        for ji in range(0, gm.kjpindex, 1):
            if humrel[ji, jv] > gm.min_sechiba and soiltile[ji, jst] * gm.vegtot[ji] > gm.min_sechiba:
                gm.tr_ns[ji, jst] = gm.tr_ns[ji, jst] + transpir[ji, jv] * (gm.humrelv[ji, jv, jst] / humrel[ji, jv]) / (soiltile[ji, jst] * gm.vegtot[ji])
    gm.rootsink[:, :, :] = gm.zero
    if gm.ok_hydrol_arch:
        for jv in range(0, gm.nvm, 1):
            jst = gm.pref_soil_veg[jv] - 1
            for jsl in range(0, gm.nslm, 1):
                for ji in range(0, gm.kjpindex, 1):
                    if humrel[ji, jv] > gm.min_sechiba and soiltile[ji, jst] * gm.vegtot[ji] > gm.min_sechiba:
                        if gm.is_tuzet_hydrol_arch:
                            gm.rootsink[ji, jsl, jst] = gm.rootsink[ji, jsl, jst] + F_absorption[ji, jv] * e_frac[ji, jv, jsl, jst] * gm.dt_sechiba * gm.kilo_to_unit
                        else:
                            gm.rootsink[ji, jsl, jst] = gm.rootsink[ji, jsl, jst] + transpir[ji, jv] * e_frac[ji, jv, jsl, jst] / (soiltile[ji, jst] * gm.vegtot[ji])
    else:
        for jv in range(0, gm.nvm, 1):
            jst = gm.pref_soil_veg[jv] - 1
            for jsl in range(0, gm.nslm, 1):
                for ji in range(0, gm.kjpindex, 1):
                    if humrel[ji, jv] > gm.min_sechiba and soiltile[ji, jst] * gm.vegtot[ji] > gm.min_sechiba:
                        gm.rootsink[ji, jsl, jst] = gm.rootsink[ji, jsl, jst] + transpir[ji, jv] * (us[ji, jv, jst, jsl] / humrel[ji, jv]) / (soiltile[ji, jst] * gm.vegtot[ji])
    if gm.check_cwrr:
        error = False
        tmp_check1[:] = gm.zero
        for jst in range(0, gm.nstm, 1):
            for ji in range(0, gm.kjpindex, 1):
                tmp_check1[ji] = tmp_check1[ji] + gm.precisol_ns[ji, jst] * soiltile[ji, jst] * gm.vegtot[ji]
        tmp_check2[:] = gm.zero
        for jv in range(0, gm.nvm, 1):
            for ji in range(0, gm.kjpindex, 1):
                tmp_check2[ji] = tmp_check2[ji] + gm.precisol[ji, jv]
        for ji in range(0, gm.kjpindex, 1):
            if np.abs(tmp_check1[ji] - tmp_check2[ji]) > gm.allowed_err:
                print(f'PRECISOL SPLIT FALSE:ji={ji},{tmp_check1[ji]},{tmp_check2[ji]}')
                print(f'err{np.abs(tmp_check1[ji] - tmp_check2[ji])}')
                print(f'vegtot{gm.vegtot[ji]}')
                for jv in range(0, gm.nvm, 1):
                    print(f'jv,veget_max, precisol, vegetmax_soil{jv},{veget_max[ji, jv]},{gm.precisol[ji, jv]},{gm.vegetmax_soil[ji, jv, :]}')
                for jst in range(0, gm.nstm, 1):
                    print(f'jst,precisol_ns{jst},{gm.precisol_ns[ji, jst]}')
                    print(f'soiltile{soiltile[ji, jst]}')
                error = True
                ipslerr_p(2, 'hydrol_split_soil', 'We will STOP in the end of this subroutine.', 'check_CWRR', 'PRECISOL SPLIT FALSE')
        tmp_check1[:] = gm.zero
        for jst in range(0, gm.nstm, 1):
            for ji in range(0, gm.kjpindex, 1):
                tmp_check1[ji] = tmp_check1[ji] + gm.ae_ns[ji, jst] * soiltile[ji, jst] * gm.vegtot[ji]
        for ji in range(0, gm.kjpindex, 1):
            if np.abs(tmp_check1[ji] - vevapnu[ji]) > gm.allowed_err:
                print(f'VEVAPNU SPLIT FALSE:ji, Sum(ae_ns), vevapnu ={ji},{tmp_check1[ji]},{vevapnu[ji]}')
                print(f'err{np.abs(tmp_check1[ji] - vevapnu[ji])}')
                print(f'ae_ns{gm.ae_ns[ji, :]}')
                print(f'vegtot{gm.vegtot[ji]}')
                print(f'evap_bare_lim, evap_bare_lim_ns{evap_bare_lim[ji]},{evap_bare_lim_ns[ji, :]}')
                for jst in range(0, gm.nstm, 1):
                    print(f'jst,ae_ns{jst},{gm.ae_ns[ji, jst]}')
                    print(f'soiltile{soiltile[ji, jst]}')
                error = True
                ipslerr_p(2, 'hydrol_split_soil', 'We will STOP in the end of this subroutine.', 'check_CWRR', 'VEVAPNU SPLIT FALSE')
        tmp_check1[:] = gm.zero
        for jst in range(0, gm.nstm, 1):
            for ji in range(0, gm.kjpindex, 1):
                tmp_check1[ji] = tmp_check1[ji] + gm.tr_ns[ji, jst] * soiltile[ji, jst] * gm.vegtot[ji]
        tmp_check2[:] = gm.zero
        for jv in range(0, gm.nvm, 1):
            for ji in range(0, gm.kjpindex, 1):
                tmp_check2[ji] = tmp_check2[ji] + transpir[ji, jv]
        for ji in range(0, gm.kjpindex, 1):
            if np.abs(tmp_check1[ji] - tmp_check2[ji]) > gm.allowed_err:
                print(f'TRANSPIR SPLIT FALSE:ji={ji},{tmp_check1[ji]},{tmp_check2[ji]}')
                print(f'err{np.abs(tmp_check1[ji] - tmp_check2[ji])}')
                print(f'vegtot{gm.vegtot[ji]}')
                for jv in range(0, gm.nvm, 1):
                    print(f'jv,veget_max, transpir{jv},{veget_max[ji, jv]},{transpir[ji, jv]}')
                    for jst in range(0, gm.nstm, 1):
                        print(f'vegetmax_soil:ji,jv,jst{ji},{jv},{jst},{gm.vegetmax_soil[ji, jv, jst]}')
                for jst in range(0, gm.nstm, 1):
                    print(f'jst,tr_ns{jst},{gm.tr_ns[ji, jst]}')
                    print(f'soiltile{soiltile[ji, jst]}')
                error = True
                ipslerr_p(2, 'hydrol_split_soil', 'We will STOP in the end of this subroutine.', 'check_CWRR', 'TRANSPIR SPLIT FALSE')
        tmp_check3[:, :] = gm.zero
        for jst in range(0, gm.nstm, 1):
            for jsl in range(0, gm.nslm, 1):
                for ji in range(0, gm.kjpindex, 1):
                    tmp_check3[ji, jst] = tmp_check3[ji, jst] + gm.rootsink[ji, jsl, jst]
        for jst in range(0, gm.nstm, 1):
            for ji in range(0, gm.kjpindex, 1):
                if np.abs(tmp_check3[ji, jst] - gm.tr_ns[ji, jst]) > gm.allowed_err:
                    print(f'ROOTSINK SPLIT FALSE:ji,jst={ji},{jst},{tmp_check3[ji, jst]},{gm.tr_ns[ji, jst]}')
                    print(f'err{np.abs(tmp_check3[ji, jst] - gm.tr_ns[ji, jst])}')
                    print(f'HUMREL(jv=1:13){humrel[ji, :]}')
                    print(f'TRANSPIR{transpir[ji, :]}')
                    for jv in range(0, gm.nvm, 1):
                        print(f'jv={jv},us={us[ji, jv, jst, :]}')
                    error = True
                    ipslerr_p(2, 'hydrol_split_soil', 'We will STOP in the end of this subroutine.', 'check_CWRR', 'ROOTSINK SPLIT FALSE')
        if error:
            print('One or more errors have been detected in hydrol_split_soil. Model stops.')
            ipslerr_p(3, 'hydrol_split_soil', 'We will STOP now.', 'One or several fatal errors were found previously.', '')

def test_hydrol_split_soil(gm):
    print('--- inside the test function for hydrol_split_soil ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_split_soil/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['ae_ns', 'precisol_ns', 'rootsink', 'tr_ns']
    for (variable, value) in zip(modif_var, [gm.ae_ns, gm.precisol_ns, gm.rootsink, gm.tr_ns]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gm = global_module_split_soil()
    gm.declaration_initialization()
    e_frac = np.zeros((gm.kjpindex, gm.nvm, gm.nslm, gm.nstm), dtype=np.float64)
    evap_bare_lim = np.zeros((gm.kjpindex,), dtype=np.float64)
    evap_bare_lim_ns = np.zeros((gm.kjpindex, gm.nstm), dtype=np.float64)
    F_absorption = np.zeros((gm.kjpindex, gm.nvm), dtype=np.float64)
    humrel = np.zeros((gm.kjpindex, gm.nvm), dtype=np.float64)
    soiltile = np.zeros((gm.kjpindex, gm.nstm), dtype=np.float64)
    tot_bare_soil = np.zeros((gm.kjpindex,), dtype=np.float64)
    transpir = np.zeros((gm.kjpindex, gm.nvm), dtype=np.float64)
    us = np.zeros((gm.kjpindex, gm.nvm, gm.nstm, gm.nslm), dtype=np.float64)
    veget_max = np.zeros((gm.kjpindex, gm.nvm), dtype=np.float64)
    vevapnu = np.zeros((gm.kjpindex,), dtype=np.float64)
    read_dummy(e_frac, evap_bare_lim, evap_bare_lim_ns, F_absorption, humrel, soiltile, tot_bare_soil, transpir, us, veget_max, vevapnu)
    hydrol_split_soil(gm, veget_max, soiltile, vevapnu, transpir, humrel, evap_bare_lim, evap_bare_lim_ns, tot_bare_soil, us, e_frac, F_absorption)
    test_hydrol_split_soil(gm)
if __name__ == '__main__':
    main()
```

## hydrol_soil_tridiag

```python
import os
import time
import functools
import numpy as np
from scipy.io import FortranFile
# from global_module import Global_module

def read_dummy(ins):
    print(f'--- inside the read dummy routine for hydrol_soil_tridiag ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_tridiag/dummy.bin'
    ffile = FortranFile(path, 'r')
    ins = ffile.read_ints(np.int32)[0]
    return ins

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        return result
    return wrapper

@timer
def hydrol_soil_tridiag(gm, ins):
    bet = np.zeros((gm.kjpindex,), dtype=np.float64)
    gam = np.zeros((gm.kjpindex, gm.nslm), dtype=np.float64)
    for ji in range(0, gm.kjpindex, 1):
        if gm.resolv[ji]:
            bet[ji] = gm.tmat[ji, 0, 1]
            gm.mcl[ji, 0, ins - 1] = gm.rhs[ji, 0] / bet[ji]
    for jsl in range(1, gm.nslm, 1):
        for ji in range(0, gm.kjpindex, 1):
            if gm.resolv[ji]:
                gam[ji, jsl] = gm.tmat[ji, jsl - 1, 2] / bet[ji]
                bet[ji] = gm.tmat[ji, jsl, 1] - gm.tmat[ji, jsl, 0] * gam[ji, jsl]
                gm.mcl[ji, jsl, ins - 1] = (gm.rhs[ji, jsl] - gm.tmat[ji, jsl, 0] * gm.mcl[ji, jsl - 1, ins - 1]) / bet[ji]
    for ji in range(0, gm.kjpindex, 1):
        if gm.resolv[ji]:
            for jsl in range(gm.nslm - 2, -1, -1):
                gm.mcl[ji, jsl, ins - 1] = gm.mcl[ji, jsl, ins - 1] - gam[ji, jsl + 1] * gm.mcl[ji, jsl + 1, ins - 1]

def test_hydrol_soil_tridiag(gm):
    print('--- inside the test function for hydrol_soil_tridiag ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_tridiag/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['mcl']
    for (variable, value) in zip(modif_var, [gm.mcl]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gm = global_module_soil_tridiag()
    gm.declaration_initialization()
    ins = np.int32(0)
    ins = read_dummy(ins)
    hydrol_soil_tridiag(gm, ins)
    test_hydrol_soil_tridiag(gm)
if __name__ == '__main__':
    main()
```

## hydrol_root_profile

```python
import os
import time
import functools
import numpy as np
import logging
from scipy.io import FortranFile
# from module_global import Global_module_hydrol_root_profile

def read_dummy(altmax, sm, smw):
    print(f'--- inside the read dummy routine for hydrol_root_profile ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_root_profile/dummy.bin'
    ffile = FortranFile(path, 'r')
    for var in [altmax, sm, smw]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                var[:] = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif var.dtype == np.int32:
                var[:] = ffile.read_ints(np.int32).reshape(arr_shape, order='F')

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        path = '/data/ssivanes/Fgpt/benchmark/hydrol_root_profile/time.txt'
        with open(path, 'a') as f:
            f.write(f"[TIMER] Python: '{func.__name__}' executed in {duration:.6f} seconds")
        return result
    return wrapper

@timer
def hydrol_root_profile(gmhrp, altmax, sm, smw, root_profile, root_depth):
    count = np.zeros((gmhrp.kjpindex,), dtype=np.float64)
    root_profile_tmp = np.zeros((gmhrp.nslm,), dtype=np.float64)
    root_depth[:, :, :] = gmhrp.zero
    for ji in range(0, gmhrp.kjpindex, 1):
        for jv in range(0, gmhrp.nvm, 1):
            root_depth_tmp = np.minimum(np.minimum(altmax[ji, jv], gmhrp.maxaltmax), np.minimum(gmhrp.zdr[gmhrp.nslm], gmhrp.max_root_depth[jv]))
            root_depth[ji, jv, gmhrp.inode - 1] = np.argmin(np.abs(root_depth_tmp - gmhrp.znh[:]), axis=0) + 1
            root_depth[ji, jv, gmhrp.iinterface - 1] = np.argmin(np.abs(root_depth_tmp - gmhrp.zdr[:]), axis=0)
    root_depth[:, :, gmhrp.inode - 1] = np.where(root_depth[:, :, gmhrp.inode - 1] <= 2, 2, root_depth[:, :, gmhrp.inode - 1])
    root_depth[:, :, gmhrp.iinterface - 1] = np.where(root_depth[:, :, gmhrp.iinterface - 1] <= 2, 2, root_depth[:, :, gmhrp.iinterface - 1])
    root_profile[:, :, :, gmhrp.istruc - 1] = gmhrp.zero
    for jv in range(0, gmhrp.nvm, 1):
        for ji in range(0, gmhrp.kjpindex, 1):
            z_top = gmhrp.zdr[1]
            z_bottom = gmhrp.zdr[int(root_depth[ji, jv, gmhrp.iinterface - 1])]
            rpc = gmhrp.un / (np.exp(-(z_top * gmhrp.humcste[jv])) - np.exp(-(z_bottom * gmhrp.humcste[jv])))
            for jsl in range(1, int(root_depth[ji, jv, gmhrp.iinterface - 1]), 1):
                root_profile[ji, jv, jsl, gmhrp.istruc - 1] = rpc * (np.exp(-(gmhrp.zdr[jsl] * gmhrp.humcste[jv])) - np.exp(-(gmhrp.zdr[jsl + 1] * gmhrp.humcste[jv])))
            root_profile[ji, jv, 0, gmhrp.istruc - 1] = gmhrp.zero
            if gmhrp.err_act > 1:
                if np.abs(np.sum(root_profile[ji, jv, :, gmhrp.istruc - 1]) - gmhrp.un) > 100 * np.finfo(np.float64).eps:
                    logging.info(f'pixel, PFT, sum of structural root_profile,{ji},{jv},{np.sum(root_profile[ji, jv, :, gmhrp.istruc - 1])}')
                    logging.info(f'hydrol_root_profile, structural root_profile,{root_profile[ji, jv, :, gmhrp.istruc - 1]}')
                    logging.error('Exception: hydrol.f90 structural root profile does not add up to 1 Check its calculation ')
    root_profile[:, :, :, gmhrp.ifunc - 1] = gmhrp.zero
    for jv in range(1, gmhrp.nvm, 1):
        for ji in range(0, gmhrp.kjpindex, 1):
            root_profile_tmp[:] = gmhrp.zero
            for jsl in range(1, int(root_depth[ji, jv, gmhrp.iinterface - 1]), 1):
                root_profile_tmp[jsl] = np.maximum(gmhrp.zero, sm[ji, jsl] - smw[ji, jsl])
            if np.sum(root_profile_tmp[:]) > gmhrp.min_sechiba:
                root_profile[ji, jv, :, gmhrp.ifunc - 1] = root_profile_tmp[:] / np.sum(root_profile_tmp[:])
            else:
                root_profile[ji, jv, :, gmhrp.ifunc - 1] = gmhrp.zero
            root_profile[ji, jv, 0, gmhrp.ifunc - 1] = gmhrp.zero
            if gmhrp.err_act > 1:
                if np.abs(np.sum(root_profile[ji, jv, :, gmhrp.ifunc - 1]) - gmhrp.un) > 100 * np.finfo(np.float64).eps and np.sum(root_profile_tmp[:]) > gmhrp.min_sechiba:
                    logging.info(f'pixel, PFT, sum of functional root_profile,{ji},{jv},{np.sum(root_profile[ji, jv, :, gmhrp.ifunc - 1])}')
                    logging.info(f'hydrol_root_profile, functional root_profile,{root_profile[ji, jv, :, gmhrp.ifunc - 1]}')
                    logging.error('Exception: hydrol.f90 functional root profile does not add up to 1 Check its calculation ')

def test_hydrol_root_profile(root_depth, root_profile):
    print('--- inside the test function for hydrol_root_profile ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_root_profile/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['root_depth', 'root_profile']
    for (variable, value) in zip(modif_var, [root_depth, root_profile]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gmhrp = global_module_root_profile()
    gmhrp.declaration_initialization()
    altmax = np.zeros((gmhrp.kjpindex, gmhrp.nvm), dtype=np.float64)
    root_depth = np.zeros((gmhrp.kjpindex, gmhrp.nvm, gmhrp.ndepths), dtype=np.float64)
    root_profile = np.zeros((gmhrp.kjpindex, gmhrp.nvm, gmhrp.nslm, gmhrp.nroot_prof), dtype=np.float64)
    sm = np.zeros((gmhrp.kjpindex, gmhrp.nslm), dtype=np.float64)
    smw = np.zeros((gmhrp.kjpindex, gmhrp.nslm), dtype=np.float64)
    read_dummy(altmax, sm, smw)
    hydrol_root_profile(gmhrp, altmax, sm, smw, root_profile, root_depth)
    test_hydrol_root_profile(root_depth, root_profile)
if __name__ == '__main__':
    main()
```

```python
import os
import time
import functools
import numpy as np
from scipy.io import FortranFile
# from global_module import Global_module

def read_dummy(ins, mcr, njsc):
    print(f'--- inside the read dummy routine for hydrol_soil_smooth_under_mcr ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_smooth_under_mcr/dummy.bin'
    ffile = FortranFile(path, 'r')
    ins = ffile.read_ints(np.int32)[0]
    for var in [mcr, njsc]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                var[:] = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif var.dtype == np.int32:
                var[:] = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
    return ins

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        return result
    return wrapper

@timer
def hydrol_soil_smooth_under_mcr(gm, mcr, ins, njsc, is_under_mcr, check):
    excessji = np.zeros((gm.kjpindex,), dtype=np.float64)
    tmci = np.zeros((gm.kjpindex,), dtype=np.float64)
    tmcf = np.zeros((gm.kjpindex,), dtype=np.float64)
    if gm.check_cwrr:
        tmci[:] = gm.dz[1] * (gm.trois * gm.mc[:, 0, ins - 1] + gm.mc[:, 1, ins - 1]) / gm.huit
        for jsl in range(1, gm.nslm - 1, 1):
            tmci[:] = tmci[:] + gm.dz[jsl] * (gm.trois * gm.mc[:, jsl, ins - 1] + gm.mc[:, jsl - 1, ins - 1]) / gm.huit + gm.dz[jsl + 1] * (gm.trois * gm.mc[:, jsl, ins - 1] + gm.mc[:, jsl + 1, ins - 1]) / gm.huit
        tmci[:] = tmci[:] + gm.dz[gm.nslm - 1] * (gm.trois * gm.mc[:, gm.nslm - 1, ins - 1] + gm.mc[:, gm.nslm - 2, ins - 1]) / gm.huit
    for jsl in range(0, gm.nslm - 2, 1):
        for ji in range(0, gm.kjpindex, 1):
            excess = np.maximum(mcr[ji] - gm.mc[ji, jsl, ins - 1], gm.zero)
            gm.mc[ji, jsl, ins - 1] = gm.mc[ji, jsl, ins - 1] + excess
            gm.mc[ji, jsl + 1, ins - 1] = gm.mc[ji, jsl + 1, ins - 1] - excess * (gm.dz[jsl] + gm.dz[jsl + 1]) / (gm.dz[jsl + 1] + gm.dz[jsl + 2])
    jsl = gm.nslm - 2
    for ji in range(0, gm.kjpindex, 1):
        excess = np.maximum(mcr[ji] - gm.mc[ji, jsl, ins - 1], gm.zero)
        gm.mc[ji, jsl, ins - 1] = gm.mc[ji, jsl, ins - 1] + excess
        gm.mc[ji, jsl + 1, ins - 1] = gm.mc[ji, jsl + 1, ins - 1] - excess * (gm.dz[jsl] + gm.dz[jsl + 1]) / gm.dz[jsl + 1]
    jsl = gm.nslm - 1
    for ji in range(0, gm.kjpindex, 1):
        excess = np.maximum(mcr[ji] - gm.mc[ji, jsl, ins - 1], gm.zero)
        gm.mc[ji, jsl, ins - 1] = gm.mc[ji, jsl, ins - 1] + excess
        gm.mc[ji, jsl - 1, ins - 1] = gm.mc[ji, jsl - 1, ins - 1] - excess * gm.dz[jsl] / (gm.dz[jsl - 1] + gm.dz[jsl])
    for jsl in range(gm.nslm - 2, -2, -1):
        for ji in range(0, gm.kjpindex, 1):
            excess = np.maximum(mcr[ji] - gm.mc[ji, jsl, ins - 1], gm.zero)
            gm.mc[ji, jsl, ins - 1] = gm.mc[ji, jsl, ins - 1] + excess
            gm.mc[ji, jsl - 1, ins - 1] = gm.mc[ji, jsl - 1, ins - 1] - excess * (gm.dz[jsl] + gm.dz[jsl + 1]) / (gm.dz[jsl - 1] + gm.dz[jsl])
    for ji in range(0, gm.kjpindex, 1):
        excessji[ji] = gm.mask_soiltile[ji, ins - 1] * np.maximum(mcr[ji] - gm.mc[ji, 0, ins - 1], gm.zero)
    for ji in range(0, gm.kjpindex, 1):
        gm.mc[ji, 0, ins - 1] = gm.mc[ji, 0, ins - 1] + excessji[ji]
        is_under_mcr[ji, ins - 1] = excessji[ji] > gm.min_sechiba
    for jsl in range(0, gm.nslm, 1):
        for ji in range(0, gm.kjpindex, 1):
            gm.mc[ji, jsl, ins - 1] = gm.mc[ji, jsl, ins - 1] - excessji[ji] * gm.dz[1] / (gm.deux * gm.zmaxh * gm.mille)
    for jsl in range(0, gm.nslm, 1):
        for ji in range(0, gm.kjpindex, 1):
            gm.mc[ji, jsl, ins - 1] = gm.mask_soiltile[ji, ins - 1] * gm.mc[ji, jsl, ins - 1]
    if gm.check_cwrr:
        tmcf[:] = gm.dz[1] * (gm.trois * gm.mc[:, 0, ins - 1] + gm.mc[:, 1, ins - 1]) / gm.huit
        for jsl in range(1, gm.nslm - 1, 1):
            tmcf[:] = tmcf[:] + gm.dz[jsl] * (gm.trois * gm.mc[:, jsl, ins - 1] + gm.mc[:, jsl - 1, ins - 1]) / gm.huit + gm.dz[jsl + 1] * (gm.trois * gm.mc[:, jsl, ins - 1] + gm.mc[:, jsl + 1, ins - 1]) / gm.huit
        tmcf[:] = tmcf[:] + gm.dz[gm.nslm - 1] * (gm.trois * gm.mc[:, gm.nslm - 1, ins - 1] + gm.mc[:, gm.nslm - 2, ins - 1]) / gm.huit
        check[:, ins - 1] = tmcf[:] - tmci[:]

def test_hydrol_soil_smooth_under_mcr(gm, is_under_mcr, check):
    print('--- inside the test function for hydrol_soil_smooth_under_mcr ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_smooth_under_mcr/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['check', 'is_under_mcr', 'mc']
    for (variable, value) in zip(modif_var, [check, is_under_mcr, gm.mc]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gm = global_module_soil_smooth_under_mcr()
    gm.declaration_initialization()
    check = np.zeros((gm.kjpindex, gm.nstm), dtype=np.float64)
    ins = np.int32(0)
    is_under_mcr = np.zeros((gm.kjpindex, gm.nstm), dtype=np.int32)
    mcr = np.zeros((gm.kjpindex,), dtype=np.float64)
    njsc = np.zeros((gm.kjpindex,), dtype=np.int32)
    ins = read_dummy(ins, mcr, njsc)
    hydrol_soil_smooth_under_mcr(gm, mcr, ins, njsc, is_under_mcr, check)
    test_hydrol_soil_smooth_under_mcr(gm, is_under_mcr, check)
if __name__ == '__main__':
    main()
```
