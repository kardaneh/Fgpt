## hydrol_soil_froz

```python
import numpy as np
import os
from scipy.io import FortranFile

class global_module_soil_froz:

    def __init__(self):
        i_std = np.int32
        r_std = np.float64
        self.nsnow = i_std(3)
        self.nslm = i_std(11)
        self.nvm = i_std(15)
        self.nstm = i_std(3)
        self.kjpindex = i_std(4717)
        self.ier = i_std(0)
        self.ic0 = i_std(0)
        self.ic = i_std(0)
        self.icr = r_std(0.0)
        self.start_time = r_std(0.0)
        self.stop_time = r_std(0.0)
        self.zero = np.float64(0.0)
        self.min_sechiba = np.float64(1e-08)
        self.lhf = np.float64(0.3336 * 1000000.0)
        self.ZeroCelsius = np.float64(273.15)
        self.fr_center = np.float64(0)
        self.fr_dT = np.float64(0)
        self.froz_frac_corr = np.float64(0)
        self.max_froz_hydro = np.float64(0)
        self.ok_thermodynamical_freezing = np.bool(False)
        self.smtot_corr = np.float64(0)
        self.dh = np.zeros((self.nslm,), dtype=np.float64)
        self.mc = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)
        self.profil_froz_hydro_ns = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)

    def declaration_initialization(self):
        print(f'--- add the declaration and initialization in module global ---')
        path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_froz/global.bin'
        ffile = FortranFile(path, 'r')
        self.fr_center = ffile.read_reals(np.float64)[0]
        self.fr_dT = ffile.read_reals(np.float64)[0]
        self.froz_frac_corr = ffile.read_reals(np.float64)[0]
        self.max_froz_hydro = ffile.read_reals(np.float64)[0]
        self.ok_thermodynamical_freezing = np.bool(ffile.read_ints(np.int32)[0])
        self.smtot_corr = ffile.read_reals(np.float64)[0]
        for attr_name in ['dh', 'mc', 'profil_froz_hydro_ns']:
            attribute = getattr(self, attr_name)
            if attribute.dtype == np.float64 and isinstance(attribute, np.ndarray):
                arr_shape = attribute.shape
                setattr(self, attr_name, ffile.read_reals(np.float64).reshape(arr_shape, order='F'))
            elif attribute.dtype == np.int32 and isinstance(attribute, np.ndarray):
                arr_shape = attribute.shape
                setattr(self, attr_name, ffile.read_ints(np.int32).reshape(arr_shape, order='F'))
```

## hydrol_diag_soil

```python
import os
import time
import functools
import numpy as np
from scipy.io import FortranFile
# from global_module import Global_module

def read_dummy(avan, evapot, frac_snow_nobio, irrigation, ks, mcfc, mcr, mcs, mcw, njsc, nvan, precip_rain, reinfiltration, returnflow, soiltile, tot_melt, totfrac_nobio, us, veget_max, vevapnu):
    print(f'--- inside the read dummy routine for hydrol_diag_soil ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_diag_soil/dummy.bin'
    ffile = FortranFile(path, 'r')
    for var in [avan, evapot, frac_snow_nobio, irrigation, ks, mcfc, mcr, mcs, mcw, njsc, nvan, precip_rain, reinfiltration, returnflow, soiltile, tot_melt, totfrac_nobio, us, veget_max, vevapnu]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                var[:] = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif var.dtype == np.int32:
                var[:] = ffile.read_ints(np.int32).reshape(arr_shape, order='F')

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        return result
    return wrapper

@timer
def hydrol_diag_soil(gm, ks, nvan, avan, mcr, mcs, mcfc, mcw, veget_max, soiltile, njsc, runoff, drainage, evapot, vevapnu, returnflow, reinfiltration, irrigation, shumdiag, shumdiag_perma, k_litt, litterhumdiag, humrel, vegstress, drysoil_frac, tot_melt, us, precip_rain, totfrac_nobio, frac_snow_nobio):
    mask_vegtot = np.zeros((gm.kjpindex,), dtype=np.float64)
    for jst in range(0, gm.nstm, 1):
        for ji in range(0, gm.kjpindex, 1):
            gm.ae_ns[ji, jst] = gm.ae_ns[ji, jst] * gm.mask_soiltile[ji, jst]
            gm.dr_ns[ji, jst] = gm.dr_ns[ji, jst] * gm.mask_soiltile[ji, jst]
            gm.ru_ns[ji, jst] = gm.ru_ns[ji, jst] * gm.mask_soiltile[ji, jst]
            gm.tmc[ji, jst] = gm.tmc[ji, jst] * gm.mask_soiltile[ji, jst]
            for jv in range(0, gm.nvm, 1):
                gm.humrelv[ji, jv, jst] = gm.humrelv[ji, jv, jst] * gm.mask_soiltile[ji, jst]
                for jsl in range(0, gm.nslm, 1):
                    us[ji, jv, jst, jsl] = us[ji, jv, jst, jsl] * gm.mask_soiltile[ji, jst]
            for jsl in range(0, gm.nslm, 1):
                gm.mc[ji, jsl, jst] = gm.mc[ji, jsl, jst] * gm.mask_soiltile[ji, jst]
    runoff[:] = gm.zero
    drainage[:] = gm.zero
    gm.humtot[:] = gm.zero
    shumdiag[:, :] = gm.zero
    shumdiag_perma[:, :] = gm.zero
    k_litt[:] = gm.zero
    litterhumdiag[:] = gm.zero
    gm.tmc_litt_dry_mea[:] = gm.zero
    gm.tmc_litt_wet_mea[:] = gm.zero
    gm.tmc_litt_mea[:] = gm.zero
    humrel[:, :] = gm.zero
    vegstress[:, :] = gm.zero
    if gm.ok_freeze_cwrr:
        gm.profil_froz_hydro[:, :] = gm.zero
    for ji in range(0, gm.kjpindex, 1):
        mask_vegtot[ji] = 0
        if gm.vegtot[ji] > gm.min_sechiba:
            mask_vegtot[ji] = 1
    for ji in range(0, gm.kjpindex, 1):
        gm.ae_ns[ji, :] = mask_vegtot[ji] * gm.ae_ns[ji, :] * gm.frac_bare_ns[ji, :]
    for jst in range(0, gm.nstm, 1):
        for ji in range(0, gm.kjpindex, 1):
            drainage[ji] = mask_vegtot[ji] * (drainage[ji] + gm.vegtot[ji] * soiltile[ji, jst] * gm.dr_ns[ji, jst])
            runoff[ji] = mask_vegtot[ji] * (runoff[ji] + gm.vegtot[ji] * soiltile[ji, jst] * gm.ru_ns[ji, jst]) + (1 - mask_vegtot[ji]) * (tot_melt[ji] + irrigation[ji] + returnflow[ji] + reinfiltration[ji])
            gm.humtot[ji] = mask_vegtot[ji] * (gm.humtot[ji] + gm.vegtot[ji] * soiltile[ji, jst] * gm.tmc[ji, jst])
            if gm.ok_freeze_cwrr:
                gm.profil_froz_hydro[ji, :] = mask_vegtot[ji] * (gm.profil_froz_hydro[ji, :] + gm.vegtot[ji] * soiltile[ji, jst] * gm.profil_froz_hydro_ns[ji, :, jst])
    for ji in range(0, gm.kjpindex, 1):
        if gm.vegtot[ji] != 0.0:
            vevapnu[ji] = vevapnu[ji] + gm.subsinksoil[ji] * gm.vegtot[ji]
        else:
            vevapnu[ji] = vevapnu[ji] + gm.subsinksoil[ji]
        runoff[ji] = runoff[ji] + precip_rain[ji] * totfrac_nobio[ji] * (1 - frac_snow_nobio[ji, gm.iice - 1])
    for jst in range(0, gm.nstm, 1):
        for jv in range(0, gm.nvm, 1):
            for ji in range(0, gm.kjpindex, 1):
                if veget_max[ji, jv] > gm.min_sechiba:
                    vegstress[ji, jv] = vegstress[ji, jv] + gm.vegstressv[ji, jv, jst]
                    vegstress[ji, jv] = np.maximum(vegstress[ji, jv], gm.zero)
    for jst in range(0, gm.nstm, 1):
        for jv in range(0, gm.nvm, 1):
            for ji in range(0, gm.kjpindex, 1):
                humrel[ji, jv] = humrel[ji, jv] + gm.humrelv[ji, jv, jst]
                humrel[ji, jv] = np.maximum(humrel[ji, jv], gm.zero)
    for jst in range(0, gm.nstm, 1):
        for ji in range(0, gm.kjpindex, 1):
            if gm.tmc_litter[ji, jst] < gm.tmc_litter_res[ji, jst]:
                i = gm.imin
            else:
                tmc_litter_ratio = (gm.tmc_litter[ji, jst] - gm.tmc_litter_res[ji, jst]) / (gm.tmc_litter_sat[ji, jst] - gm.tmc_litter_res[ji, jst])
                i = np.maximum(np.minimum(int((gm.imax - gm.imin) * tmc_litter_ratio) + gm.imin, gm.imax - 1), gm.imin)
            k_tmp = np.maximum(gm.k_lin[i - 1, 0, ji] * ks[ji], gm.zero)
            k_litt[ji] = k_litt[ji] + gm.vegtot[ji] * soiltile[ji, jst] * np.sqrt(k_tmp)
        for ji in range(0, gm.kjpindex, 1):
            litterhumdiag[ji] = litterhumdiag[ji] + gm.soil_wet_litter[ji, jst] * soiltile[ji, jst]
            gm.tmc_litt_wet_mea[ji] = gm.tmc_litt_wet_mea[ji] + gm.tmc_litter_awet[ji, jst] * soiltile[ji, jst]
            gm.tmc_litt_dry_mea[ji] = gm.tmc_litt_dry_mea[ji] + gm.tmc_litter_adry[ji, jst] * soiltile[ji, jst]
            gm.tmc_litt_mea[ji] = gm.tmc_litt_mea[ji] + gm.tmc_litter[ji, jst] * soiltile[ji, jst]
    for ji in range(0, gm.kjpindex, 1):
        if gm.tmc_litt_wet_mea[ji] - gm.tmc_litt_dry_mea[ji] > gm.zero:
            drysoil_frac[ji] = gm.un + np.maximum(np.minimum((gm.tmc_litt_dry_mea[ji] - gm.tmc_litt_mea[ji]) / (gm.tmc_litt_wet_mea[ji] - gm.tmc_litt_dry_mea[ji]), gm.zero), -gm.un)
        else:
            drysoil_frac[ji] = gm.zero
    gm.soilmoist[:, :] = gm.zero
    for jst in range(0, gm.nstm, 1):
        for ji in range(0, gm.kjpindex, 1):
            gm.soilmoist[ji, 0] = gm.soilmoist[ji, 0] + soiltile[ji, jst] * gm.dz[1] * (gm.trois * gm.mc[ji, 0, jst] + gm.mc[ji, 1, jst]) / gm.huit
            for jsl in range(1, gm.nslm - 1, 1):
                gm.soilmoist[ji, jsl] = gm.soilmoist[ji, jsl] + soiltile[ji, jst] * (gm.dz[jsl] * (gm.trois * gm.mc[ji, jsl, jst] + gm.mc[ji, jsl - 1, jst]) / gm.huit + gm.dz[jsl + 1] * (gm.trois * gm.mc[ji, jsl, jst] + gm.mc[ji, jsl + 1, jst]) / gm.huit)
            gm.soilmoist[ji, gm.nslm - 1] = gm.soilmoist[ji, gm.nslm - 1] + soiltile[ji, jst] * gm.dz[gm.nslm - 1] * (gm.trois * gm.mc[ji, gm.nslm - 1, jst] + gm.mc[ji, gm.nslm - 2, jst]) / gm.huit
    for ji in range(0, gm.kjpindex, 1):
        gm.soilmoist[ji, :] = gm.soilmoist[ji, :] * gm.vegtot[ji]
    gm.soilmoist_s[:, :, :] = gm.zero
    for jst in range(0, gm.nstm, 1):
        for ji in range(0, gm.kjpindex, 1):
            gm.soilmoist_s[ji, 0, gm.nstm - 1] = gm.soilmoist_s[ji, 0, gm.nstm - 1] + soiltile[ji, jst] * gm.dz[1] * (gm.trois * gm.mc[ji, 0, jst] + gm.mc[ji, 1, jst]) / gm.huit
            for jsl in range(1, gm.nslm - 1, 1):
                gm.soilmoist_s[ji, jsl, gm.nstm - 1] = gm.soilmoist_s[ji, jsl, gm.nstm - 1] + soiltile[ji, jst] * (gm.dz[jsl] * (gm.trois * gm.mc[ji, jsl, jst] + gm.mc[ji, jsl - 1, jst]) / gm.huit + gm.dz[jsl + 1] * (gm.trois * gm.mc[ji, jsl, jst] + gm.mc[ji, jsl + 1, jst]) / gm.huit)
            gm.soilmoist_s[ji, gm.nslm - 1, gm.nstm - 1] = gm.soilmoist_s[ji, gm.nslm - 1, gm.nstm - 1] + soiltile[ji, jst] * gm.dz[gm.nslm - 1] * (gm.trois * gm.mc[ji, gm.nslm - 1, jst] + gm.mc[ji, gm.nslm - 2, jst]) / gm.huit
    for ji in range(0, gm.kjpindex, 1):
        gm.soilmoist_s[ji, :, :] = gm.soilmoist_s[ji, :, :] * gm.vegtot[ji]
    gm.soilmoist_liquid[:, :] = gm.zero
    for jst in range(0, gm.nstm, 1):
        for ji in range(0, gm.kjpindex, 1):
            gm.soilmoist_liquid[ji, 0] = gm.soilmoist_liquid[ji, 0] + soiltile[ji, jst] * gm.dz[1] * (gm.trois * gm.mcl[ji, 0, jst] + gm.mcl[ji, 1, jst]) / gm.huit
            for jsl in range(1, gm.nslm - 1, 1):
                gm.soilmoist_liquid[ji, jsl] = gm.soilmoist_liquid[ji, jsl] + soiltile[ji, jst] * (gm.dz[jsl] * (gm.trois * gm.mcl[ji, jsl, jst] + gm.mcl[ji, jsl - 1, jst]) / gm.huit + gm.dz[jsl + 1] * (gm.trois * gm.mcl[ji, jsl, jst] + gm.mcl[ji, jsl + 1, jst]) / gm.huit)
            gm.soilmoist_liquid[ji, gm.nslm - 1] = gm.soilmoist_liquid[ji, gm.nslm - 1] + soiltile[ji, jst] * gm.dz[gm.nslm - 1] * (gm.trois * gm.mcl[ji, gm.nslm - 1, jst] + gm.mcl[ji, gm.nslm - 2, jst]) / gm.huit
    for ji in range(0, gm.kjpindex, 1):
        gm.soilmoist_liquid[ji, :] = gm.soilmoist_liquid[ji, :] * gm.vegtot_old[ji]
    for jst in range(0, gm.nstm, 1):
        for jsl in range(0, gm.nslm, 1):
            for ji in range(0, gm.kjpindex, 1):
                shumdiag[ji, jsl] = shumdiag[ji, jsl] + gm.soil_wet_ns[ji, jsl, jst] * soiltile[ji, jst] * ((mcs[ji] - mcw[ji]) / (mcfc[ji] - mcw[ji]))
                shumdiag[ji, jsl] = np.maximum(np.minimum(shumdiag[ji, jsl], gm.un), gm.zero)
    for jsl in range(0, gm.nslm, 1):
        for ji in range(0, gm.kjpindex, 1):
            shumdiag_perma[ji, jsl] = gm.soilmoist[ji, jsl] / (gm.dh[jsl] * mcs[ji])
            shumdiag_perma[ji, jsl] = np.maximum(np.minimum(shumdiag_perma[ji, jsl], gm.un), gm.zero)

def test_hydrol_diag_soil(gm, vevapnu, runoff, drysoil_frac, drainage, k_litt, vegstress, litterhumdiag, us, shumdiag_perma, shumdiag, humrel):
    print('--- inside the test function for hydrol_diag_soil ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_diag_soil/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['ae_ns', 'dr_ns', 'drainage', 'drysoil_frac', 'humrel', 'humrelv', 'humtot', 'k_litt', 'litterhumdiag', 'mc', 'profil_froz_hydro', 'ru_ns', 'runoff', 'shumdiag', 'shumdiag_perma', 'soilmoist', 'soilmoist_liquid', 'soilmoist_s', 'tmc', 'tmc_litt_dry_mea', 'tmc_litt_mea', 'tmc_litt_wet_mea', 'us', 'vegstress', 'vevapnu']
    for (variable, value) in zip(modif_var, [gm.ae_ns, gm.dr_ns, drainage, drysoil_frac, humrel, gm.humrelv, gm.humtot, k_litt, litterhumdiag, gm.mc, gm.profil_froz_hydro, gm.ru_ns, runoff, shumdiag, shumdiag_perma, gm.soilmoist, gm.soilmoist_liquid, gm.soilmoist_s, gm.tmc, gm.tmc_litt_dry_mea, gm.tmc_litt_mea, gm.tmc_litt_wet_mea, us, vegstress, vevapnu]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gm = Global_module()
    gm.declaration_initialization()
    avan = np.zeros((gm.kjpindex,), dtype=np.float64)
    drainage = np.zeros((gm.kjpindex,), dtype=np.float64)
    drysoil_frac = np.zeros((gm.kjpindex,), dtype=np.float64)
    evapot = np.zeros((gm.kjpindex,), dtype=np.float64)
    frac_snow_nobio = np.zeros((gm.kjpindex, gm.nnobio), dtype=np.float64)
    humrel = np.zeros((gm.kjpindex, gm.nvm), dtype=np.float64)
    irrigation = np.zeros((gm.kjpindex,), dtype=np.float64)
    k_litt = np.zeros((gm.kjpindex,), dtype=np.float64)
    ks = np.zeros((gm.kjpindex,), dtype=np.float64)
    litterhumdiag = np.zeros((gm.kjpindex,), dtype=np.float64)
    mcfc = np.zeros((gm.kjpindex,), dtype=np.float64)
    mcr = np.zeros((gm.kjpindex,), dtype=np.float64)
    mcs = np.zeros((gm.kjpindex,), dtype=np.float64)
    mcw = np.zeros((gm.kjpindex,), dtype=np.float64)
    njsc = np.zeros((gm.kjpindex,), dtype=np.int32)
    nvan = np.zeros((gm.kjpindex,), dtype=np.float64)
    precip_rain = np.zeros((gm.kjpindex,), dtype=np.float64)
    reinfiltration = np.zeros((gm.kjpindex,), dtype=np.float64)
    returnflow = np.zeros((gm.kjpindex,), dtype=np.float64)
    runoff = np.zeros((gm.kjpindex,), dtype=np.float64)
    shumdiag = np.zeros((gm.kjpindex, gm.nslm), dtype=np.float64)
    shumdiag_perma = np.zeros((gm.kjpindex, gm.nslm), dtype=np.float64)
    soiltile = np.zeros((gm.kjpindex, gm.nstm), dtype=np.float64)
    tot_melt = np.zeros((gm.kjpindex,), dtype=np.float64)
    totfrac_nobio = np.zeros((gm.kjpindex,), dtype=np.float64)
    us = np.zeros((gm.kjpindex, gm.nvm, gm.nstm, gm.nslm), dtype=np.float64)
    veget_max = np.zeros((gm.kjpindex, gm.nvm), dtype=np.float64)
    vegstress = np.zeros((gm.kjpindex, gm.nvm), dtype=np.float64)
    vevapnu = np.zeros((gm.kjpindex,), dtype=np.float64)
    read_dummy(avan, evapot, frac_snow_nobio, irrigation, ks, mcfc, mcr, mcs, mcw, njsc, nvan, precip_rain, reinfiltration, returnflow, soiltile, tot_melt, totfrac_nobio, us, veget_max, vevapnu)
    hydrol_diag_soil(gm, ks, nvan, avan, mcr, mcs, mcfc, mcw, veget_max, soiltile, njsc, runoff, drainage, evapot, vevapnu, returnflow, reinfiltration, irrigation, shumdiag, shumdiag_perma, k_litt, litterhumdiag, humrel, vegstress, drysoil_frac, tot_melt, us, precip_rain, totfrac_nobio, frac_snow_nobio)
    test_hydrol_diag_soil(gm, vevapnu, runoff, drysoil_frac, drainage, k_litt, vegstress, litterhumdiag, us, shumdiag_perma, shumdiag, humrel)
if __name__ == '__main__':
    main()
```

## hydrol_soil_froz

```python
import os
import time
import functools
import numpy as np
from scipy.io import FortranFile
# from global_module import Global_module

def read_dummy(gm, avan, ins, mcr, mcs, njsc, nvan, stempdiag):
    print(f'--- inside the read dummy routine for hydrol_soil_froz ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_froz/dummy.bin'
    ffile = FortranFile(path, 'r')
    avan[:] = ffile.read_reals(np.float64).reshape((gm.kjpindex,), order='F')
    ins = ffile.read_ints(np.int32)[0]
    for var in [mcr, mcs, njsc, nvan, stempdiag]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                var[:] = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif var.dtype == np.int32:
                var[:] = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
    return ins

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        return result
    return wrapper

@timer
def hydrol_soil_froz(gm, nvan, avan, mcr, mcs, ins, njsc, stempdiag):
    froz_frac_moy = np.zeros((gm.kjpindex,), dtype=np.float64)
    smtot_moy = np.zeros((gm.kjpindex,), dtype=np.float64)
    mc_ns = np.zeros((gm.kjpindex, gm.nslm), dtype=np.float64)
    for jsl in range(0, gm.nslm, 1):
        for ji in range(0, gm.kjpindex, 1):
            m = 1.0 - 1.0 / nvan[ji]
            if not gm.ok_thermodynamical_freezing or gm.mc[ji, jsl, ins - 1] < mcr[ji] + gm.min_sechiba:
                if stempdiag[ji, jsl] >= gm.fr_center + gm.fr_dT / 2.0:
                    x = 1.0
                elif stempdiag[ji, jsl] >= gm.fr_center - gm.fr_dT / 2.0 and stempdiag[ji, jsl] < gm.fr_center + gm.fr_dT / 2.0:
                    x = (stempdiag[ji, jsl] - (gm.fr_center - gm.fr_dT / 2.0)) / gm.fr_dT
                else:
                    x = 0.0
            elif gm.ok_thermodynamical_freezing:
                if stempdiag[ji, jsl] >= gm.fr_center + gm.fr_dT / 2.0:
                    x = 1.0
                elif stempdiag[ji, jsl] >= gm.fr_center - gm.fr_dT / 2.0 and stempdiag[ji, jsl] < gm.fr_center + gm.fr_dT / 2.0:
                    x = np.minimum((mcs[ji] - mcr[ji]) * ((2.2 * 1000.0 * avan[ji] * (gm.fr_center + gm.fr_dT / 2.0 - stempdiag[ji, jsl]) * gm.lhf / gm.ZeroCelsius / 10.0) ** nvan[ji] + 1.0) ** (-m) / (gm.mc[ji, jsl, ins - 1] - mcr[ji]), 1.0)
                else:
                    x = 0.0
            gm.profil_froz_hydro_ns[ji, jsl, ins - 1] = 1.0 - x
            mc_ns[ji, jsl] = gm.mc[ji, jsl, ins - 1] / mcs[ji]
    froz_frac_moy[:] = gm.zero
    denom = gm.zero
    for jsl in range(0, gm.nslm, 1):
        froz_frac_moy[:] = froz_frac_moy[:] + gm.dh[jsl] * gm.profil_froz_hydro_ns[:, jsl, ins - 1]
        denom = denom + gm.dh[jsl]
    froz_frac_moy[:] = froz_frac_moy[:] / denom
    smtot_moy[:] = gm.zero
    denom = gm.zero
    for jsl in range(0, gm.nslm - 1, 1):
        smtot_moy[:] = smtot_moy[:] + gm.dh[jsl] * mc_ns[:, jsl]
        denom = denom + gm.dh[jsl]
    smtot_moy[:] = smtot_moy[:] / denom
    for jsl in range(0, gm.nslm, 1):
        gm.profil_froz_hydro_ns[:, jsl, ins - 1] = np.minimum(gm.profil_froz_hydro_ns[:, jsl, ins - 1] * froz_frac_moy[:] ** gm.froz_frac_corr * smtot_moy[:] ** gm.smtot_corr, gm.max_froz_hydro)

def test_hydrol_soil_froz(gm):
    print('--- inside the test function for hydrol_soil_froz ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_froz/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['profil_froz_hydro_ns']
    for (variable, value) in zip(modif_var, [gm.profil_froz_hydro_ns]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gm = global_module_soil_froz()
    gm.declaration_initialization()
    avan = np.zeros((gm.kjpindex,), dtype=np.float64)
    ins = np.int32(0)
    mcr = np.zeros((gm.kjpindex,), dtype=np.float64)
    mcs = np.zeros((gm.kjpindex,), dtype=np.float64)
    njsc = np.zeros((gm.kjpindex,), dtype=np.int32)
    nvan = np.zeros((gm.kjpindex,), dtype=np.float64)
    stempdiag = np.zeros((gm.kjpindex, gm.nslm), dtype=np.float64)
    ins = read_dummy(gm, avan, ins, mcr, mcs, njsc, nvan, stempdiag)
    hydrol_soil_froz(gm, nvan, avan, mcr, mcs, ins, njsc, stempdiag)
    test_hydrol_soil_froz(gm)
if __name__ == '__main__':
    main()
```

## hydrol_soil_coef

```python
import os
import time
import functools
import numpy as np
from scipy.io import FortranFile
# from global_module import Global_module

def read_dummy(ins, mcr, mcs, njsc):
    print(f'--- inside the read dummy routine for hydrol_soil_coef ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_coef/dummy.bin'
    ffile = FortranFile(path, 'r')
    ins = ffile.read_ints(np.int32)[0]
    for var in [mcr, mcs, njsc]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                var[:] = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif var.dtype == np.int32:
                var[:] = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
    return ins

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        return result
    return wrapper

@timer
def hydrol_soil_coef(gm, mcr, mcs, ins, njsc):
    if gm.ok_freeze_cwrr:
        for jsl in range(0, gm.nslm, 1):
            for ji in range(0, gm.kjpindex, 1):
                x = 1.0 - gm.profil_froz_hydro_ns[ji, jsl, ins - 1]
                mc_used = mcr[ji] + x * np.maximum(gm.mc[ji, jsl, ins - 1] - mcr[ji], gm.zero)
                i = np.maximum(gm.imin, np.minimum(gm.imax - 1, int(gm.imin + (gm.imax - gm.imin) * (mc_used - mcr[ji]) / (mcs[ji] - mcr[ji]))))
                gm.a[ji, jsl] = gm.a_lin[i - 1, jsl, ji] * gm.kfact_root[ji, jsl, ins - 1]
                gm.b[ji, jsl] = gm.b_lin[i - 1, jsl, ji] * gm.kfact_root[ji, jsl, ins - 1]
                gm.d[ji, jsl] = gm.d_lin[i - 1, jsl, ji] * gm.kfact_root[ji, jsl, ins - 1]
                gm.k[ji, jsl] = gm.kfact_root[ji, jsl, ins - 1] * np.maximum(gm.k_lin[gm.imin, jsl, ji], gm.a_lin[i - 1, jsl, ji] * mc_used + gm.b_lin[i - 1, jsl, ji])
    else:
        for jsl in range(0, gm.nslm, 1):
            for ji in range(0, gm.kjpindex, 1):
                mc_ratio = np.maximum(gm.mc[ji, jsl, ins - 1] - mcr[ji], gm.zero) / (mcs[ji] - mcr[ji])
                i = np.maximum(np.minimum(int((gm.imax - gm.imin) * mc_ratio) + gm.imin, gm.imax - 1), gm.imin)
                gm.a[ji, jsl] = gm.a_lin[i - 1, jsl, ji] * gm.kfact_root[ji, jsl, ins - 1]
                gm.b[ji, jsl] = gm.b_lin[i - 1, jsl, ji] * gm.kfact_root[ji, jsl, ins - 1]
                gm.d[ji, jsl] = gm.d_lin[i - 1, jsl, ji] * gm.kfact_root[ji, jsl, ins - 1]
                gm.k[ji, jsl] = gm.kfact_root[ji, jsl, ins - 1] * np.maximum(gm.k_lin[gm.imin, jsl, ji], gm.a_lin[i - 1, jsl, ji] * gm.mc[ji, jsl, ins - 1] + gm.b_lin[i - 1, jsl, ji])

def test_hydrol_soil_coef(gm):
    print('--- inside the test function for hydrol_soil_coef ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_coef/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['a', 'b', 'd', 'k']
    for (variable, value) in zip(modif_var, [gm.a, gm.b, gm.d, gm.k]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gm = global_module_soil_coef()
    gm.declaration_initialization()
    ins = np.int32(0)
    mcr = np.zeros((gm.kjpindex,), dtype=np.float64)
    mcs = np.zeros((gm.kjpindex,), dtype=np.float64)
    njsc = np.zeros((gm.kjpindex,), dtype=np.int32)
    ins = read_dummy(ins, mcr, mcs, njsc)
    hydrol_soil_coef(gm, mcr, mcs, ins, njsc)
    test_hydrol_soil_coef(gm)
if __name__ == '__main__':
    main()
```

## hydrol_soil_setup

```python
import os
import time
import functools
import numpy as np
from scipy.io import FortranFile
# from global_module import Global_module

def read_dummy(ins):
    print(f'--- inside the read dummy routine for hydrol_soil_setup ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_setup/dummy.bin'
    ffile = FortranFile(path, 'r')
    ins = ffile.read_ints(np.int32)[0]
    return ins

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        return result
    return wrapper

@timer
def hydrol_soil_setup(gm, ins):
    temp3 = gm.w_time * (gm.dt_sechiba / gm.one_day) / gm.deux
    temp4 = (gm.un - gm.w_time) * (gm.dt_sechiba / gm.one_day) / gm.deux
    for ji in range(0, gm.kjpindex, 1):
        gm.e[ji, 0] = gm.zero
        gm.f[ji, 0] = gm.trois * gm.dz[1] / gm.huit + temp3 * ((gm.d[ji, 0] + gm.d[ji, 1]) / gm.dz[1] + gm.a[ji, 0])
        gm.g1[ji, 0] = gm.dz[1] / gm.huit - temp3 * ((gm.d[ji, 0] + gm.d[ji, 1]) / gm.dz[1] - gm.a[ji, 1])
        gm.ep[ji, 0] = gm.zero
        gm.fp[ji, 0] = gm.trois * gm.dz[1] / gm.huit - temp4 * ((gm.d[ji, 0] + gm.d[ji, 1]) / gm.dz[1] + gm.a[ji, 0])
        gm.gp[ji, 0] = gm.dz[1] / gm.huit + temp4 * ((gm.d[ji, 0] + gm.d[ji, 1]) / gm.dz[1] - gm.a[ji, 1])
    for jsl in range(1, gm.nslm - 1, 1):
        for ji in range(0, gm.kjpindex, 1):
            gm.e[ji, jsl] = gm.dz[jsl] / gm.huit - temp3 * ((gm.d[ji, jsl] + gm.d[ji, jsl - 1]) / gm.dz[jsl] + gm.a[ji, jsl - 1])
            gm.f[ji, jsl] = gm.trois * (gm.dz[jsl] + gm.dz[jsl + 1]) / gm.huit + temp3 * ((gm.d[ji, jsl] + gm.d[ji, jsl - 1]) / gm.dz[jsl] + (gm.d[ji, jsl] + gm.d[ji, jsl + 1]) / gm.dz[jsl + 1])
            gm.g1[ji, jsl] = gm.dz[jsl + 1] / gm.huit - temp3 * ((gm.d[ji, jsl] + gm.d[ji, jsl + 1]) / gm.dz[jsl + 1] - gm.a[ji, jsl + 1])
            gm.ep[ji, jsl] = gm.dz[jsl] / gm.huit + temp4 * ((gm.d[ji, jsl] + gm.d[ji, jsl - 1]) / gm.dz[jsl] + gm.a[ji, jsl - 1])
            gm.fp[ji, jsl] = gm.trois * (gm.dz[jsl] + gm.dz[jsl + 1]) / gm.huit - temp4 * ((gm.d[ji, jsl] + gm.d[ji, jsl - 1]) / gm.dz[jsl] + (gm.d[ji, jsl] + gm.d[ji, jsl + 1]) / gm.dz[jsl + 1])
            gm.gp[ji, jsl] = gm.dz[jsl + 1] / gm.huit + temp4 * ((gm.d[ji, jsl] + gm.d[ji, jsl + 1]) / gm.dz[jsl + 1] - gm.a[ji, jsl + 1])
    for ji in range(0, gm.kjpindex, 1):
        gm.e[ji, gm.nslm - 1] = gm.dz[gm.nslm - 1] / gm.huit - temp3 * ((gm.d[ji, gm.nslm - 1] + gm.d[ji, gm.nslm - 2]) / gm.dz[gm.nslm - 1] + gm.a[ji, gm.nslm - 2])
        gm.f[ji, gm.nslm - 1] = gm.trois * gm.dz[gm.nslm - 1] / gm.huit + temp3 * ((gm.d[ji, gm.nslm - 1] + gm.d[ji, gm.nslm - 2]) / gm.dz[gm.nslm - 1] - gm.a[ji, gm.nslm - 1] * (gm.un - gm.deux * gm.free_drain_coef[ji, ins - 1]))
        gm.g1[ji, gm.nslm - 1] = gm.zero
        gm.ep[ji, gm.nslm - 1] = gm.dz[gm.nslm - 1] / gm.huit + temp4 * ((gm.d[ji, gm.nslm - 1] + gm.d[ji, gm.nslm - 2]) / gm.dz[gm.nslm - 1] + gm.a[ji, gm.nslm - 2])
        gm.fp[ji, gm.nslm - 1] = gm.trois * gm.dz[gm.nslm - 1] / gm.huit - temp4 * ((gm.d[ji, gm.nslm - 1] + gm.d[ji, gm.nslm - 2]) / gm.dz[gm.nslm - 1] - gm.a[ji, gm.nslm - 1] * (gm.un - gm.deux * gm.free_drain_coef[ji, ins - 1]))
        gm.gp[ji, gm.nslm - 1] = gm.zero

def test_hydrol_soil_setup(gm):
    print('--- inside the test function for hydrol_soil_setup ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_setup/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['e', 'ep', 'f', 'fp', 'g1', 'gp']
    for (variable, value) in zip(modif_var, [gm.e, gm.ep, gm.f, gm.fp, gm.g1, gm.gp]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gm = global_module_soil_setup()
    gm.declaration_initialization()
    ins = np.int32(0)
    ins = read_dummy(ins)
    hydrol_soil_setup(gm, ins)
    test_hydrol_soil_setup(gm)
if __name__ == '__main__':
    main()
```

## hydrol_split_soil

```python
import os
import time
import functools
import numpy as np
from scipy.io import FortranFile
# from global_module import Global_module

def read_dummy(e_frac, evap_bare_lim, evap_bare_lim_ns, F_absorption, humrel, soiltile, tot_bare_soil, transpir, us, veget_max, vevapnu):
    print(f'--- inside the read dummy routine for hydrol_split_soil ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_split_soil/dummy.bin'
    ffile = FortranFile(path, 'r')
    for var in [e_frac, evap_bare_lim, evap_bare_lim_ns, F_absorption, humrel, soiltile, tot_bare_soil, transpir, us, veget_max, vevapnu]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                var[:] = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif var.dtype == np.int32:
                var[:] = ffile.read_ints(np.int32).reshape(arr_shape, order='F')

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        return result
    return wrapper

@timer
def hydrol_split_soil(gm, veget_max, soiltile, vevapnu, transpir, humrel, evap_bare_lim, evap_bare_lim_ns, tot_bare_soil, us, e_frac, F_absorption):
    tmp_check1 = np.zeros((gm.kjpindex,), dtype=np.float64)
    tmp_check2 = np.zeros((gm.kjpindex,), dtype=np.float64)
    tmp_check3 = np.zeros((gm.kjpindex, gm.nstm), dtype=np.float64)
    gm.precisol_ns[:, :] = gm.zero
    for jv in range(0, gm.nvm, 1):
        for ji in range(0, gm.kjpindex, 1):
            jst = gm.pref_soil_veg[jv] - 1
            if veget_max[ji, jv] > gm.min_sechiba and soiltile[ji, jst] * gm.vegtot[ji] > gm.min_sechiba:
                gm.precisol_ns[ji, jst] = gm.precisol_ns[ji, jst] + gm.precisol[ji, jv] / (soiltile[ji, jst] * gm.vegtot[ji])
    gm.ae_ns[:, :] = gm.zero
    for jst in range(0, gm.nstm, 1):
        for ji in range(0, gm.kjpindex, 1):
            if evap_bare_lim[ji] > gm.min_sechiba:
                gm.ae_ns[ji, jst] = vevapnu[ji] * evap_bare_lim_ns[ji, jst] / evap_bare_lim[ji]
    gm.tr_ns[:, :] = gm.zero
    for jv in range(0, gm.nvm, 1):
        jst = gm.pref_soil_veg[jv] - 1
        for ji in range(0, gm.kjpindex, 1):
            if humrel[ji, jv] > gm.min_sechiba and soiltile[ji, jst] * gm.vegtot[ji] > gm.min_sechiba:
                gm.tr_ns[ji, jst] = gm.tr_ns[ji, jst] + transpir[ji, jv] * (gm.humrelv[ji, jv, jst] / humrel[ji, jv]) / (soiltile[ji, jst] * gm.vegtot[ji])
    gm.rootsink[:, :, :] = gm.zero
    if gm.ok_hydrol_arch:
        for jv in range(0, gm.nvm, 1):
            jst = gm.pref_soil_veg[jv] - 1
            for jsl in range(0, gm.nslm, 1):
                for ji in range(0, gm.kjpindex, 1):
                    if humrel[ji, jv] > gm.min_sechiba and soiltile[ji, jst] * gm.vegtot[ji] > gm.min_sechiba:
                        if gm.is_tuzet_hydrol_arch:
                            gm.rootsink[ji, jsl, jst] = gm.rootsink[ji, jsl, jst] + F_absorption[ji, jv] * e_frac[ji, jv, jsl, jst] * gm.dt_sechiba * gm.kilo_to_unit
                        else:
                            gm.rootsink[ji, jsl, jst] = gm.rootsink[ji, jsl, jst] + transpir[ji, jv] * e_frac[ji, jv, jsl, jst] / (soiltile[ji, jst] * gm.vegtot[ji])
    else:
        for jv in range(0, gm.nvm, 1):
            jst = gm.pref_soil_veg[jv] - 1
            for jsl in range(0, gm.nslm, 1):
                for ji in range(0, gm.kjpindex, 1):
                    if humrel[ji, jv] > gm.min_sechiba and soiltile[ji, jst] * gm.vegtot[ji] > gm.min_sechiba:
                        gm.rootsink[ji, jsl, jst] = gm.rootsink[ji, jsl, jst] + transpir[ji, jv] * (us[ji, jv, jst, jsl] / humrel[ji, jv]) / (soiltile[ji, jst] * gm.vegtot[ji])
    if gm.check_cwrr:
        error = False
        tmp_check1[:] = gm.zero
        for jst in range(0, gm.nstm, 1):
            for ji in range(0, gm.kjpindex, 1):
                tmp_check1[ji] = tmp_check1[ji] + gm.precisol_ns[ji, jst] * soiltile[ji, jst] * gm.vegtot[ji]
        tmp_check2[:] = gm.zero
        for jv in range(0, gm.nvm, 1):
            for ji in range(0, gm.kjpindex, 1):
                tmp_check2[ji] = tmp_check2[ji] + gm.precisol[ji, jv]
        for ji in range(0, gm.kjpindex, 1):
            if np.abs(tmp_check1[ji] - tmp_check2[ji]) > gm.allowed_err:
                print(f'PRECISOL SPLIT FALSE:ji={ji},{tmp_check1[ji]},{tmp_check2[ji]}')
                print(f'err{np.abs(tmp_check1[ji] - tmp_check2[ji])}')
                print(f'vegtot{gm.vegtot[ji]}')
                for jv in range(0, gm.nvm, 1):
                    print(f'jv,veget_max, precisol, vegetmax_soil{jv},{veget_max[ji, jv]},{gm.precisol[ji, jv]},{gm.vegetmax_soil[ji, jv, :]}')
                for jst in range(0, gm.nstm, 1):
                    print(f'jst,precisol_ns{jst},{gm.precisol_ns[ji, jst]}')
                    print(f'soiltile{soiltile[ji, jst]}')
                error = True
                ipslerr_p(2, 'hydrol_split_soil', 'We will STOP in the end of this subroutine.', 'check_CWRR', 'PRECISOL SPLIT FALSE')
        tmp_check1[:] = gm.zero
        for jst in range(0, gm.nstm, 1):
            for ji in range(0, gm.kjpindex, 1):
                tmp_check1[ji] = tmp_check1[ji] + gm.ae_ns[ji, jst] * soiltile[ji, jst] * gm.vegtot[ji]
        for ji in range(0, gm.kjpindex, 1):
            if np.abs(tmp_check1[ji] - vevapnu[ji]) > gm.allowed_err:
                print(f'VEVAPNU SPLIT FALSE:ji, Sum(ae_ns), vevapnu ={ji},{tmp_check1[ji]},{vevapnu[ji]}')
                print(f'err{np.abs(tmp_check1[ji] - vevapnu[ji])}')
                print(f'ae_ns{gm.ae_ns[ji, :]}')
                print(f'vegtot{gm.vegtot[ji]}')
                print(f'evap_bare_lim, evap_bare_lim_ns{evap_bare_lim[ji]},{evap_bare_lim_ns[ji, :]}')
                for jst in range(0, gm.nstm, 1):
                    print(f'jst,ae_ns{jst},{gm.ae_ns[ji, jst]}')
                    print(f'soiltile{soiltile[ji, jst]}')
                error = True
                ipslerr_p(2, 'hydrol_split_soil', 'We will STOP in the end of this subroutine.', 'check_CWRR', 'VEVAPNU SPLIT FALSE')
        tmp_check1[:] = gm.zero
        for jst in range(0, gm.nstm, 1):
            for ji in range(0, gm.kjpindex, 1):
                tmp_check1[ji] = tmp_check1[ji] + gm.tr_ns[ji, jst] * soiltile[ji, jst] * gm.vegtot[ji]
        tmp_check2[:] = gm.zero
        for jv in range(0, gm.nvm, 1):
            for ji in range(0, gm.kjpindex, 1):
                tmp_check2[ji] = tmp_check2[ji] + transpir[ji, jv]
        for ji in range(0, gm.kjpindex, 1):
            if np.abs(tmp_check1[ji] - tmp_check2[ji]) > gm.allowed_err:
                print(f'TRANSPIR SPLIT FALSE:ji={ji},{tmp_check1[ji]},{tmp_check2[ji]}')
                print(f'err{np.abs(tmp_check1[ji] - tmp_check2[ji])}')
                print(f'vegtot{gm.vegtot[ji]}')
                for jv in range(0, gm.nvm, 1):
                    print(f'jv,veget_max, transpir{jv},{veget_max[ji, jv]},{transpir[ji, jv]}')
                    for jst in range(0, gm.nstm, 1):
                        print(f'vegetmax_soil:ji,jv,jst{ji},{jv},{jst},{gm.vegetmax_soil[ji, jv, jst]}')
                for jst in range(0, gm.nstm, 1):
                    print(f'jst,tr_ns{jst},{gm.tr_ns[ji, jst]}')
                    print(f'soiltile{soiltile[ji, jst]}')
                error = True
                ipslerr_p(2, 'hydrol_split_soil', 'We will STOP in the end of this subroutine.', 'check_CWRR', 'TRANSPIR SPLIT FALSE')
        tmp_check3[:, :] = gm.zero
        for jst in range(0, gm.nstm, 1):
            for jsl in range(0, gm.nslm, 1):
                for ji in range(0, gm.kjpindex, 1):
                    tmp_check3[ji, jst] = tmp_check3[ji, jst] + gm.rootsink[ji, jsl, jst]
        for jst in range(0, gm.nstm, 1):
            for ji in range(0, gm.kjpindex, 1):
                if np.abs(tmp_check3[ji, jst] - gm.tr_ns[ji, jst]) > gm.allowed_err:
                    print(f'ROOTSINK SPLIT FALSE:ji,jst={ji},{jst},{tmp_check3[ji, jst]},{gm.tr_ns[ji, jst]}')
                    print(f'err{np.abs(tmp_check3[ji, jst] - gm.tr_ns[ji, jst])}')
                    print(f'HUMREL(jv=1:13){humrel[ji, :]}')
                    print(f'TRANSPIR{transpir[ji, :]}')
                    for jv in range(0, gm.nvm, 1):
                        print(f'jv={jv},us={us[ji, jv, jst, :]}')
                    error = True
                    ipslerr_p(2, 'hydrol_split_soil', 'We will STOP in the end of this subroutine.', 'check_CWRR', 'ROOTSINK SPLIT FALSE')
        if error:
            print('One or more errors have been detected in hydrol_split_soil. Model stops.')
            ipslerr_p(3, 'hydrol_split_soil', 'We will STOP now.', 'One or several fatal errors were found previously.', '')

def test_hydrol_split_soil(gm):
    print('--- inside the test function for hydrol_split_soil ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_split_soil/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['ae_ns', 'precisol_ns', 'rootsink', 'tr_ns']
    for (variable, value) in zip(modif_var, [gm.ae_ns, gm.precisol_ns, gm.rootsink, gm.tr_ns]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gm = global_module_split_soil()
    gm.declaration_initialization()
    e_frac = np.zeros((gm.kjpindex, gm.nvm, gm.nslm, gm.nstm), dtype=np.float64)
    evap_bare_lim = np.zeros((gm.kjpindex,), dtype=np.float64)
    evap_bare_lim_ns = np.zeros((gm.kjpindex, gm.nstm), dtype=np.float64)
    F_absorption = np.zeros((gm.kjpindex, gm.nvm), dtype=np.float64)
    humrel = np.zeros((gm.kjpindex, gm.nvm), dtype=np.float64)
    soiltile = np.zeros((gm.kjpindex, gm.nstm), dtype=np.float64)
    tot_bare_soil = np.zeros((gm.kjpindex,), dtype=np.float64)
    transpir = np.zeros((gm.kjpindex, gm.nvm), dtype=np.float64)
    us = np.zeros((gm.kjpindex, gm.nvm, gm.nstm, gm.nslm), dtype=np.float64)
    veget_max = np.zeros((gm.kjpindex, gm.nvm), dtype=np.float64)
    vevapnu = np.zeros((gm.kjpindex,), dtype=np.float64)
    read_dummy(e_frac, evap_bare_lim, evap_bare_lim_ns, F_absorption, humrel, soiltile, tot_bare_soil, transpir, us, veget_max, vevapnu)
    hydrol_split_soil(gm, veget_max, soiltile, vevapnu, transpir, humrel, evap_bare_lim, evap_bare_lim_ns, tot_bare_soil, us, e_frac, F_absorption)
    test_hydrol_split_soil(gm)
if __name__ == '__main__':
    main()
```

## hydrol_soil_tridiag

```python
import os
import time
import functools
import numpy as np
from scipy.io import FortranFile
# from global_module import Global_module

def read_dummy(ins):
    print(f'--- inside the read dummy routine for hydrol_soil_tridiag ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_tridiag/dummy.bin'
    ffile = FortranFile(path, 'r')
    ins = ffile.read_ints(np.int32)[0]
    return ins

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        return result
    return wrapper

@timer
def hydrol_soil_tridiag(gm, ins):
    bet = np.zeros((gm.kjpindex,), dtype=np.float64)
    gam = np.zeros((gm.kjpindex, gm.nslm), dtype=np.float64)
    for ji in range(0, gm.kjpindex, 1):
        if gm.resolv[ji]:
            bet[ji] = gm.tmat[ji, 0, 1]
            gm.mcl[ji, 0, ins - 1] = gm.rhs[ji, 0] / bet[ji]
    for jsl in range(1, gm.nslm, 1):
        for ji in range(0, gm.kjpindex, 1):
            if gm.resolv[ji]:
                gam[ji, jsl] = gm.tmat[ji, jsl - 1, 2] / bet[ji]
                bet[ji] = gm.tmat[ji, jsl, 1] - gm.tmat[ji, jsl, 0] * gam[ji, jsl]
                gm.mcl[ji, jsl, ins - 1] = (gm.rhs[ji, jsl] - gm.tmat[ji, jsl, 0] * gm.mcl[ji, jsl - 1, ins - 1]) / bet[ji]
    for ji in range(0, gm.kjpindex, 1):
        if gm.resolv[ji]:
            for jsl in range(gm.nslm - 2, -1, -1):
                gm.mcl[ji, jsl, ins - 1] = gm.mcl[ji, jsl, ins - 1] - gam[ji, jsl + 1] * gm.mcl[ji, jsl + 1, ins - 1]

def test_hydrol_soil_tridiag(gm):
    print('--- inside the test function for hydrol_soil_tridiag ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_tridiag/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['mcl']
    for (variable, value) in zip(modif_var, [gm.mcl]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gm = global_module_soil_tridiag()
    gm.declaration_initialization()
    ins = np.int32(0)
    ins = read_dummy(ins)
    hydrol_soil_tridiag(gm, ins)
    test_hydrol_soil_tridiag(gm)
if __name__ == '__main__':
    main()
```

## hydrol_root_profile

```python
import os
import time
import functools
import numpy as np
import logging
from scipy.io import FortranFile
# from module_global import Global_module_hydrol_root_profile

def read_dummy(altmax, sm, smw):
    print(f'--- inside the read dummy routine for hydrol_root_profile ---')
    path = f'/home/ssivanes/Fgpt/benchmark/hydrol_root_profile/dummy.bin'
    ffile = FortranFile(path, 'r')
    for var in [altmax, sm, smw]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                var[:] = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif var.dtype == np.int32:
                var[:] = ffile.read_ints(np.int32).reshape(arr_shape, order='F')

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        path = '/home/ssivanes/Fgpt/benchmark/hydrol_root_profile/time.txt'
        with open(path, 'a') as f:
            f.write(f"[TIMER] Python: '{func.__name__}' executed in {duration:.6f} seconds \n")
        return result
    return wrapper

@timer
def hydrol_root_profile(altmax, sm, smw, root_profile, root_depth, gmhrp):
    count = np.zeros((gmhrp.kjpindex,), dtype=np.float64)
    root_profile_tmp = np.zeros((gmhrp.nslm,), dtype=np.float64)
    root_depth[:, :, :] = gmhrp.zero
    for ji in range(0, gmhrp.kjpindex, 1):
        for jv in range(0, gmhrp.nvm, 1):
            root_depth_tmp = np.minimum(np.minimum(altmax[ji, jv], gmhrp.maxaltmax), np.minimum(gmhrp.zdr[gmhrp.nslm], gmhrp.max_root_depth[jv]))
            root_depth[ji, jv, gmhrp.inode - 1] = np.argmin(np.abs(root_depth_tmp - gmhrp.znh[:]), axis=0) + 1
            root_depth[ji, jv, gmhrp.iinterface - 1] = np.argmin(np.abs(root_depth_tmp - gmhrp.zdr[:]), axis=0)
    root_depth[:, :, gmhrp.inode - 1] = np.where(root_depth[:, :, gmhrp.inode - 1] <= 2, 2, root_depth[:, :, gmhrp.inode - 1])
    root_depth[:, :, gmhrp.iinterface - 1] = np.where(root_depth[:, :, gmhrp.iinterface - 1] <= 2, 2, root_depth[:, :, gmhrp.iinterface - 1])
    root_profile[:, :, :, gmhrp.istruc - 1] = gmhrp.zero
    for jv in range(0, gmhrp.nvm, 1):
        for ji in range(0, gmhrp.kjpindex, 1):
            z_top = gmhrp.zdr[1]
            z_bottom = gmhrp.zdr[int(root_depth[ji, jv, gmhrp.iinterface - 1])]
            rpc = gmhrp.un / (np.exp(-(z_top * gmhrp.humcste[jv])) - np.exp(-(z_bottom * gmhrp.humcste[jv])))
            for jsl in range(1, int(root_depth[ji, jv, gmhrp.iinterface - 1]), 1):
                root_profile[ji, jv, jsl, gmhrp.istruc - 1] = rpc * (np.exp(-(gmhrp.zdr[jsl] * gmhrp.humcste[jv])) - np.exp(-(gmhrp.zdr[jsl + 1] * gmhrp.humcste[jv])))
            root_profile[ji, jv, 0, gmhrp.istruc - 1] = gmhrp.zero
            if gmhrp.err_act > 1:
                if np.abs(np.sum(root_profile[ji, jv, :, gmhrp.istruc - 1]) - gmhrp.un) > 100 * np.finfo(np.float64).eps:
                    logging.info(f'pixel, PFT, sum of structural root_profile,{ji},{jv},{np.sum(root_profile[ji, jv, :, istruc - 1])}')
                    logging.info(f'hydrol_root_profile, structural root_profile,{root_profile[ji, jv, :, istruc - 1]}')
                    logging.error('Exception: hydrol.f90 structural root profile does not add up to 1 Check its calculation ')
    root_profile[:, :, :, gmhrp.ifunc - 1] = gmhrp.zero
    for jv in range(1, gmhrp.nvm, 1):
        for ji in range(0, gmhrp.kjpindex, 1):
            root_profile_tmp[:] = gmhrp.zero
            for jsl in range(1, int(root_depth[ji, jv, gmhrp.iinterface - 1]), 1):
                root_profile_tmp[jsl] = np.maximum(gmhrp.zero, sm[ji, jsl] - smw[ji, jsl])
            if np.sum(root_profile_tmp[:]) > gmhrp.min_sechiba:
                root_profile[ji, jv, :, gmhrp.ifunc - 1] = root_profile_tmp[:] / np.sum(root_profile_tmp[:])
            else:
                root_profile[ji, jv, :, gmhrp.ifunc - 1] = gmhrp.zero
            root_profile[ji, jv, 0, gmhrp.ifunc - 1] = gmhrp.zero
            if gmhrp.err_act > 1:
                if np.abs(np.sum(root_profile[ji, jv, :, gmhrp.ifunc - 1]) - gmhrp.un) > 100 * np.finfo(np.float64).eps and np.sum(root_profile_tmp[:]) > gmhrp.min_sechiba:
                    logging.info(f'pixel, PFT, sum of functional root_profile,{ji},{jv},{np.sum(root_profile[ji, jv, :, ifunc - 1])}')
                    logging.info(f'hydrol_root_profile, functional root_profile,{root_profile[ji, jv, :, ifunc - 1]}')
                    logging.error('Exception: hydrol.f90 functional root profile does not add up to 1 Check its calculation ')

def test_hydrol_root_profile(root_profile, root_depth):
    print('--- inside the test function for hydrol_root_profile ---')
    path = f'/home/ssivanes/Fgpt/benchmark/hydrol_root_profile/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['root_depth', 'root_profile']
    for (variable, value) in zip(modif_var, [root_depth, root_profile]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gmhrp = global_module_root_profile()
    gmhrp.declaration_initialization()
    altmax = np.zeros((gmhrp.kjpindex, gmhrp.nvm), dtype=np.float64)
    root_depth = np.zeros((gmhrp.kjpindex, gmhrp.nvm, gmhrp.ndepths), dtype=np.float64)
    root_profile = np.zeros((gmhrp.kjpindex, gmhrp.nvm, gmhrp.nslm, gmhrp.nroot_prof), dtype=np.float64)
    sm = np.zeros((gmhrp.kjpindex, gmhrp.nslm), dtype=np.float64)
    smw = np.zeros((gmhrp.kjpindex, gmhrp.nslm), dtype=np.float64)
    read_dummy(altmax, sm, smw)
    hydrol_root_profile(altmax, sm, smw, root_profile, root_depth, gmhrp)
    test_hydrol_root_profile(root_profile, root_depth)
if __name__ == '__main__':
    main()
```

## hydrol_soil_infilt

```python
import os
import time
import functools
import numpy as np
from scipy.io import FortranFile
# from global_module import Global_module

def read_dummy(gm, avan, flux_infilt, ins, ks, mcfc, mcr, mcs, mcw, njsc, nvan, stempdiag):
    print(f'--- inside the read dummy routine for hydrol_soil_infilt ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_infilt/dummy.bin'
    ffile = FortranFile(path, 'r')
    avan[:] = ffile.read_reals(np.float64).reshape((gm.kjpindex,), order='F')
    flux_infilt[:] = ffile.read_reals(np.float64).reshape((gm.kjpindex,), order='F')
    ins = ffile.read_ints(np.int32)[0]
    for var in [ks, mcfc, mcr, mcs, mcw, njsc, nvan, stempdiag]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                var[:] = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif var.dtype == np.int32:
                var[:] = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
    return ins

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        return result
    return wrapper

@timer
def hydrol_soil_infilt(gm, ks, nvan, avan, mcr, mcs, mcfc, mcw, ins, njsc, flux_infilt, stempdiag, qinfilt_ns, ru_infilt, check):
    wat_inf_pot = np.zeros((gm.kjpindex,), dtype=np.float64)
    wat_inf = np.zeros((gm.kjpindex,), dtype=np.float64)
    dt_tmp = np.zeros((gm.kjpindex,), dtype=np.float64)
    dt_inf = np.zeros((gm.kjpindex,), dtype=np.float64)
    infilt_tmp = np.zeros((gm.kjpindex,), dtype=np.float64)
    infilt_tot = np.zeros((gm.kjpindex,), dtype=np.float64)
    flux_tmp = np.zeros((gm.kjpindex,), dtype=np.float64)
    tmci = np.zeros((gm.kjpindex,), dtype=np.float64)
    tmcf = np.zeros((gm.kjpindex,), dtype=np.float64)
    if gm.check_cwrr:
        tmci[:] = gm.dz[1] * (gm.trois * gm.mc[:, 0, ins - 1] + gm.mc[:, 1, ins - 1]) / gm.huit
        for jsl in range(1, gm.nslm - 1, 1):
            tmci[:] = tmci[:] + gm.dz[jsl] * (gm.trois * gm.mc[:, jsl, ins - 1] + gm.mc[:, jsl - 1, ins - 1]) / gm.huit + gm.dz[jsl + 1] * (gm.trois * gm.mc[:, jsl, ins - 1] + gm.mc[:, jsl + 1, ins - 1]) / gm.huit
        tmci[:] = tmci[:] + gm.dz[gm.nslm - 1] * (gm.trois * gm.mc[:, gm.nslm - 1, ins - 1] + gm.mc[:, gm.nslm - 2, ins - 1]) / gm.huit
    for ji in range(0, gm.kjpindex, 1):
        wat_inf_pot[ji] = np.maximum((mcs[ji] - gm.mc[ji, 0, ins - 1]) * gm.dz[1] / gm.deux, gm.zero)
        wat_inf[ji] = np.minimum(wat_inf_pot[ji], flux_infilt[ji])
        gm.mc[ji, 0, ins - 1] = gm.mc[ji, 0, ins - 1] + wat_inf[ji] * gm.deux / gm.dz[1]
    dt_tmp[:] = gm.dt_sechiba / gm.one_day
    infilt_tot[:] = wat_inf[:]
    flux_tmp[:] = (flux_infilt[:] - wat_inf[:]) / dt_tmp[:]
    for jsl in range(1, gm.nslm - 1, 1):
        for ji in range(0, gm.kjpindex, 1):
            k_m = (gm.k[ji, jsl] + ks[ji] * gm.kfact[jsl - 1, ji] * gm.kfact_root[ji, jsl, ins - 1]) / gm.deux
            if gm.ok_freeze_cwrr:
                if stempdiag[ji, jsl] < gm.ZeroCelsius:
                    k_m = gm.k[ji, jsl]
            infilt_tmp[ji] = k_m * (gm.un - np.exp(-(flux_tmp[ji] / k_m)))
            wat_inf_pot[ji] = np.maximum((mcs[ji] - gm.mc[ji, jsl, ins - 1]) * (gm.dz[jsl] + gm.dz[jsl + 1]) / gm.deux, gm.zero)
            if infilt_tmp[ji] > gm.min_sechiba:
                dt_inf[ji] = np.minimum(wat_inf_pot[ji] / infilt_tmp[ji], dt_tmp[ji])
                if dt_inf[ji] * infilt_tmp[ji] > flux_infilt[ji] - infilt_tot[ji]:
                    dt_inf[ji] = np.maximum(flux_infilt[ji] - infilt_tot[ji], gm.zero) / infilt_tmp[ji]
            else:
                dt_inf[ji] = dt_tmp[ji]
            wat_inf[ji] = dt_inf[ji] * infilt_tmp[ji]
            gm.mc[ji, jsl, ins - 1] = gm.mc[ji, jsl, ins - 1] + wat_inf[ji] * gm.deux / (gm.dz[jsl] + gm.dz[jsl + 1])
            dt_tmp[ji] = dt_tmp[ji] - dt_inf[ji]
            infilt_tot[ji] = infilt_tot[ji] + infilt_tmp[ji] * dt_inf[ji]
    ru_infilt[:, ins - 1] = flux_infilt[:] - infilt_tot[:]
    qinfilt_ns[:, ins - 1] = infilt_tot[:]
    if gm.check_cwrr:
        tmcf[:] = gm.dz[1] * (gm.trois * gm.mc[:, 0, ins - 1] + gm.mc[:, 1, ins - 1]) / gm.huit
        for jsl in range(1, gm.nslm - 1, 1):
            tmcf[:] = tmcf[:] + gm.dz[jsl] * (gm.trois * gm.mc[:, jsl, ins - 1] + gm.mc[:, jsl - 1, ins - 1]) / gm.huit + gm.dz[jsl + 1] * (gm.trois * gm.mc[:, jsl, ins - 1] + gm.mc[:, jsl + 1, ins - 1]) / gm.huit
        tmcf[:] = tmcf[:] + gm.dz[gm.nslm - 1] * (gm.trois * gm.mc[:, gm.nslm - 1, ins - 1] + gm.mc[:, gm.nslm - 2, ins - 1]) / gm.huit
        check[:, ins - 1] = tmcf[:] - (tmci[:] + infilt_tot[:])
    for ji in range(0, gm.kjpindex, 1):
        if infilt_tot[ji] < -gm.min_sechiba or infilt_tot[ji] > flux_infilt[ji] + gm.min_sechiba:
            print(f'Error in the calculation of infilt tot{infilt_tot[ji]}')
            print(f'k, ji, jst, mc{gm.k[ji, :2]},{ji},{ins},{gm.mc[ji, 0, ins - 1]}')
            ipslerr_p(3, 'hydrol_soil_infilt', 'We will STOP now.', 'Error in calculation of infilt tot', '')

def test_hydrol_soil_infilt(gm, qinfilt_ns, ru_infilt, check):
    print('--- inside the test function for hydrol_soil_infilt ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_infilt/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['check', 'mc', 'qinfilt_ns', 'ru_infilt']
    for (variable, value) in zip(modif_var, [check, gm.mc, qinfilt_ns, ru_infilt]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gm = global_module_soil_infilt()
    gm.declaration_initialization()
    avan = np.zeros((gm.kjpindex,), dtype=np.float64)
    check = np.zeros((gm.kjpindex, gm.nstm), dtype=np.float64)
    flux_infilt = np.zeros((gm.kjpindex,), dtype=np.float64)
    ins = np.int32(0)
    ks = np.zeros((gm.kjpindex,), dtype=np.float64)
    mcfc = np.zeros((gm.kjpindex,), dtype=np.float64)
    mcr = np.zeros((gm.kjpindex,), dtype=np.float64)
    mcs = np.zeros((gm.kjpindex,), dtype=np.float64)
    mcw = np.zeros((gm.kjpindex,), dtype=np.float64)
    njsc = np.zeros((gm.kjpindex,), dtype=np.int32)
    nvan = np.zeros((gm.kjpindex,), dtype=np.float64)
    qinfilt_ns = np.zeros((gm.kjpindex, gm.nstm), dtype=np.float64)
    ru_infilt = np.zeros((gm.kjpindex, gm.nstm), dtype=np.float64)
    stempdiag = np.zeros((gm.kjpindex, gm.nslm), dtype=np.float64)
    ins = read_dummy(gm, avan, flux_infilt, ins, ks, mcfc, mcr, mcs, mcw, njsc, nvan, stempdiag)
    hydrol_soil_infilt(gm, ks, nvan, avan, mcr, mcs, mcfc, mcw, ins, njsc, flux_infilt, stempdiag, qinfilt_ns, ru_infilt, check)
    test_hydrol_soil_infilt(gm, qinfilt_ns, ru_infilt, check)
if __name__ == '__main__':
    main()
```

## hydrol_soil_smooth_over_mcs2

```python
import os
import time
import functools
import numpy as np
from scipy.io import FortranFile
# from global_module import Global_module

def read_dummy(ins, mcs, njsc):
    print(f'--- inside the read dummy routine for hydrol_soil_smooth_over_mcs2 ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_smooth_over_mcs2/dummy.bin'
    ffile = FortranFile(path, 'r')
    ins = ffile.read_ints(np.int32)[0]
    for var in [mcs, njsc]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                var[:] = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif var.dtype == np.int32:
                var[:] = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
    return ins

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        return result
    return wrapper

@timer
def hydrol_soil_smooth_over_mcs2(gm, mcs, ins, njsc, is_over_mcs, rudr_corr, check):
    excess = np.zeros((gm.kjpindex, gm.nslm), dtype=np.float64)
    tmci = np.zeros((gm.kjpindex,), dtype=np.float64)
    tmcf = np.zeros((gm.kjpindex,), dtype=np.float64)
    if gm.check_cwrr:
        tmci[:] = gm.dz[1] * (gm.trois * gm.mc[:, 0, ins - 1] + gm.mc[:, 1, ins - 1]) / gm.huit
        for jsl in range(1, gm.nslm - 1, 1):
            tmci[:] = tmci[:] + gm.dz[jsl] * (gm.trois * gm.mc[:, jsl, ins - 1] + gm.mc[:, jsl - 1, ins - 1]) / gm.huit + gm.dz[jsl + 1] * (gm.trois * gm.mc[:, jsl, ins - 1] + gm.mc[:, jsl + 1, ins - 1]) / gm.huit
        tmci[:] = tmci[:] + gm.dz[gm.nslm - 1] * (gm.trois * gm.mc[:, gm.nslm - 1, ins - 1] + gm.mc[:, gm.nslm - 2, ins - 1]) / gm.huit
    for jsl in range(0, gm.nslm, 1):
        for ji in range(0, gm.kjpindex, 1):
            excess[ji, jsl] = np.maximum(gm.mc[ji, jsl, ins - 1] - mcs[ji], gm.zero)
            gm.mc[ji, jsl, ins - 1] = gm.mc[ji, jsl, ins - 1] - excess[ji, jsl]
    for ji in range(0, gm.kjpindex, 1):
        rudr_corr[ji, ins - 1] = gm.dz[1] * (gm.trois * excess[ji, 0] + excess[ji, 1]) / gm.huit
    for jsl in range(1, gm.nslm - 1, 1):
        for ji in range(0, gm.kjpindex, 1):
            rudr_corr[ji, ins - 1] = rudr_corr[ji, ins - 1] + gm.dz[jsl] * (gm.trois * excess[ji, jsl] + excess[ji, jsl - 1]) / gm.huit + gm.dz[jsl + 1] * (gm.trois * excess[ji, jsl] + excess[ji, jsl + 1]) / gm.huit
    for ji in range(0, gm.kjpindex, 1):
        rudr_corr[ji, ins - 1] = rudr_corr[ji, ins - 1] + gm.dz[gm.nslm - 1] * (gm.trois * excess[ji, gm.nslm - 1] + excess[ji, gm.nslm - 2]) / gm.huit
        is_over_mcs[ji] = False
    if gm.check_cwrr:
        tmcf[:] = gm.dz[1] * (gm.trois * gm.mc[:, 0, ins - 1] + gm.mc[:, 1, ins - 1]) / gm.huit
        for jsl in range(1, gm.nslm - 1, 1):
            tmcf[:] = tmcf[:] + gm.dz[jsl] * (gm.trois * gm.mc[:, jsl, ins - 1] + gm.mc[:, jsl - 1, ins - 1]) / gm.huit + gm.dz[jsl + 1] * (gm.trois * gm.mc[:, jsl, ins - 1] + gm.mc[:, jsl + 1, ins - 1]) / gm.huit
        tmcf[:] = tmcf[:] + gm.dz[gm.nslm - 1] * (gm.trois * gm.mc[:, gm.nslm - 1, ins - 1] + gm.mc[:, gm.nslm - 2, ins - 1]) / gm.huit
        check[:, ins - 1] = tmcf[:] - (tmci[:] - rudr_corr[:, ins - 1])

def test_hydrol_soil_smooth_over_mcs2(gm, is_over_mcs, rudr_corr, check):
    print('--- inside the test function for hydrol_soil_smooth_over_mcs2 ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_smooth_over_mcs2/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['check', 'is_over_mcs', 'mc', 'rudr_corr']
    for (variable, value) in zip(modif_var, [check, is_over_mcs, gm.mc, rudr_corr]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gm = global_module_soil_smooth_over_mcs2()
    gm.declaration_initialization()
    check = np.zeros((gm.kjpindex, gm.nstm), dtype=np.float64)
    ins = np.int32(0)
    is_over_mcs = np.zeros((gm.kjpindex,), dtype=np.int32)
    mcs = np.zeros((gm.kjpindex,), dtype=np.float64)
    njsc = np.zeros((gm.kjpindex,), dtype=np.int32)
    rudr_corr = np.zeros((gm.kjpindex, gm.nstm), dtype=np.float64)
    ins = read_dummy(ins, mcs, njsc)
    hydrol_soil_smooth_over_mcs2(gm, mcs, ins, njsc, is_over_mcs, rudr_corr, check)
    test_hydrol_soil_smooth_over_mcs2(gm, is_over_mcs, rudr_corr, check)
if __name__ == '__main__':
    main()
```

## hydrol_soil_smooth_under_mcr

```python
import os
import time
import functools
import numpy as np
from scipy.io import FortranFile
# from global_module import Global_module

def read_dummy(ins, mcr, njsc):
    print(f'--- inside the read dummy routine for hydrol_soil_smooth_under_mcr ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_smooth_under_mcr/dummy.bin'
    ffile = FortranFile(path, 'r')
    ins = ffile.read_ints(np.int32)[0]
    for var in [mcr, njsc]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                var[:] = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif var.dtype == np.int32:
                var[:] = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
    return ins

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        return result
    return wrapper

@timer
def hydrol_soil_smooth_under_mcr(gm, mcr, ins, njsc, is_under_mcr, check):
    excessji = np.zeros((gm.kjpindex,), dtype=np.float64)
    tmci = np.zeros((gm.kjpindex,), dtype=np.float64)
    tmcf = np.zeros((gm.kjpindex,), dtype=np.float64)
    if gm.check_cwrr:
        tmci[:] = gm.dz[1] * (gm.trois * gm.mc[:, 0, ins - 1] + gm.mc[:, 1, ins - 1]) / gm.huit
        for jsl in range(1, gm.nslm - 1, 1):
            tmci[:] = tmci[:] + gm.dz[jsl] * (gm.trois * gm.mc[:, jsl, ins - 1] + gm.mc[:, jsl - 1, ins - 1]) / gm.huit + gm.dz[jsl + 1] * (gm.trois * gm.mc[:, jsl, ins - 1] + gm.mc[:, jsl + 1, ins - 1]) / gm.huit
        tmci[:] = tmci[:] + gm.dz[gm.nslm - 1] * (gm.trois * gm.mc[:, gm.nslm - 1, ins - 1] + gm.mc[:, gm.nslm - 2, ins - 1]) / gm.huit
    for jsl in range(0, gm.nslm - 2, 1):
        for ji in range(0, gm.kjpindex, 1):
            excess = np.maximum(mcr[ji] - gm.mc[ji, jsl, ins - 1], gm.zero)
            gm.mc[ji, jsl, ins - 1] = gm.mc[ji, jsl, ins - 1] + excess
            gm.mc[ji, jsl + 1, ins - 1] = gm.mc[ji, jsl + 1, ins - 1] - excess * (gm.dz[jsl] + gm.dz[jsl + 1]) / (gm.dz[jsl + 1] + gm.dz[jsl + 2])
    jsl = gm.nslm - 2
    for ji in range(0, gm.kjpindex, 1):
        excess = np.maximum(mcr[ji] - gm.mc[ji, jsl, ins - 1], gm.zero)
        gm.mc[ji, jsl, ins - 1] = gm.mc[ji, jsl, ins - 1] + excess
        gm.mc[ji, jsl + 1, ins - 1] = gm.mc[ji, jsl + 1, ins - 1] - excess * (gm.dz[jsl] + gm.dz[jsl + 1]) / gm.dz[jsl + 1]
    jsl = gm.nslm - 1
    for ji in range(0, gm.kjpindex, 1):
        excess = np.maximum(mcr[ji] - gm.mc[ji, jsl, ins - 1], gm.zero)
        gm.mc[ji, jsl, ins - 1] = gm.mc[ji, jsl, ins - 1] + excess
        gm.mc[ji, jsl - 1, ins - 1] = gm.mc[ji, jsl - 1, ins - 1] - excess * gm.dz[jsl] / (gm.dz[jsl - 1] + gm.dz[jsl])
    for jsl in range(gm.nslm - 2, -2, -1):
        for ji in range(0, gm.kjpindex, 1):
            excess = np.maximum(mcr[ji] - gm.mc[ji, jsl, ins - 1], gm.zero)
            gm.mc[ji, jsl, ins - 1] = gm.mc[ji, jsl, ins - 1] + excess
            gm.mc[ji, jsl - 1, ins - 1] = gm.mc[ji, jsl - 1, ins - 1] - excess * (gm.dz[jsl] + gm.dz[jsl + 1]) / (gm.dz[jsl - 1] + gm.dz[jsl])
    for ji in range(0, gm.kjpindex, 1):
        excessji[ji] = gm.mask_soiltile[ji, ins - 1] * np.maximum(mcr[ji] - gm.mc[ji, 0, ins - 1], gm.zero)
    for ji in range(0, gm.kjpindex, 1):
        gm.mc[ji, 0, ins - 1] = gm.mc[ji, 0, ins - 1] + excessji[ji]
        is_under_mcr[ji, ins - 1] = excessji[ji] > gm.min_sechiba
    for jsl in range(0, gm.nslm, 1):
        for ji in range(0, gm.kjpindex, 1):
            gm.mc[ji, jsl, ins - 1] = gm.mc[ji, jsl, ins - 1] - excessji[ji] * gm.dz[1] / (gm.deux * gm.zmaxh * gm.mille)
    for jsl in range(0, gm.nslm, 1):
        for ji in range(0, gm.kjpindex, 1):
            gm.mc[ji, jsl, ins - 1] = gm.mask_soiltile[ji, ins - 1] * gm.mc[ji, jsl, ins - 1]
    if gm.check_cwrr:
        tmcf[:] = gm.dz[1] * (gm.trois * gm.mc[:, 0, ins - 1] + gm.mc[:, 1, ins - 1]) / gm.huit
        for jsl in range(1, gm.nslm - 1, 1):
            tmcf[:] = tmcf[:] + gm.dz[jsl] * (gm.trois * gm.mc[:, jsl, ins - 1] + gm.mc[:, jsl - 1, ins - 1]) / gm.huit + gm.dz[jsl + 1] * (gm.trois * gm.mc[:, jsl, ins - 1] + gm.mc[:, jsl + 1, ins - 1]) / gm.huit
        tmcf[:] = tmcf[:] + gm.dz[gm.nslm - 1] * (gm.trois * gm.mc[:, gm.nslm - 1, ins - 1] + gm.mc[:, gm.nslm - 2, ins - 1]) / gm.huit
        check[:, ins - 1] = tmcf[:] - tmci[:]

def test_hydrol_soil_smooth_under_mcr(gm, is_under_mcr, check):
    print('--- inside the test function for hydrol_soil_smooth_under_mcr ---')
    path = f'/data/ssivanes/Fgpt/benchmark/hydrol_soil_smooth_under_mcr/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['check', 'is_under_mcr', 'mc']
    for (variable, value) in zip(modif_var, [check, is_under_mcr, gm.mc]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gm = global_module_soil_smooth_under_mcr()
    gm.declaration_initialization()
    check = np.zeros((gm.kjpindex, gm.nstm), dtype=np.float64)
    ins = np.int32(0)
    is_under_mcr = np.zeros((gm.kjpindex, gm.nstm), dtype=np.int32)
    mcr = np.zeros((gm.kjpindex,), dtype=np.float64)
    njsc = np.zeros((gm.kjpindex,), dtype=np.int32)
    ins = read_dummy(ins, mcr, njsc)
    hydrol_soil_smooth_under_mcr(gm, mcr, ins, njsc, is_under_mcr, check)
    test_hydrol_soil_smooth_under_mcr(gm, is_under_mcr, check)
if __name__ == '__main__':
    main()
```

```python
import numpy as np
import os
from scipy.io import FortranFile

class Global_module_hydrol_vegupd:

    def __init__(self):
        self.nsnow = np.int32(3)
        self.nslm = np.int32(11)
        self.nvm = np.int32(15)
        self.nstm = np.int32(3)
        self.kjpindex = np.int32(4717)
        self.ier = np.int32(0)
        self.ic0 = np.int32(0)
        self.ic = np.int32(0)
        self.icr = np.float64(0.0)
        self.start_time = np.float64(0.0)
        self.stop_time = np.float64(0.0)
        self.printlev = np.int32(2)
        self.zero = np.float64(0.0)
        self.huit = np.float64(8.0)
        self.numout = np.int32(6)
        self.min_sechiba = np.float64(1e-08)
        self.trois = np.float64(3.0)
        self.un = np.float64(1.0)
        self.ok_bare_soil_new = np.bool(False)
        self.dz = np.zeros((self.nslm,), dtype=np.float64)
        self.frac_bare_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.humtot = np.zeros((self.kjpindex,), dtype=np.float64)
        self.mask_soiltile = np.zeros((self.kjpindex, self.nstm), dtype=np.int32)
        self.mask_veget = np.zeros((self.kjpindex, self.nvm), dtype=np.int32)
        self.mc = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)
        self.pref_soil_veg = np.zeros((self.nvm,), dtype=np.int32)
        self.resdist = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.tmc = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.vegetmax_soil = np.zeros((self.kjpindex, self.nvm, self.nstm), dtype=np.float64)
        self.vegtot = np.zeros((self.kjpindex,), dtype=np.float64)
        self.vegtot_old = np.zeros((self.kjpindex,), dtype=np.float64)
        self.water2infilt = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)

    def declaration_initialization(self):
        print(f'--- add the declaration and initialization in module global ---')
        path = f'/home/ssivanes/Fgpt/benchmark/hydrol_vegupd/global.bin'
        ffile = FortranFile(path, 'r')
        self.ok_bare_soil_new = np.bool(ffile.read_ints(np.int32)[0])
        for attr_name in ['dz', 'frac_bare_ns', 'humtot', 'mask_soiltile', 'mask_veget', 'mc', 'pref_soil_veg', 'resdist', 'tmc', 'vegetmax_soil', 'vegtot', 'vegtot_old', 'water2infilt']:
            attribute = getattr(self, attr_name)
            if attribute.dtype == np.float64 and isinstance(attribute, np.ndarray):
                arr_shape = attribute.shape
                setattr(self, attr_name, ffile.read_reals(np.float64).reshape(arr_shape, order='F'))
            elif attribute.dtype == np.int32 and isinstance(attribute, np.ndarray):
                arr_shape = attribute.shape
                setattr(self, attr_name, ffile.read_ints(np.int32).reshape(arr_shape, order='F'))
```

```python
import os
import time
import functools
import numpy as np
import logging
from scipy.io import FortranFile
# from module_global import Global_module_hydrol_vegupd

class Hydrol_vegupd:

    def __init__(self):
        self.nsnow = np.int32(3)
        self.nslm = np.int32(11)
        self.nvm = np.int32(15)
        self.nstm = np.int32(3)
        self.kjpindex = np.int32(4717)
        self.ier = np.int32(0)
        self.ic0 = np.int32(0)
        self.ic = np.int32(0)
        self.icr = np.float64(0.0)
        self.start_time = np.float64(0.0)
        self.stop_time = np.float64(0.0)
        self.gmhv = Global_module_hydrol_vegupd()
        self.gmhv.declaration_initialization()
        self.drain_upd = np.zeros((self.kjpindex,), dtype=np.float64)
        self.frac_bare = np.zeros((self.kjpindex, self.nvm), dtype=np.float64)
        self.qsintveg = np.zeros((self.kjpindex, self.nvm), dtype=np.float64)
        self.runoff_upd = np.zeros((self.kjpindex,), dtype=np.float64)
        self.soiltile = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.veget = np.zeros((self.kjpindex, self.nvm), dtype=np.float64)
        self.veget_max = np.zeros((self.kjpindex, self.nvm), dtype=np.float64)

    def hydrol_tmc_update(self):
        vmr = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        vmr_sum = np.zeros((self.kjpindex,), dtype=np.float64)
        delvegtot = np.zeros((self.kjpindex,), dtype=np.float64)
        mc_dilu = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        infil_dilu = np.zeros((self.kjpindex,), dtype=np.float64)
        tmc_old = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        water2infilt_old = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        qsintveg_old = np.zeros((self.kjpindex, self.nvm), dtype=np.float64)
        test = np.zeros((self.kjpindex,), dtype=np.float64)
        mcaux = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)
        for ji in range(0, self.kjpindex, 1):
            if self.gmhv.vegtot_old[ji] > self.gmhv.min_sechiba:
                for jv in range(0, self.nvm, 1):
                    if self.veget_max[ji, jv] < self.gmhv.min_sechiba and self.qsintveg[ji, jv] > 0.0:
                        jst = self.gmhv.pref_soil_veg[jv] - 1
                        if self.gmhv.resdist[ji, jst] > self.gmhv.zero:
                            index = jst
                        else:
                            index = np.argmax(self.gmhv.resdist[ji, :], axis=0) + 1
                            if self.gmhv.resdist[ji, index] <= self.gmhv.zero:
                                logging.info(f'ipts, index, resdist,{ji},{index},{self.gmhv.resdist[ji, :]}')
                                logging.error('Exception: hydrol_tmc_update if all resdist - see above- are zero the last vegetation may have been replaced by a non biological land cover This transfer has not yet been implemented in the code')
                        self.gmhv.water2infilt[ji, index] = self.gmhv.water2infilt[ji, index] + self.qsintveg[ji, jv] / (self.gmhv.resdist[ji, index] * self.gmhv.vegtot_old[ji])
                        self.qsintveg[ji, jv] = self.gmhv.zero
        vegtot_upd = np.sum(np.abs(self.gmhv.vegtot[:] - self.gmhv.vegtot_old[:])) > self.gmhv.zero
        self.runoff_upd[:] = self.gmhv.zero
        self.drain_upd[:] = self.gmhv.zero
        if vegtot_upd:
            delvegtot[:] = self.gmhv.vegtot[:] - self.gmhv.vegtot_old[:]
            for jst in range(0, self.nstm, 1):
                for ji in range(0, self.kjpindex, 1):
                    if delvegtot[ji] > self.gmhv.min_sechiba:
                        self.gmhv.mc[ji, :, jst] = self.gmhv.mc[ji, :, jst] * self.gmhv.vegtot_old[ji] / self.gmhv.vegtot[ji]
                        self.gmhv.water2infilt[ji, jst] = self.gmhv.water2infilt[ji, jst] * self.gmhv.vegtot_old[ji] / self.gmhv.vegtot[ji]
                    else:
                        if self.gmhv.vegtot[ji] > self.gmhv.min_sechiba:
                            mcaux[ji, :, jst] = self.gmhv.mc[ji, :, jst] * (self.gmhv.vegtot_old[ji] - self.gmhv.vegtot[ji]) / self.gmhv.vegtot[ji]
                        else:
                            mcaux[ji, :, jst] = self.gmhv.mc[ji, :, jst]
                        self.drain_upd[ji] = self.drain_upd[ji] + self.gmhv.dz[1] * (self.gmhv.trois * mcaux[ji, 0, jst] + mcaux[ji, 1, jst]) / self.gmhv.huit
                        for jsl in range(1, self.nslm - 1, 1):
                            self.drain_upd[ji] = self.drain_upd[ji] + self.gmhv.dz[jsl] * (self.gmhv.trois * mcaux[ji, jsl, jst] + mcaux[ji, jsl - 1, jst]) / self.gmhv.huit + self.gmhv.dz[jsl + 1] * (self.gmhv.trois * mcaux[ji, jsl, jst] + mcaux[ji, jsl + 1, jst]) / self.gmhv.huit
                        self.drain_upd[ji] = self.drain_upd[ji] + self.gmhv.dz[self.nslm - 1] * (self.gmhv.trois * mcaux[ji, self.nslm - 1, jst] + mcaux[ji, self.nslm - 2, jst]) / self.gmhv.huit
                        if self.gmhv.vegtot[ji] > self.gmhv.min_sechiba:
                            self.runoff_upd[ji] = self.runoff_upd[ji] + self.gmhv.water2infilt[ji, jst] * (self.gmhv.vegtot_old[ji] - self.gmhv.vegtot[ji]) / self.gmhv.vegtot[ji]
                        else:
                            self.runoff_upd[ji] = self.runoff_upd[ji] + self.gmhv.water2infilt[ji, jst]
        soil_upd = np.sum(np.abs(self.soiltile[:, :] - self.gmhv.resdist[:, :])) > self.gmhv.zero
        if self.gmhv.printlev >= 3:
            logging.info(f'soil_upd{soil_upd}')
        if soil_upd:
            vmr[:, :] = self.soiltile[:, :] - self.gmhv.resdist[:, :]
            for ji in range(0, self.kjpindex, 1):
                vmr_sum[ji] = np.sum(vmr[ji, :], MASK=vmr[ji, :] < self.gmhv.zero)
            mc_dilu[:, :] = self.gmhv.zero
            for jst in range(0, self.nstm, 1):
                for jsl in range(0, self.nslm, 1):
                    for ji in range(0, self.kjpindex, 1):
                        if vmr[ji, jst] < -self.gmhv.min_sechiba:
                            mc_dilu[ji, jsl] = mc_dilu[ji, jsl] + self.gmhv.mc[ji, jsl, jst] * vmr[ji, jst] / vmr_sum[ji]
            infil_dilu[:] = self.gmhv.zero
            for jst in range(0, self.nstm, 1):
                for ji in range(0, self.kjpindex, 1):
                    if vmr[ji, jst] < -self.gmhv.min_sechiba:
                        infil_dilu[ji] = infil_dilu[ji] + self.gmhv.water2infilt[ji, jst] * vmr[ji, jst] / vmr_sum[ji]
            for jst in range(0, self.nstm, 1):
                for jsl in range(0, self.nslm, 1):
                    for ji in range(0, self.kjpindex, 1):
                        if vmr[ji, jst] > self.gmhv.min_sechiba:
                            self.gmhv.mc[ji, jsl, jst] = (self.gmhv.mc[ji, jsl, jst] * self.gmhv.resdist[ji, jst] + mc_dilu[ji, jsl] * vmr[ji, jst]) / self.soiltile[ji, jst]
            for jst in range(0, self.nstm, 1):
                for ji in range(0, self.kjpindex, 1):
                    if vmr[ji, jst] > self.gmhv.min_sechiba:
                        self.gmhv.water2infilt[ji, jst] = (self.gmhv.water2infilt[ji, jst] * self.gmhv.resdist[ji, jst] + infil_dilu[ji] * vmr[ji, jst]) / self.soiltile[ji, jst]
            for jst in range(0, self.nstm, 1):
                for ji in range(0, self.kjpindex, 1):
                    if self.soiltile[ji, jst] < self.gmhv.min_sechiba:
                        self.gmhv.water2infilt[ji, jst] = self.gmhv.zero
                        self.gmhv.mc[ji, :, jst] = self.gmhv.zero
        for jst in range(0, self.nstm, 1):
            for ji in range(0, self.kjpindex, 1):
                self.gmhv.tmc[ji, jst] = self.gmhv.dz[1] * (self.gmhv.trois * self.gmhv.mc[ji, 0, jst] + self.gmhv.mc[ji, 1, jst]) / self.gmhv.huit
                for jsl in range(1, self.nslm - 1, 1):
                    self.gmhv.tmc[ji, jst] = self.gmhv.tmc[ji, jst] + self.gmhv.dz[jsl] * (self.gmhv.trois * self.gmhv.mc[ji, jsl, jst] + self.gmhv.mc[ji, jsl - 1, jst]) / self.gmhv.huit + self.gmhv.dz[jsl + 1] * (self.gmhv.trois * self.gmhv.mc[ji, jsl, jst] + self.gmhv.mc[ji, jsl + 1, jst]) / self.gmhv.huit
                self.gmhv.tmc[ji, jst] = self.gmhv.tmc[ji, jst] + self.gmhv.dz[self.nslm - 1] * (self.gmhv.trois * self.gmhv.mc[ji, self.nslm - 1, jst] + self.gmhv.mc[ji, self.nslm - 2, jst]) / self.gmhv.huit
                self.gmhv.tmc[ji, jst] = self.gmhv.tmc[ji, jst] + self.gmhv.water2infilt[ji, jst]
        self.gmhv.humtot[:] = self.gmhv.zero
        for jst in range(0, self.nstm, 1):
            for ji in range(0, self.kjpindex, 1):
                self.gmhv.humtot[ji] = self.gmhv.humtot[ji] + self.gmhv.vegtot[ji] * self.soiltile[ji, jst] * self.gmhv.tmc[ji, jst]
        self.gmhv.resdist[:, :] = self.soiltile[:, :]
        if self.gmhv.printlev >= 3:
            logging.info('hydrol_tmc_update done')

    def hydrol_vegupd(self):
        self.hydrol_tmc_update()
        self.gmhv.mask_veget[:, :] = 0
        self.gmhv.mask_soiltile[:, :] = 0
        for jst in range(0, self.nstm, 1):
            for ji in range(0, self.kjpindex, 1):
                if self.soiltile[ji, jst] > self.gmhv.min_sechiba:
                    self.gmhv.mask_soiltile[ji, jst] = 1
        for jv in range(0, self.nvm, 1):
            for ji in range(0, self.kjpindex, 1):
                if self.veget_max[ji, jv] > self.gmhv.min_sechiba:
                    self.gmhv.mask_veget[ji, jv] = 1
        self.gmhv.vegetmax_soil[:, :, :] = self.gmhv.zero
        for jv in range(0, self.nvm, 1):
            jst = self.gmhv.pref_soil_veg[jv] - 1
            for ji in range(0, self.kjpindex, 1):
                if self.gmhv.mask_soiltile[ji, jst] > 0 and self.gmhv.vegtot[ji] > self.gmhv.min_sechiba:
                    self.gmhv.vegetmax_soil[ji, jv, jst] = self.veget_max[ji, jv] / self.soiltile[ji, jst]
        for ji in range(0, self.kjpindex, 1):
            if self.veget_max[ji, 0] > self.gmhv.min_sechiba:
                self.frac_bare[ji, 0] = self.gmhv.un
            else:
                self.frac_bare[ji, 0] = self.gmhv.zero
        if self.gmhv.ok_bare_soil_new:
            self.frac_bare[:, 1:self.nvm] = self.gmhv.zero
        else:
            for jv in range(1, self.nvm, 1):
                for ji in range(0, self.kjpindex, 1):
                    if self.veget_max[ji, jv] > self.gmhv.min_sechiba:
                        self.frac_bare[ji, jv] = self.gmhv.un - self.veget[ji, jv] / self.veget_max[ji, jv]
                    else:
                        self.frac_bare[ji, jv] = self.gmhv.zero
        self.gmhv.frac_bare_ns[:, :] = self.gmhv.zero
        for jst in range(0, self.nstm, 1):
            for jv in range(0, self.nvm, 1):
                for ji in range(0, self.kjpindex, 1):
                    if self.gmhv.vegtot[ji] > self.gmhv.min_sechiba:
                        self.gmhv.frac_bare_ns[ji, jst] = self.gmhv.frac_bare_ns[ji, jst] + self.gmhv.vegetmax_soil[ji, jv, jst] * self.frac_bare[ji, jv] / self.gmhv.vegtot[ji]
        if self.gmhv.printlev >= 3:
            logging.info('hydrol_vegupd done')

    def read_dummy(self):
        print(f'--- inside the read dummy routine for hydrol_vegupd ---')
        path = f'/home/ssivanes/Fgpt/benchmark/hydrol_vegupd/dummy.bin'
        ffile = FortranFile(path, 'r')
        for var in [self.frac_bare, self.qsintveg, self.soiltile, self.veget, self.veget_max]:
            if isinstance(var, np.ndarray):
                arr_shape = var.shape
                if var.dtype == np.float64:
                    data = ffile.read_reals(np.float64)
                elif var.dtype == np.int32 or var.dtype == np.bool:
                    data = ffile.read_ints(np.int32)
                if data.size != np.prod(arr_shape):
                    continue
                var[:] = data.reshape(arr_shape, order='F')

    def test_hydrol_vegupd(self):
        print('--- inside the test function for hydrol_vegupd ---')
        path = f'/home/ssivanes/Fgpt/benchmark/hydrol_vegupd/output.bin'
        ffile = FortranFile(path, 'r')
        modif_var = ['frac_bare', 'frac_bare_ns', 'mask_soiltile', 'mask_veget', 'vegetmax_soil']
        for (variable, value) in zip(modif_var, [self.frac_bare, self.gmhv.frac_bare_ns, self.gmhv.mask_soiltile, self.gmhv.mask_veget, self.gmhv.vegetmax_soil]):
            fortran_value = None
            if isinstance(value, np.ndarray):
                arr_shape = value.shape
                if value.dtype == np.float64:
                    fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
                elif value.dtype == np.int32 or value.dtype == np.bool:
                    fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
                if np.allclose(value, fortran_value):
                    print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
                else:
                    print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                    print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                    print(f' Min and max of python value: {value.min()}, {value.max()}')
                    print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
            elif not isinstance(value, np.ndarray):
                if value.dtype == np.bool or value.dtype == np.int32:
                    fortran_value = ffile.read_ints(np.int32)[-1]
                elif value.dtype == np.float64:
                    fortran_value = ffile.read_reals(np.float64)[-1]
                if np.isclose(value, fortran_value):
                    print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
                else:
                    print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    hv = Hydrol_vegupd()
    hv.read_dummy()
    hv.hydrol_vegupd()
    hv.test_hydrol_vegupd()
if __name__ == '__main__':
    main()
```

```python
import numpy as np
import os
from scipy.io import FortranFile

class Global_module_hydrol_soil:

    def __init__(self):
        self.nsnow = np.int32(3)
        self.nslm = np.int32(11)
        self.nvm = np.int32(15)
        self.nstm = np.int32(3)
        self.kjpindex = np.int32(4717)
        self.ier = np.int32(0)
        self.ic0 = np.int32(0)
        self.ic = np.int32(0)
        self.icr = np.float64(0.0)
        self.start_time = np.float64(0.0)
        self.stop_time = np.float64(0.0)
        self.zero = np.float64(0.0)
        self.imin = np.int32(1)
        self.lhf = np.float64(0.3336 * 1000000.0)
        self.min_sechiba = np.float64(1e-08)
        self.ZeroCelsius = np.float64(273.15)
        self.trois = np.float64(3.0)
        self.huit = np.float64(8.0)
        self.numout = np.int32(6)
        self.un = np.float64(1.0)
        self.nnobio = np.int32(1)
        self.iice = np.int32(1)
        self.deux = np.float64(2.0)
        self.mille = np.float64(1000.0)
        self.iinterface = np.int32(2)
        self.maxaltmax = np.float64(2.0)
        self.plev = np.int32(0)
        self.ifunc = np.int32(2)
        self.istruc = np.int32(1)
        self.nroot_prof = np.int32(2)
        self.ndepths = np.int32(2)
        self.err_act = np.int32(1)
        self.inode = np.int32(1)
        self.w_time = np.float64(1.0)
        self.kilo_to_unit = np.float64(1000.0)
        self.is_tuzet_hydrol_arch = np.bool(False)
        self.allowed_err = np.float64(2e-08)
        self.iroot = np.int32(6)
        self.irrig_st = np.int32(3)
        self.min_wind = np.float64(0.1)
        self.undef_sechiba = np.float64(1e+20)
        self.nelements = np.int32(2)
        self.doponds = np.bool(False)
        self.ks_usda = np.array([7128.0, 3501.6, 1060.8, 108.0, 60.0, 249.6, 314.4, 16.8, 62.4, 28.8, 4.8, 48.0, 6131.4], dtype=np.float64)
        self.new_watstress = np.bool(False)
        self.beta_irrig = np.float64(0.9)
        self.alpha_watstress = np.float64(1.0)
        self.quatre = np.float64(4.0)
        self.old_irrig_scheme = np.bool(False)
        self.min_stomate = np.float64(1e-08)
        self.nparts = np.int32(9)
        self.printlev = np.int32(2)
        self.icarbon = np.int32(1)
        self.ncirc = np.int32(1)
        self.xios_default_val = np.float64(0.0)
        self.lai_irrig_min = np.float64(0.1)
        self.irrigated_soiltile = np.bool(False)
        self.do_rsoil = np.bool(True)
        self.nscm_usda = np.int32(13)
        self.nscm = self.nscm_usda
        self.alpha_nudge_mc = np.float64(0)
        self.check_cwrr = np.bool(False)
        self.dt_sechiba = np.float64(0)
        self.fr_center = np.float64(0)
        self.fr_dT = np.float64(0)
        self.froz_frac_corr = np.float64(0)
        self.imax = np.int32(0)
        self.kfact_root_const = np.bool(False)
        self.max_froz_hydro = np.float64(0)
        self.ok_freeze_cwrr = np.bool(False)
        self.ok_hydrol_arch = np.bool(False)
        self.ok_nudge_mc = np.bool(False)
        self.ok_thermodynamical_freezing = np.bool(False)
        self.one_day = np.float64(0)
        self.smtot_corr = np.float64(0)
        self.zmaxh = np.float64(0)
        self.a = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.ae_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.avan_mod_tab = np.zeros((self.nslm, self.kjpindex), dtype=np.float64)
        self.b = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.check_top_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.d = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.dh = np.zeros((self.nslm,), dtype=np.float64)
        self.dr_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.dz = np.zeros((self.nslm,), dtype=np.float64)
        self.e = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.ep = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.ext_coeff_vegetfrac = np.zeros((self.nvm,), dtype=np.float64)
        self.f = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.fp = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.frac_bare_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.free_drain_coef = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.g1 = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.gp = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.humcste = np.zeros((self.nvm,), dtype=np.float64)
        self.humrelv = np.zeros((self.kjpindex, self.nvm, self.nstm), dtype=np.float64)
        self.humtot = np.zeros((self.kjpindex,), dtype=np.float64)
        self.k = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.kfact = np.zeros((self.nslm, self.kjpindex), dtype=np.float64)
        self.kfact_root = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)
        self.kk = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)
        self.kk_moy = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.mask_soiltile = np.zeros((self.kjpindex, self.nstm), dtype=np.int32)
        self.max_root_depth = np.zeros((self.nvm,), dtype=np.float64)
        self.mc = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)
        self.mc_read_current = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)
        self.mcl = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)
        self.natural = np.zeros((self.nvm,), dtype=np.bool)
        self.nslm_root = np.zeros((self.kjpindex,), dtype=np.int32)
        self.nvan_mod_tab = np.zeros((self.nslm, self.kjpindex), dtype=np.float64)
        self.pcent = np.zeros((self.nscm,), dtype=np.float64)
        self.precisol = np.zeros((self.kjpindex, self.nvm), dtype=np.float64)
        self.precisol_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.pref_soil_veg = np.zeros((self.nvm,), dtype=np.int32)
        self.profil_froz_hydro = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.profil_froz_hydro_ns = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)
        self.qflux_ns = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)
        self.resolv = np.zeros((self.kjpindex,), dtype=np.bool)
        self.rhs = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.rootsink = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)
        self.ru_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.soil_wet_litter = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.soil_wet_ns = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)
        self.soilmoist = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.soilmoist_liquid = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.soilmoist_s = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)
        self.srhs = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.stmat = np.zeros((self.kjpindex, self.nslm, 3), dtype=np.float64)
        self.subsinksoil = np.zeros((self.kjpindex,), dtype=np.float64)
        self.tmat = np.zeros((self.kjpindex, self.nslm, 3), dtype=np.float64)
        self.tmc = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.tmc_aux = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.tmc_litt_dry_mea = np.zeros((self.kjpindex,), dtype=np.float64)
        self.tmc_litt_mea = np.zeros((self.kjpindex,), dtype=np.float64)
        self.tmc_litt_wet_mea = np.zeros((self.kjpindex,), dtype=np.float64)
        self.tmc_litter = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.tmc_litter_adry = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.tmc_litter_awet = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.tmc_litter_field = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.tmc_litter_res = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.tmc_litter_sat = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.tmc_litter_wilt = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.tr_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.undermcr = np.zeros((self.kjpindex,), dtype=np.float64)
        self.vegetmax_soil = np.zeros((self.kjpindex, self.nvm, self.nstm), dtype=np.float64)
        self.vegstressv = np.zeros((self.kjpindex, self.nvm, self.nstm), dtype=np.float64)
        self.vegtot = np.zeros((self.kjpindex,), dtype=np.float64)
        self.vegtot_old = np.zeros((self.kjpindex,), dtype=np.float64)
        self.water2infilt = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.zdr = np.zeros((self.nslm - 0 + 1,), dtype=np.float64)
        self.znh = np.zeros((self.nslm,), dtype=np.float64)
        self.zwt_force = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.zz = np.zeros((self.nslm,), dtype=np.float64)

    def declaration_initialization(self):
        print(f'--- add the declaration and initialization in module global ---')
        path = f'/home/ssivanes/Fgpt/benchmark/hydrol_soil/global.bin'
        ffile = FortranFile(path, 'r')
        self.alpha_nudge_mc = ffile.read_reals(np.float64)[0]
        self.check_cwrr = np.bool(ffile.read_ints(np.int32)[0])
        self.dt_sechiba = ffile.read_reals(np.float64)[0]
        self.fr_center = ffile.read_reals(np.float64)[0]
        self.fr_dT = ffile.read_reals(np.float64)[0]
        self.froz_frac_corr = ffile.read_reals(np.float64)[0]
        self.imax = ffile.read_ints(np.int32)[0]
        self.b_lin = np.zeros((self.imax - self.imin + 1, self.nslm, self.kjpindex), dtype=np.float64)
        self.d_lin = np.zeros((self.imax - self.imin + 1, self.nslm, self.kjpindex), dtype=np.float64)
        self.a_lin = np.zeros((self.imax - self.imin + 1, self.nslm, self.kjpindex), dtype=np.float64)
        self.k_lin = np.zeros((self.imax - self.imin + 1, self.nslm, self.kjpindex), dtype=np.float64)
        self.kfact_root_const = np.bool(ffile.read_ints(np.int32)[0])
        self.max_froz_hydro = ffile.read_reals(np.float64)[0]
        self.ok_freeze_cwrr = np.bool(ffile.read_ints(np.int32)[0])
        self.ok_hydrol_arch = np.bool(ffile.read_ints(np.int32)[0])
        self.ok_nudge_mc = np.bool(ffile.read_ints(np.int32)[0])
        self.ok_thermodynamical_freezing = np.bool(ffile.read_ints(np.int32)[0])
        self.one_day = ffile.read_reals(np.float64)[0]
        self.smtot_corr = ffile.read_reals(np.float64)[0]
        self.zmaxh = ffile.read_reals(np.float64)[0]
        for attr_name in ['a', 'a_lin', 'ae_ns', 'avan_mod_tab', 'b', 'b_lin', 'check_top_ns', 'd', 'd_lin', 'dh', 'dr_ns', 'dz', 'e', 'ep', 'ext_coeff_vegetfrac', 'f', 'fp', 'frac_bare_ns', 'free_drain_coef', 'g1', 'gp', 'humcste', 'humrelv', 'humtot', 'k', 'k_lin', 'kfact', 'kfact_root', 'kk', 'kk_moy', 'mask_soiltile', 'max_root_depth', 'mc', 'mc_read_current', 'mcl', 'natural', 'nslm_root', 'nvan_mod_tab', 'pcent', 'precisol', 'precisol_ns', 'pref_soil_veg', 'profil_froz_hydro', 'profil_froz_hydro_ns', 'qflux_ns', 'resolv', 'rhs', 'rootsink', 'ru_ns', 'soil_wet_litter', 'soil_wet_ns', 'soilmoist', 'soilmoist_liquid', 'soilmoist_s', 'srhs', 'stmat', 'subsinksoil', 'tmat', 'tmc', 'tmc_aux', 'tmc_litt_dry_mea', 'tmc_litt_mea', 'tmc_litt_wet_mea', 'tmc_litter', 'tmc_litter_adry', 'tmc_litter_awet', 'tmc_litter_field', 'tmc_litter_res', 'tmc_litter_sat', 'tmc_litter_wilt', 'tr_ns', 'undermcr', 'vegetmax_soil', 'vegstressv', 'vegtot', 'vegtot_old', 'water2infilt', 'zdr', 'znh', 'zwt_force', 'zz']:
            attribute = getattr(self, attr_name)
            if isinstance(attribute, np.ndarray):
                arr_shape = attribute.shape
                if attribute.dtype == np.float64:
                    data = ffile.read_reals(np.float64)
                elif attribute.dtype == np.int32 or attribute.dtype == np.bool:
                    data = ffile.read_ints(np.int32)
                if data.size != np.prod(arr_shape):
                    continue
                value = data.reshape(arr_shape, order='F')
                setattr(self, attr_name, value)
```

```python
import os
import time
import functools
import numpy as np
import logging
from scipy.io import FortranFile
# from module_global import Global_module_hydrol_soil

class Hydrol_soil:

    def __init__(self):
        self.nsnow = np.int32(3)
        self.nslm = np.int32(11)
        self.nvm = np.int32(15)
        self.nstm = np.int32(3)
        self.kjpindex = np.int32(4717)
        self.ier = np.int32(0)
        self.ic0 = np.int32(0)
        self.ic = np.int32(0)
        self.icr = np.float64(0.0)
        self.start_time = np.float64(0.0)
        self.stop_time = np.float64(0.0)
        self.gmhs = Global_module_hydrol_soil()
        self.gmhs.declaration_initialization()
        self.altmax = np.zeros((self.kjpindex, self.nvm), dtype=np.float64)
        self.avan = np.zeros((self.kjpindex,), dtype=np.float64)
        self.circ_class_biomass = np.zeros((self.kjpindex, self.nvm, self.gmhs.ncirc, self.gmhs.nparts, self.gmhs.nelements), dtype=np.float64)
        self.drainage = np.zeros((self.kjpindex,), dtype=np.float64)
        self.drysoil_frac = np.zeros((self.kjpindex,), dtype=np.float64)
        self.e_frac = np.zeros((self.kjpindex, self.nvm, self.nslm, self.nstm), dtype=np.float64)
        self.evap_bare_lim = np.zeros((self.kjpindex,), dtype=np.float64)
        self.evap_bare_lim_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.evapot = np.zeros((self.kjpindex,), dtype=np.float64)
        self.evapot_penm = np.zeros((self.kjpindex,), dtype=np.float64)
        self.F_absorption = np.zeros((self.kjpindex, self.nvm), dtype=np.float64)
        self.frac_snow_nobio = np.zeros((self.kjpindex, self.gmhs.nnobio), dtype=np.float64)
        self.humrel = np.zeros((self.kjpindex, self.nvm), dtype=np.float64)
        self.irrigation = np.zeros((self.kjpindex,), dtype=np.float64)
        self.k_litt = np.zeros((self.kjpindex,), dtype=np.float64)
        self.ks = np.zeros((self.kjpindex,), dtype=np.float64)
        self.ksoil = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)
        self.litterhumdiag = np.zeros((self.kjpindex,), dtype=np.float64)
        self.mc_layh = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.mc_layh_s = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)
        self.mcfc = np.zeros((self.kjpindex,), dtype=np.float64)
        self.mcl_layh = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.mcl_layh_s = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)
        self.mcr = np.zeros((self.kjpindex,), dtype=np.float64)
        self.mcs = np.zeros((self.kjpindex,), dtype=np.float64)
        self.mcw = np.zeros((self.kjpindex,), dtype=np.float64)
        self.njsc = np.zeros((self.kjpindex,), dtype=np.int32)
        self.nvan = np.zeros((self.kjpindex,), dtype=np.float64)
        self.precip_rain = np.zeros((self.kjpindex,), dtype=np.float64)
        self.reinf_slope_soil = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.reinfiltration = np.zeros((self.kjpindex,), dtype=np.float64)
        self.returnflow = np.zeros((self.kjpindex,), dtype=np.float64)
        self.root_deficit = np.zeros((self.kjpindex,), dtype=np.float64)
        self.root_depth = np.zeros((self.kjpindex, self.nvm, self.gmhs.ndepths), dtype=np.float64)
        self.root_profile = np.zeros((self.kjpindex, self.nvm, self.nslm, self.gmhs.nroot_prof), dtype=np.float64)
        self.runoff = np.zeros((self.kjpindex,), dtype=np.float64)
        self.shumdiag = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.shumdiag_perma = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.snow = np.zeros((self.kjpindex,), dtype=np.float64)
        self.snowdz = np.zeros((self.kjpindex, self.nsnow), dtype=np.float64)
        self.soiltile = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.stempdiag = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        self.tot_bare_soil = np.zeros((self.kjpindex,), dtype=np.float64)
        self.tot_melt = np.zeros((self.kjpindex,), dtype=np.float64)
        self.totfrac_nobio = np.zeros((self.kjpindex,), dtype=np.float64)
        self.tq_cdrag = np.zeros((self.kjpindex,), dtype=np.float64)
        self.transpir = np.zeros((self.kjpindex, self.nvm), dtype=np.float64)
        self.u = np.zeros((self.kjpindex,), dtype=np.float64)
        self.us = np.zeros((self.kjpindex, self.nvm, self.nstm, self.nslm), dtype=np.float64)
        self.v = np.zeros((self.kjpindex,), dtype=np.float64)
        self.veget = np.zeros((self.kjpindex, self.nvm), dtype=np.float64)
        self.veget_max = np.zeros((self.kjpindex, self.nvm), dtype=np.float64)
        self.vegstress = np.zeros((self.kjpindex, self.nvm), dtype=np.float64)
        self.vevapnu = np.zeros((self.kjpindex,), dtype=np.float64)

    def hydrol_diag_soil_flux(self, ins, mclint, flux_top):
        check_temp = np.zeros((self.kjpindex,), dtype=np.float64)
        for ji in range(0, self.kjpindex, 1):
            jsl = self.nslm - 1
            self.gmhs.qflux_ns[ji, jsl, ins] = self.gmhs.dr_ns[ji, ins]
            jsl = self.nslm - 2
            self.gmhs.qflux_ns[ji, jsl, ins] = self.gmhs.qflux_ns[ji, jsl + 1, ins] + (self.gmhs.mcl[ji, jsl, ins] - mclint[ji, jsl] + self.gmhs.trois * self.gmhs.mcl[ji, jsl + 1, ins] - self.gmhs.trois * mclint[ji, jsl + 1]) * (self.gmhs.dz[jsl + 1] / self.gmhs.huit) + self.gmhs.rootsink[ji, jsl + 1, ins]
        for jsl in range(self.nslm - 3, -1, -1):
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.qflux_ns[ji, jsl, ins] = self.gmhs.qflux_ns[ji, jsl + 1, ins] + (self.gmhs.mcl[ji, jsl, ins] - mclint[ji, jsl] + self.gmhs.trois * self.gmhs.mcl[ji, jsl + 1, ins] - self.gmhs.trois * mclint[ji, jsl + 1]) * (self.gmhs.dz[jsl + 1] / self.gmhs.huit) + self.gmhs.rootsink[ji, jsl + 1, ins] + self.gmhs.dz[jsl + 2] / self.gmhs.huit * (self.gmhs.trois * self.gmhs.mcl[ji, jsl + 1, ins] - self.gmhs.trois * mclint[ji, jsl + 1] + self.gmhs.mcl[ji, jsl + 2, ins] - mclint[ji, jsl + 2])
        for ji in range(0, self.kjpindex, 1):
            check_temp[ji] = self.gmhs.qflux_ns[ji, 0, ins] + self.gmhs.dz[1] / self.gmhs.huit * (self.gmhs.trois * (self.gmhs.mcl[ji, 0, ins] - mclint[ji, 0]) + (self.gmhs.mcl[ji, 1, ins] - mclint[ji, 1])) + self.gmhs.rootsink[ji, 0, ins]
            self.gmhs.check_top_ns[ji, ins] = flux_top[ji] + check_temp[ji]
            if np.abs(self.gmhs.check_top_ns[ji, ins]) / self.gmhs.dt_sechiba > self.gmhs.min_sechiba:
                logging.info(f'Problem in the water balance, qflux_ns computation, surface fluxes{flux_top[ji]},{check_temp[ji]}')
                logging.info(f'Diagnosed and imposed fluxes differ by more than 1.e-8 mm/s:{self.gmhs.check_top_ns[ji, ins]}')
                logging.info(f'ji{ji},jsl{jsl},ins{ins}')
                logging.info(f'mclint{mclint[ji, :]}')
                logging.info(f'mcl{self.gmhs.mcl[ji, :, ins]}')
                logging.info(f'rootsink{self.gmhs.rootsink[ji, 0, ins]}')
                logging.error('Exception: hydrol_diag_soil_flux NOTE: Problem in the water balance, qflux_ns computation ')

    def hydrol_soil_tridiag(self, ins):
        bet = np.zeros((self.kjpindex,), dtype=np.float64)
        gam = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        for ji in range(0, self.kjpindex, 1):
            if self.gmhs.resolv[ji]:
                bet[ji] = self.gmhs.tmat[ji, 0, 1]
                self.gmhs.mcl[ji, 0, ins] = self.gmhs.rhs[ji, 0] / bet[ji]
        for jsl in range(1, self.nslm, 1):
            for ji in range(0, self.kjpindex, 1):
                if self.gmhs.resolv[ji]:
                    gam[ji, jsl] = self.gmhs.tmat[ji, jsl - 1, 2] / bet[ji]
                    bet[ji] = self.gmhs.tmat[ji, jsl, 1] - self.gmhs.tmat[ji, jsl, 0] * gam[ji, jsl]
                    self.gmhs.mcl[ji, jsl, ins] = (self.gmhs.rhs[ji, jsl] - self.gmhs.tmat[ji, jsl, 0] * self.gmhs.mcl[ji, jsl - 1, ins]) / bet[ji]
        for ji in range(0, self.kjpindex, 1):
            if self.gmhs.resolv[ji]:
                for jsl in range(self.nslm - 2, -1, -1):
                    self.gmhs.mcl[ji, jsl, ins] = self.gmhs.mcl[ji, jsl, ins] - gam[ji, jsl + 1] * self.gmhs.mcl[ji, jsl + 1, ins]

    def hydrol_diag_soil(self):
        mask_vegtot = np.zeros((self.kjpindex,), dtype=np.float64)
        for jst in range(0, self.nstm, 1):
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.ae_ns[ji, jst] = self.gmhs.ae_ns[ji, jst] * self.gmhs.mask_soiltile[ji, jst]
                self.gmhs.dr_ns[ji, jst] = self.gmhs.dr_ns[ji, jst] * self.gmhs.mask_soiltile[ji, jst]
                self.gmhs.ru_ns[ji, jst] = self.gmhs.ru_ns[ji, jst] * self.gmhs.mask_soiltile[ji, jst]
                self.gmhs.tmc[ji, jst] = self.gmhs.tmc[ji, jst] * self.gmhs.mask_soiltile[ji, jst]
                for jv in range(0, self.nvm, 1):
                    self.gmhs.humrelv[ji, jv, jst] = self.gmhs.humrelv[ji, jv, jst] * self.gmhs.mask_soiltile[ji, jst]
                    for jsl in range(0, self.nslm, 1):
                        self.us[ji, jv, jst, jsl] = self.us[ji, jv, jst, jsl] * self.gmhs.mask_soiltile[ji, jst]
                for jsl in range(0, self.nslm, 1):
                    self.gmhs.mc[ji, jsl, jst] = self.gmhs.mc[ji, jsl, jst] * self.gmhs.mask_soiltile[ji, jst]
        self.runoff[:] = self.gmhs.zero
        self.drainage[:] = self.gmhs.zero
        self.gmhs.humtot[:] = self.gmhs.zero
        self.shumdiag[:, :] = self.gmhs.zero
        self.shumdiag_perma[:, :] = self.gmhs.zero
        self.k_litt[:] = self.gmhs.zero
        self.litterhumdiag[:] = self.gmhs.zero
        self.gmhs.tmc_litt_dry_mea[:] = self.gmhs.zero
        self.gmhs.tmc_litt_wet_mea[:] = self.gmhs.zero
        self.gmhs.tmc_litt_mea[:] = self.gmhs.zero
        self.humrel[:, :] = self.gmhs.zero
        self.vegstress[:, :] = self.gmhs.zero
        if self.gmhs.ok_freeze_cwrr:
            self.gmhs.profil_froz_hydro[:, :] = self.gmhs.zero
        for ji in range(0, self.kjpindex, 1):
            mask_vegtot[ji] = 0
            if self.gmhs.vegtot[ji] > self.gmhs.min_sechiba:
                mask_vegtot[ji] = 1
        for ji in range(0, self.kjpindex, 1):
            self.gmhs.ae_ns[ji, :] = mask_vegtot[ji] * self.gmhs.ae_ns[ji, :] * self.gmhs.frac_bare_ns[ji, :]
        for jst in range(0, self.nstm, 1):
            for ji in range(0, self.kjpindex, 1):
                self.drainage[ji] = mask_vegtot[ji] * (self.drainage[ji] + self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * self.gmhs.dr_ns[ji, jst])
                self.runoff[ji] = mask_vegtot[ji] * (self.runoff[ji] + self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * self.gmhs.ru_ns[ji, jst]) + (1 - mask_vegtot[ji]) * (self.tot_melt[ji] + self.irrigation[ji] + self.returnflow[ji] + self.reinfiltration[ji])
                self.gmhs.humtot[ji] = mask_vegtot[ji] * (self.gmhs.humtot[ji] + self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * self.gmhs.tmc[ji, jst])
                if self.gmhs.ok_freeze_cwrr:
                    self.gmhs.profil_froz_hydro[ji, :] = mask_vegtot[ji] * (self.gmhs.profil_froz_hydro[ji, :] + self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * self.gmhs.profil_froz_hydro_ns[ji, :, jst])
        for ji in range(0, self.kjpindex, 1):
            if self.gmhs.vegtot[ji] != 0.0:
                self.vevapnu[ji] = self.vevapnu[ji] + self.gmhs.subsinksoil[ji] * self.gmhs.vegtot[ji]
            else:
                self.vevapnu[ji] = self.vevapnu[ji] + self.gmhs.subsinksoil[ji]
            self.runoff[ji] = self.runoff[ji] + self.precip_rain[ji] * self.totfrac_nobio[ji] * (1 - self.frac_snow_nobio[ji, self.gmhs.iice - 1])
        for jst in range(0, self.nstm, 1):
            for jv in range(0, self.nvm, 1):
                for ji in range(0, self.kjpindex, 1):
                    if self.veget_max[ji, jv] > self.gmhs.min_sechiba:
                        self.vegstress[ji, jv] = self.vegstress[ji, jv] + self.gmhs.vegstressv[ji, jv, jst]
                        self.vegstress[ji, jv] = np.maximum(self.vegstress[ji, jv], self.gmhs.zero)
        for jst in range(0, self.nstm, 1):
            for jv in range(0, self.nvm, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.humrel[ji, jv] = self.humrel[ji, jv] + self.gmhs.humrelv[ji, jv, jst]
                    self.humrel[ji, jv] = np.maximum(self.humrel[ji, jv], self.gmhs.zero)
        for jst in range(0, self.nstm, 1):
            for ji in range(0, self.kjpindex, 1):
                if self.gmhs.tmc_litter[ji, jst] < self.gmhs.tmc_litter_res[ji, jst]:
                    i = self.gmhs.imin
                else:
                    tmc_litter_ratio = (self.gmhs.tmc_litter[ji, jst] - self.gmhs.tmc_litter_res[ji, jst]) / (self.gmhs.tmc_litter_sat[ji, jst] - self.gmhs.tmc_litter_res[ji, jst])
                    i = np.maximum(np.minimum(int((self.gmhs.imax - self.gmhs.imin) * tmc_litter_ratio) + self.gmhs.imin, self.gmhs.imax - 1), self.gmhs.imin)
                k_tmp = np.maximum(self.gmhs.k_lin[i - 1, 0, ji] * self.ks[ji], self.gmhs.zero)
                self.k_litt[ji] = self.k_litt[ji] + self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * np.sqrt(k_tmp)
            for ji in range(0, self.kjpindex, 1):
                self.litterhumdiag[ji] = self.litterhumdiag[ji] + self.gmhs.soil_wet_litter[ji, jst] * self.soiltile[ji, jst]
                self.gmhs.tmc_litt_wet_mea[ji] = self.gmhs.tmc_litt_wet_mea[ji] + self.gmhs.tmc_litter_awet[ji, jst] * self.soiltile[ji, jst]
                self.gmhs.tmc_litt_dry_mea[ji] = self.gmhs.tmc_litt_dry_mea[ji] + self.gmhs.tmc_litter_adry[ji, jst] * self.soiltile[ji, jst]
                self.gmhs.tmc_litt_mea[ji] = self.gmhs.tmc_litt_mea[ji] + self.gmhs.tmc_litter[ji, jst] * self.soiltile[ji, jst]
        for ji in range(0, self.kjpindex, 1):
            if self.gmhs.tmc_litt_wet_mea[ji] - self.gmhs.tmc_litt_dry_mea[ji] > self.gmhs.zero:
                self.drysoil_frac[ji] = self.gmhs.un + np.maximum(np.minimum((self.gmhs.tmc_litt_dry_mea[ji] - self.gmhs.tmc_litt_mea[ji]) / (self.gmhs.tmc_litt_wet_mea[ji] - self.gmhs.tmc_litt_dry_mea[ji]), self.gmhs.zero), -self.gmhs.un)
            else:
                self.drysoil_frac[ji] = self.gmhs.zero
        self.gmhs.soilmoist[:, :] = self.gmhs.zero
        for jst in range(0, self.nstm, 1):
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.soilmoist[ji, 0] = self.gmhs.soilmoist[ji, 0] + self.soiltile[ji, jst] * self.gmhs.dz[1] * (self.gmhs.trois * self.gmhs.mc[ji, 0, jst] + self.gmhs.mc[ji, 1, jst]) / self.gmhs.huit
                for jsl in range(1, self.nslm - 1, 1):
                    self.gmhs.soilmoist[ji, jsl] = self.gmhs.soilmoist[ji, jsl] + self.soiltile[ji, jst] * (self.gmhs.dz[jsl] * (self.gmhs.trois * self.gmhs.mc[ji, jsl, jst] + self.gmhs.mc[ji, jsl - 1, jst]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * self.gmhs.mc[ji, jsl, jst] + self.gmhs.mc[ji, jsl + 1, jst]) / self.gmhs.huit)
                self.gmhs.soilmoist[ji, self.nslm - 1] = self.gmhs.soilmoist[ji, self.nslm - 1] + self.soiltile[ji, jst] * self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * self.gmhs.mc[ji, self.nslm - 1, jst] + self.gmhs.mc[ji, self.nslm - 2, jst]) / self.gmhs.huit
        for ji in range(0, self.kjpindex, 1):
            self.gmhs.soilmoist[ji, :] = self.gmhs.soilmoist[ji, :] * self.gmhs.vegtot[ji]
        self.gmhs.soilmoist_s[:, :, :] = self.gmhs.zero
        for jst in range(0, self.nstm, 1):
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.soilmoist_s[ji, 0, self.nstm - 1] = self.gmhs.soilmoist_s[ji, 0, self.nstm - 1] + self.soiltile[ji, jst] * self.gmhs.dz[1] * (self.gmhs.trois * self.gmhs.mc[ji, 0, jst] + self.gmhs.mc[ji, 1, jst]) / self.gmhs.huit
                for jsl in range(1, self.nslm - 1, 1):
                    self.gmhs.soilmoist_s[ji, jsl, self.nstm - 1] = self.gmhs.soilmoist_s[ji, jsl, self.nstm - 1] + self.soiltile[ji, jst] * (self.gmhs.dz[jsl] * (self.gmhs.trois * self.gmhs.mc[ji, jsl, jst] + self.gmhs.mc[ji, jsl - 1, jst]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * self.gmhs.mc[ji, jsl, jst] + self.gmhs.mc[ji, jsl + 1, jst]) / self.gmhs.huit)
                self.gmhs.soilmoist_s[ji, self.nslm - 1, self.nstm - 1] = self.gmhs.soilmoist_s[ji, self.nslm - 1, self.nstm - 1] + self.soiltile[ji, jst] * self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * self.gmhs.mc[ji, self.nslm - 1, jst] + self.gmhs.mc[ji, self.nslm - 2, jst]) / self.gmhs.huit
        for ji in range(0, self.kjpindex, 1):
            self.gmhs.soilmoist_s[ji, :, :] = self.gmhs.soilmoist_s[ji, :, :] * self.gmhs.vegtot[ji]
        self.gmhs.soilmoist_liquid[:, :] = self.gmhs.zero
        for jst in range(0, self.nstm, 1):
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.soilmoist_liquid[ji, 0] = self.gmhs.soilmoist_liquid[ji, 0] + self.soiltile[ji, jst] * self.gmhs.dz[1] * (self.gmhs.trois * self.gmhs.mcl[ji, 0, jst] + self.gmhs.mcl[ji, 1, jst]) / self.gmhs.huit
                for jsl in range(1, self.nslm - 1, 1):
                    self.gmhs.soilmoist_liquid[ji, jsl] = self.gmhs.soilmoist_liquid[ji, jsl] + self.soiltile[ji, jst] * (self.gmhs.dz[jsl] * (self.gmhs.trois * self.gmhs.mcl[ji, jsl, jst] + self.gmhs.mcl[ji, jsl - 1, jst]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * self.gmhs.mcl[ji, jsl, jst] + self.gmhs.mcl[ji, jsl + 1, jst]) / self.gmhs.huit)
                self.gmhs.soilmoist_liquid[ji, self.nslm - 1] = self.gmhs.soilmoist_liquid[ji, self.nslm - 1] + self.soiltile[ji, jst] * self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * self.gmhs.mcl[ji, self.nslm - 1, jst] + self.gmhs.mcl[ji, self.nslm - 2, jst]) / self.gmhs.huit
        for ji in range(0, self.kjpindex, 1):
            self.gmhs.soilmoist_liquid[ji, :] = self.gmhs.soilmoist_liquid[ji, :] * self.gmhs.vegtot_old[ji]
        for jst in range(0, self.nstm, 1):
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.shumdiag[ji, jsl] = self.shumdiag[ji, jsl] + self.gmhs.soil_wet_ns[ji, jsl, jst] * self.soiltile[ji, jst] * ((self.mcs[ji] - self.mcw[ji]) / (self.mcfc[ji] - self.mcw[ji]))
                    self.shumdiag[ji, jsl] = np.maximum(np.minimum(self.shumdiag[ji, jsl], self.gmhs.un), self.gmhs.zero)
        for jsl in range(0, self.nslm, 1):
            for ji in range(0, self.kjpindex, 1):
                self.shumdiag_perma[ji, jsl] = self.gmhs.soilmoist[ji, jsl] / (self.gmhs.dh[jsl] * self.mcs[ji])
                self.shumdiag_perma[ji, jsl] = np.maximum(np.minimum(self.shumdiag_perma[ji, jsl], self.gmhs.un), self.gmhs.zero)

    def hydrol_soil_smooth_under_mcr(self, ins, is_under_mcr, check):
        excessji = np.zeros((self.kjpindex,), dtype=np.float64)
        tmci = np.zeros((self.kjpindex,), dtype=np.float64)
        tmcf = np.zeros((self.kjpindex,), dtype=np.float64)
        if self.gmhs.check_cwrr:
            tmci[:] = self.gmhs.dz[1] * (self.gmhs.trois * self.gmhs.mc[:, 0, ins] + self.gmhs.mc[:, 1, ins]) / self.gmhs.huit
            for jsl in range(1, self.nslm - 1, 1):
                tmci[:] = tmci[:] + self.gmhs.dz[jsl] * (self.gmhs.trois * self.gmhs.mc[:, jsl, ins] + self.gmhs.mc[:, jsl - 1, ins]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * self.gmhs.mc[:, jsl, ins] + self.gmhs.mc[:, jsl + 1, ins]) / self.gmhs.huit
            tmci[:] = tmci[:] + self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * self.gmhs.mc[:, self.nslm - 1, ins] + self.gmhs.mc[:, self.nslm - 2, ins]) / self.gmhs.huit
        for jsl in range(0, self.nslm - 2, 1):
            for ji in range(0, self.kjpindex, 1):
                excess = np.maximum(self.mcr[ji] - self.gmhs.mc[ji, jsl, ins], self.gmhs.zero)
                self.gmhs.mc[ji, jsl, ins] = self.gmhs.mc[ji, jsl, ins] + excess
                self.gmhs.mc[ji, jsl + 1, ins] = self.gmhs.mc[ji, jsl + 1, ins] - excess * (self.gmhs.dz[jsl] + self.gmhs.dz[jsl + 1]) / (self.gmhs.dz[jsl + 1] + self.gmhs.dz[jsl + 2])
        jsl = self.nslm - 2
        for ji in range(0, self.kjpindex, 1):
            excess = np.maximum(self.mcr[ji] - self.gmhs.mc[ji, jsl, ins], self.gmhs.zero)
            self.gmhs.mc[ji, jsl, ins] = self.gmhs.mc[ji, jsl, ins] + excess
            self.gmhs.mc[ji, jsl + 1, ins] = self.gmhs.mc[ji, jsl + 1, ins] - excess * (self.gmhs.dz[jsl] + self.gmhs.dz[jsl + 1]) / self.gmhs.dz[jsl + 1]
        jsl = self.nslm - 1
        for ji in range(0, self.kjpindex, 1):
            excess = np.maximum(self.mcr[ji] - self.gmhs.mc[ji, jsl, ins], self.gmhs.zero)
            self.gmhs.mc[ji, jsl, ins] = self.gmhs.mc[ji, jsl, ins] + excess
            self.gmhs.mc[ji, jsl - 1, ins] = self.gmhs.mc[ji, jsl - 1, ins] - excess * self.gmhs.dz[jsl] / (self.gmhs.dz[jsl - 1] + self.gmhs.dz[jsl])
        for jsl in range(self.nslm - 2, -2, -1):
            for ji in range(0, self.kjpindex, 1):
                excess = np.maximum(self.mcr[ji] - self.gmhs.mc[ji, jsl, ins], self.gmhs.zero)
                self.gmhs.mc[ji, jsl, ins] = self.gmhs.mc[ji, jsl, ins] + excess
                self.gmhs.mc[ji, jsl - 1, ins] = self.gmhs.mc[ji, jsl - 1, ins] - excess * (self.gmhs.dz[jsl] + self.gmhs.dz[jsl + 1]) / (self.gmhs.dz[jsl - 1] + self.gmhs.dz[jsl])
        for ji in range(0, self.kjpindex, 1):
            excessji[ji] = self.gmhs.mask_soiltile[ji, ins] * np.maximum(self.mcr[ji] - self.gmhs.mc[ji, 0, ins], self.gmhs.zero)
        for ji in range(0, self.kjpindex, 1):
            self.gmhs.mc[ji, 0, ins] = self.gmhs.mc[ji, 0, ins] + excessji[ji]
            is_under_mcr[ji, ins] = excessji[ji] > self.gmhs.min_sechiba
        for jsl in range(0, self.nslm, 1):
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.mc[ji, jsl, ins] = self.gmhs.mc[ji, jsl, ins] - excessji[ji] * self.gmhs.dz[1] / (self.gmhs.deux * self.gmhs.zmaxh * self.gmhs.mille)
        for jsl in range(0, self.nslm, 1):
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.mc[ji, jsl, ins] = self.gmhs.mask_soiltile[ji, ins] * self.gmhs.mc[ji, jsl, ins]
        if self.gmhs.check_cwrr:
            tmcf[:] = self.gmhs.dz[1] * (self.gmhs.trois * self.gmhs.mc[:, 0, ins] + self.gmhs.mc[:, 1, ins]) / self.gmhs.huit
            for jsl in range(1, self.nslm - 1, 1):
                tmcf[:] = tmcf[:] + self.gmhs.dz[jsl] * (self.gmhs.trois * self.gmhs.mc[:, jsl, ins] + self.gmhs.mc[:, jsl - 1, ins]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * self.gmhs.mc[:, jsl, ins] + self.gmhs.mc[:, jsl + 1, ins]) / self.gmhs.huit
            tmcf[:] = tmcf[:] + self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * self.gmhs.mc[:, self.nslm - 1, ins] + self.gmhs.mc[:, self.nslm - 2, ins]) / self.gmhs.huit
            check[:, ins] = tmcf[:] - tmci[:]

    def hydrol_split_soil(self):
        tmp_check1 = np.zeros((self.kjpindex,), dtype=np.float64)
        tmp_check2 = np.zeros((self.kjpindex,), dtype=np.float64)
        tmp_check3 = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        self.gmhs.precisol_ns[:, :] = self.gmhs.zero
        for jv in range(0, self.nvm, 1):
            for ji in range(0, self.kjpindex, 1):
                jst = self.gmhs.pref_soil_veg[jv] - 1
                if self.veget_max[ji, jv] > self.gmhs.min_sechiba and self.soiltile[ji, jst] * self.gmhs.vegtot[ji] > self.gmhs.min_sechiba:
                    self.gmhs.precisol_ns[ji, jst] = self.gmhs.precisol_ns[ji, jst] + self.gmhs.precisol[ji, jv] / (self.soiltile[ji, jst] * self.gmhs.vegtot[ji])
        self.gmhs.ae_ns[:, :] = self.gmhs.zero
        for jst in range(0, self.nstm, 1):
            for ji in range(0, self.kjpindex, 1):
                if self.evap_bare_lim[ji] > self.gmhs.min_sechiba:
                    self.gmhs.ae_ns[ji, jst] = self.vevapnu[ji] * self.evap_bare_lim_ns[ji, jst] / self.evap_bare_lim[ji]
        self.gmhs.tr_ns[:, :] = self.gmhs.zero
        for jv in range(0, self.nvm, 1):
            jst = self.gmhs.pref_soil_veg[jv] - 1
            for ji in range(0, self.kjpindex, 1):
                if self.humrel[ji, jv] > self.gmhs.min_sechiba and self.soiltile[ji, jst] * self.gmhs.vegtot[ji] > self.gmhs.min_sechiba:
                    self.gmhs.tr_ns[ji, jst] = self.gmhs.tr_ns[ji, jst] + self.transpir[ji, jv] * (self.gmhs.humrelv[ji, jv, jst] / self.humrel[ji, jv]) / (self.soiltile[ji, jst] * self.gmhs.vegtot[ji])
        self.gmhs.rootsink[:, :, :] = self.gmhs.zero
        if self.gmhs.ok_hydrol_arch:
            for jv in range(0, self.nvm, 1):
                jst = self.gmhs.pref_soil_veg[jv] - 1
                for jsl in range(0, self.nslm, 1):
                    for ji in range(0, self.kjpindex, 1):
                        if self.humrel[ji, jv] > self.gmhs.min_sechiba and self.soiltile[ji, jst] * self.gmhs.vegtot[ji] > self.gmhs.min_sechiba:
                            if self.gmhs.is_tuzet_hydrol_arch:
                                self.gmhs.rootsink[ji, jsl, jst] = self.gmhs.rootsink[ji, jsl, jst] + self.F_absorption[ji, jv] * self.e_frac[ji, jv, jsl, jst] * self.gmhs.dt_sechiba * self.gmhs.kilo_to_unit
                            else:
                                self.gmhs.rootsink[ji, jsl, jst] = self.gmhs.rootsink[ji, jsl, jst] + self.transpir[ji, jv] * self.e_frac[ji, jv, jsl, jst] / (self.soiltile[ji, jst] * self.gmhs.vegtot[ji])
        else:
            for jv in range(0, self.nvm, 1):
                jst = self.gmhs.pref_soil_veg[jv] - 1
                for jsl in range(0, self.nslm, 1):
                    for ji in range(0, self.kjpindex, 1):
                        if self.humrel[ji, jv] > self.gmhs.min_sechiba and self.soiltile[ji, jst] * self.gmhs.vegtot[ji] > self.gmhs.min_sechiba:
                            self.gmhs.rootsink[ji, jsl, jst] = self.gmhs.rootsink[ji, jsl, jst] + self.transpir[ji, jv] * (self.us[ji, jv, jst, jsl] / self.humrel[ji, jv]) / (self.soiltile[ji, jst] * self.gmhs.vegtot[ji])
        if self.gmhs.check_cwrr:
            error = False
            tmp_check1[:] = self.gmhs.zero
            for jst in range(0, self.nstm, 1):
                for ji in range(0, self.kjpindex, 1):
                    tmp_check1[ji] = tmp_check1[ji] + self.gmhs.precisol_ns[ji, jst] * self.soiltile[ji, jst] * self.gmhs.vegtot[ji]
            tmp_check2[:] = self.gmhs.zero
            for jv in range(0, self.nvm, 1):
                for ji in range(0, self.kjpindex, 1):
                    tmp_check2[ji] = tmp_check2[ji] + self.gmhs.precisol[ji, jv]
            for ji in range(0, self.kjpindex, 1):
                if np.abs(tmp_check1[ji] - tmp_check2[ji]) > self.gmhs.allowed_err:
                    logging.info(f'PRECISOL SPLIT FALSE:ji={ji},{tmp_check1[ji]},{tmp_check2[ji]}')
                    logging.info(f'err{np.abs(tmp_check1[ji] - tmp_check2[ji])}')
                    logging.info(f'vegtot{self.gmhs.vegtot[ji]}')
                    for jv in range(0, self.nvm, 1):
                        logging.info(f'jv,veget_max, precisol, vegetmax_soil{jv},{self.veget_max[ji, jv]},{self.gmhs.precisol[ji, jv]},{self.gmhs.vegetmax_soil[ji, jv, :]}')
                    for jst in range(0, self.nstm, 1):
                        logging.info(f'jst,precisol_ns{jst},{self.gmhs.precisol_ns[ji, jst]}')
                        logging.info(f'soiltile{self.soiltile[ji, jst]}')
                    error = True
                    logging.error('Exception: hydrol_split_soil We will STOP in the end of this subroutine. check_CWRR PRECISOL SPLIT FALSE')
            tmp_check1[:] = self.gmhs.zero
            for jst in range(0, self.nstm, 1):
                for ji in range(0, self.kjpindex, 1):
                    tmp_check1[ji] = tmp_check1[ji] + self.gmhs.ae_ns[ji, jst] * self.soiltile[ji, jst] * self.gmhs.vegtot[ji]
            for ji in range(0, self.kjpindex, 1):
                if np.abs(tmp_check1[ji] - self.vevapnu[ji]) > self.gmhs.allowed_err:
                    logging.info(f'VEVAPNU SPLIT FALSE:ji, Sum(ae_ns), vevapnu ={ji},{tmp_check1[ji]},{self.vevapnu[ji]}')
                    logging.info(f'err{np.abs(tmp_check1[ji] - self.vevapnu[ji])}')
                    logging.info(f'ae_ns{self.gmhs.ae_ns[ji, :]}')
                    logging.info(f'vegtot{self.gmhs.vegtot[ji]}')
                    logging.info(f'evap_bare_lim, evap_bare_lim_ns{self.evap_bare_lim[ji]},{self.evap_bare_lim_ns[ji, :]}')
                    for jst in range(0, self.nstm, 1):
                        logging.info(f'jst,ae_ns{jst},{self.gmhs.ae_ns[ji, jst]}')
                        logging.info(f'soiltile{self.soiltile[ji, jst]}')
                    error = True
                    logging.error('Exception: hydrol_split_soil We will STOP in the end of this subroutine. check_CWRR VEVAPNU SPLIT FALSE')
            tmp_check1[:] = self.gmhs.zero
            for jst in range(0, self.nstm, 1):
                for ji in range(0, self.kjpindex, 1):
                    tmp_check1[ji] = tmp_check1[ji] + self.gmhs.tr_ns[ji, jst] * self.soiltile[ji, jst] * self.gmhs.vegtot[ji]
            tmp_check2[:] = self.gmhs.zero
            for jv in range(0, self.nvm, 1):
                for ji in range(0, self.kjpindex, 1):
                    tmp_check2[ji] = tmp_check2[ji] + self.transpir[ji, jv]
            for ji in range(0, self.kjpindex, 1):
                if np.abs(tmp_check1[ji] - tmp_check2[ji]) > self.gmhs.allowed_err:
                    logging.info(f'TRANSPIR SPLIT FALSE:ji={ji},{tmp_check1[ji]},{tmp_check2[ji]}')
                    logging.info(f'err{np.abs(tmp_check1[ji] - tmp_check2[ji])}')
                    logging.info(f'vegtot{self.gmhs.vegtot[ji]}')
                    for jv in range(0, self.nvm, 1):
                        logging.info(f'jv,veget_max, transpir{jv},{self.veget_max[ji, jv]},{self.transpir[ji, jv]}')
                        for jst in range(0, self.nstm, 1):
                            logging.info(f'vegetmax_soil:ji,jv,jst{ji},{jv},{jst},{self.gmhs.vegetmax_soil[ji, jv, jst]}')
                    for jst in range(0, self.nstm, 1):
                        logging.info(f'jst,tr_ns{jst},{self.gmhs.tr_ns[ji, jst]}')
                        logging.info(f'soiltile{self.soiltile[ji, jst]}')
                    error = True
                    logging.error('Exception: hydrol_split_soil We will STOP in the end of this subroutine. check_CWRR TRANSPIR SPLIT FALSE')
            tmp_check3[:, :] = self.gmhs.zero
            for jst in range(0, self.nstm, 1):
                for jsl in range(0, self.nslm, 1):
                    for ji in range(0, self.kjpindex, 1):
                        tmp_check3[ji, jst] = tmp_check3[ji, jst] + self.gmhs.rootsink[ji, jsl, jst]
            for jst in range(0, self.nstm, 1):
                for ji in range(0, self.kjpindex, 1):
                    if np.abs(tmp_check3[ji, jst] - self.gmhs.tr_ns[ji, jst]) > self.gmhs.allowed_err:
                        logging.info(f'ROOTSINK SPLIT FALSE:ji,jst={ji},{jst},{tmp_check3[ji, jst]},{self.gmhs.tr_ns[ji, jst]}')
                        logging.info(f'err{np.abs(tmp_check3[ji, jst] - self.gmhs.tr_ns[ji, jst])}')
                        logging.info(f'HUMREL(jv=1:13){self.humrel[ji, :]}')
                        logging.info(f'TRANSPIR{self.transpir[ji, :]}')
                        for jv in range(0, self.nvm, 1):
                            logging.info(f'jv={jv},us={self.us[ji, jv, jst, :]}')
                        error = True
                        logging.error('Exception: hydrol_split_soil We will STOP in the end of this subroutine. check_CWRR ROOTSINK SPLIT FALSE')
            if error:
                logging.info('One or more errors have been detected in hydrol_split_soil. Model stops.')
                logging.error('Exception: hydrol_split_soil We will STOP now. One or several fatal errors were found previously. ')

    def hydrol_soil_setup(self, ins):
        temp3 = self.gmhs.w_time * (self.gmhs.dt_sechiba / self.gmhs.one_day) / self.gmhs.deux
        temp4 = (self.gmhs.un - self.gmhs.w_time) * (self.gmhs.dt_sechiba / self.gmhs.one_day) / self.gmhs.deux
        for ji in range(0, self.kjpindex, 1):
            self.gmhs.e[ji, 0] = self.gmhs.zero
            self.gmhs.f[ji, 0] = self.gmhs.trois * self.gmhs.dz[1] / self.gmhs.huit + temp3 * ((self.gmhs.d[ji, 0] + self.gmhs.d[ji, 1]) / self.gmhs.dz[1] + self.gmhs.a[ji, 0])
            self.gmhs.g1[ji, 0] = self.gmhs.dz[1] / self.gmhs.huit - temp3 * ((self.gmhs.d[ji, 0] + self.gmhs.d[ji, 1]) / self.gmhs.dz[1] - self.gmhs.a[ji, 1])
            self.gmhs.ep[ji, 0] = self.gmhs.zero
            self.gmhs.fp[ji, 0] = self.gmhs.trois * self.gmhs.dz[1] / self.gmhs.huit - temp4 * ((self.gmhs.d[ji, 0] + self.gmhs.d[ji, 1]) / self.gmhs.dz[1] + self.gmhs.a[ji, 0])
            self.gmhs.gp[ji, 0] = self.gmhs.dz[1] / self.gmhs.huit + temp4 * ((self.gmhs.d[ji, 0] + self.gmhs.d[ji, 1]) / self.gmhs.dz[1] - self.gmhs.a[ji, 1])
        for jsl in range(1, self.nslm - 1, 1):
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.e[ji, jsl] = self.gmhs.dz[jsl] / self.gmhs.huit - temp3 * ((self.gmhs.d[ji, jsl] + self.gmhs.d[ji, jsl - 1]) / self.gmhs.dz[jsl] + self.gmhs.a[ji, jsl - 1])
                self.gmhs.f[ji, jsl] = self.gmhs.trois * (self.gmhs.dz[jsl] + self.gmhs.dz[jsl + 1]) / self.gmhs.huit + temp3 * ((self.gmhs.d[ji, jsl] + self.gmhs.d[ji, jsl - 1]) / self.gmhs.dz[jsl] + (self.gmhs.d[ji, jsl] + self.gmhs.d[ji, jsl + 1]) / self.gmhs.dz[jsl + 1])
                self.gmhs.g1[ji, jsl] = self.gmhs.dz[jsl + 1] / self.gmhs.huit - temp3 * ((self.gmhs.d[ji, jsl] + self.gmhs.d[ji, jsl + 1]) / self.gmhs.dz[jsl + 1] - self.gmhs.a[ji, jsl + 1])
                self.gmhs.ep[ji, jsl] = self.gmhs.dz[jsl] / self.gmhs.huit + temp4 * ((self.gmhs.d[ji, jsl] + self.gmhs.d[ji, jsl - 1]) / self.gmhs.dz[jsl] + self.gmhs.a[ji, jsl - 1])
                self.gmhs.fp[ji, jsl] = self.gmhs.trois * (self.gmhs.dz[jsl] + self.gmhs.dz[jsl + 1]) / self.gmhs.huit - temp4 * ((self.gmhs.d[ji, jsl] + self.gmhs.d[ji, jsl - 1]) / self.gmhs.dz[jsl] + (self.gmhs.d[ji, jsl] + self.gmhs.d[ji, jsl + 1]) / self.gmhs.dz[jsl + 1])
                self.gmhs.gp[ji, jsl] = self.gmhs.dz[jsl + 1] / self.gmhs.huit + temp4 * ((self.gmhs.d[ji, jsl] + self.gmhs.d[ji, jsl + 1]) / self.gmhs.dz[jsl + 1] - self.gmhs.a[ji, jsl + 1])
        for ji in range(0, self.kjpindex, 1):
            self.gmhs.e[ji, self.nslm - 1] = self.gmhs.dz[self.nslm - 1] / self.gmhs.huit - temp3 * ((self.gmhs.d[ji, self.nslm - 1] + self.gmhs.d[ji, self.nslm - 2]) / self.gmhs.dz[self.nslm - 1] + self.gmhs.a[ji, self.nslm - 2])
            self.gmhs.f[ji, self.nslm - 1] = self.gmhs.trois * self.gmhs.dz[self.nslm - 1] / self.gmhs.huit + temp3 * ((self.gmhs.d[ji, self.nslm - 1] + self.gmhs.d[ji, self.nslm - 2]) / self.gmhs.dz[self.nslm - 1] - self.gmhs.a[ji, self.nslm - 1] * (self.gmhs.un - self.gmhs.deux * self.gmhs.free_drain_coef[ji, ins]))
            self.gmhs.g1[ji, self.nslm - 1] = self.gmhs.zero
            self.gmhs.ep[ji, self.nslm - 1] = self.gmhs.dz[self.nslm - 1] / self.gmhs.huit + temp4 * ((self.gmhs.d[ji, self.nslm - 1] + self.gmhs.d[ji, self.nslm - 2]) / self.gmhs.dz[self.nslm - 1] + self.gmhs.a[ji, self.nslm - 2])
            self.gmhs.fp[ji, self.nslm - 1] = self.gmhs.trois * self.gmhs.dz[self.nslm - 1] / self.gmhs.huit - temp4 * ((self.gmhs.d[ji, self.nslm - 1] + self.gmhs.d[ji, self.nslm - 2]) / self.gmhs.dz[self.nslm - 1] - self.gmhs.a[ji, self.nslm - 1] * (self.gmhs.un - self.gmhs.deux * self.gmhs.free_drain_coef[ji, ins]))
            self.gmhs.gp[ji, self.nslm - 1] = self.gmhs.zero

    def hydrol_soil_froz(self, ins):
        froz_frac_moy = np.zeros((self.kjpindex,), dtype=np.float64)
        smtot_moy = np.zeros((self.kjpindex,), dtype=np.float64)
        mc_ns = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        for jsl in range(0, self.nslm, 1):
            for ji in range(0, self.kjpindex, 1):
                m = 1.0 - 1.0 / self.nvan[ji]
                if not self.gmhs.ok_thermodynamical_freezing or self.gmhs.mc[ji, jsl, ins] < self.mcr[ji] + self.gmhs.min_sechiba:
                    if self.stempdiag[ji, jsl] >= self.gmhs.fr_center + self.gmhs.fr_dT / 2.0:
                        x = 1.0
                    elif self.stempdiag[ji, jsl] >= self.gmhs.fr_center - self.gmhs.fr_dT / 2.0 and self.stempdiag[ji, jsl] < self.gmhs.fr_center + self.gmhs.fr_dT / 2.0:
                        x = (self.stempdiag[ji, jsl] - (self.gmhs.fr_center - self.gmhs.fr_dT / 2.0)) / self.gmhs.fr_dT
                    else:
                        x = 0.0
                elif self.gmhs.ok_thermodynamical_freezing:
                    if self.stempdiag[ji, jsl] >= self.gmhs.fr_center + self.gmhs.fr_dT / 2.0:
                        x = 1.0
                    elif self.stempdiag[ji, jsl] >= self.gmhs.fr_center - self.gmhs.fr_dT / 2.0 and self.stempdiag[ji, jsl] < self.gmhs.fr_center + self.gmhs.fr_dT / 2.0:
                        x = np.minimum((self.mcs[ji] - self.mcr[ji]) * ((2.2 * 1000.0 * self.avan[ji] * (self.gmhs.fr_center + self.gmhs.fr_dT / 2.0 - self.stempdiag[ji, jsl]) * self.gmhs.lhf / self.gmhs.ZeroCelsius / 10.0) ** self.nvan[ji] + 1.0) ** (-m) / (self.gmhs.mc[ji, jsl, ins] - self.mcr[ji]), 1.0)
                    else:
                        x = 0.0
                self.gmhs.profil_froz_hydro_ns[ji, jsl, ins] = 1.0 - x
                mc_ns[ji, jsl] = self.gmhs.mc[ji, jsl, ins] / self.mcs[ji]
        froz_frac_moy[:] = self.gmhs.zero
        denom = self.gmhs.zero
        for jsl in range(0, self.nslm, 1):
            froz_frac_moy[:] = froz_frac_moy[:] + self.gmhs.dh[jsl] * self.gmhs.profil_froz_hydro_ns[:, jsl, ins]
            denom = denom + self.gmhs.dh[jsl]
        froz_frac_moy[:] = froz_frac_moy[:] / denom
        smtot_moy[:] = self.gmhs.zero
        denom = self.gmhs.zero
        for jsl in range(0, self.nslm - 1, 1):
            smtot_moy[:] = smtot_moy[:] + self.gmhs.dh[jsl] * mc_ns[:, jsl]
            denom = denom + self.gmhs.dh[jsl]
        smtot_moy[:] = smtot_moy[:] / denom
        for jsl in range(0, self.nslm, 1):
            self.gmhs.profil_froz_hydro_ns[:, jsl, ins] = np.minimum(self.gmhs.profil_froz_hydro_ns[:, jsl, ins] * froz_frac_moy[:] ** self.gmhs.froz_frac_corr * smtot_moy[:] ** self.gmhs.smtot_corr, self.gmhs.max_froz_hydro)

    def hydrol_soil_smooth_over_mcs2(self, ins, is_over_mcs, rudr_corr, check):
        excess = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        tmci = np.zeros((self.kjpindex,), dtype=np.float64)
        tmcf = np.zeros((self.kjpindex,), dtype=np.float64)
        if self.gmhs.check_cwrr:
            tmci[:] = self.gmhs.dz[1] * (self.gmhs.trois * self.gmhs.mc[:, 0, ins] + self.gmhs.mc[:, 1, ins]) / self.gmhs.huit
            for jsl in range(1, self.nslm - 1, 1):
                tmci[:] = tmci[:] + self.gmhs.dz[jsl] * (self.gmhs.trois * self.gmhs.mc[:, jsl, ins] + self.gmhs.mc[:, jsl - 1, ins]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * self.gmhs.mc[:, jsl, ins] + self.gmhs.mc[:, jsl + 1, ins]) / self.gmhs.huit
            tmci[:] = tmci[:] + self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * self.gmhs.mc[:, self.nslm - 1, ins] + self.gmhs.mc[:, self.nslm - 2, ins]) / self.gmhs.huit
        for jsl in range(0, self.nslm, 1):
            for ji in range(0, self.kjpindex, 1):
                excess[ji, jsl] = np.maximum(self.gmhs.mc[ji, jsl, ins] - self.mcs[ji], self.gmhs.zero)
                self.gmhs.mc[ji, jsl, ins] = self.gmhs.mc[ji, jsl, ins] - excess[ji, jsl]
        for ji in range(0, self.kjpindex, 1):
            rudr_corr[ji, ins] = self.gmhs.dz[1] * (self.gmhs.trois * excess[ji, 0] + excess[ji, 1]) / self.gmhs.huit
        for jsl in range(1, self.nslm - 1, 1):
            for ji in range(0, self.kjpindex, 1):
                rudr_corr[ji, ins] = rudr_corr[ji, ins] + self.gmhs.dz[jsl] * (self.gmhs.trois * excess[ji, jsl] + excess[ji, jsl - 1]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * excess[ji, jsl] + excess[ji, jsl + 1]) / self.gmhs.huit
        for ji in range(0, self.kjpindex, 1):
            rudr_corr[ji, ins] = rudr_corr[ji, ins] + self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * excess[ji, self.nslm - 1] + excess[ji, self.nslm - 2]) / self.gmhs.huit
            is_over_mcs[ji] = False
        if self.gmhs.check_cwrr:
            tmcf[:] = self.gmhs.dz[1] * (self.gmhs.trois * self.gmhs.mc[:, 0, ins] + self.gmhs.mc[:, 1, ins]) / self.gmhs.huit
            for jsl in range(1, self.nslm - 1, 1):
                tmcf[:] = tmcf[:] + self.gmhs.dz[jsl] * (self.gmhs.trois * self.gmhs.mc[:, jsl, ins] + self.gmhs.mc[:, jsl - 1, ins]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * self.gmhs.mc[:, jsl, ins] + self.gmhs.mc[:, jsl + 1, ins]) / self.gmhs.huit
            tmcf[:] = tmcf[:] + self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * self.gmhs.mc[:, self.nslm - 1, ins] + self.gmhs.mc[:, self.nslm - 2, ins]) / self.gmhs.huit
            check[:, ins] = tmcf[:] - (tmci[:] - rudr_corr[:, ins])

    def hydrol_nudge_mc(self, jst, mc_loc):
        mc_aux = np.zeros((self.kjpindex, self.nslm, self.nstm), dtype=np.float64)
        mc_loc[:, :, jst] = (1 - self.gmhs.alpha_nudge_mc) * mc_loc[:, :, jst] + self.gmhs.alpha_nudge_mc * self.gmhs.mc_read_current[:, :, jst]
        mc_aux[:, :, jst] = self.gmhs.alpha_nudge_mc * (self.gmhs.mc_read_current[:, :, jst] - mc_loc[:, :, jst])
        for ji in range(0, self.kjpindex, 1):
            self.gmhs.tmc_aux[ji, jst] = self.gmhs.dz[1] * (self.gmhs.trois * mc_aux[ji, 0, jst] + mc_aux[ji, 1, jst]) / self.gmhs.huit
            for jsl in range(1, self.nslm - 1, 1):
                self.gmhs.tmc_aux[ji, jst] = self.gmhs.tmc_aux[ji, jst] + self.gmhs.dz[jsl] * (self.gmhs.trois * mc_aux[ji, jsl, jst] + mc_aux[ji, jsl - 1, jst]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * mc_aux[ji, jsl, jst] + mc_aux[ji, jsl + 1, jst]) / self.gmhs.huit
            self.gmhs.tmc_aux[ji, jst] = self.gmhs.tmc_aux[ji, jst] + self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * mc_aux[ji, self.nslm - 1, jst] + mc_aux[ji, self.nslm - 2, jst]) / self.gmhs.huit

    def hydrol_soil_coef(self, ins):
        if self.gmhs.ok_freeze_cwrr:
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    x = 1.0 - self.gmhs.profil_froz_hydro_ns[ji, jsl, ins]
                    mc_used = self.mcr[ji] + x * np.maximum(self.gmhs.mc[ji, jsl, ins] - self.mcr[ji], self.gmhs.zero)
                    i = np.maximum(self.gmhs.imin, np.minimum(self.gmhs.imax - 1, int(self.gmhs.imin + (self.gmhs.imax - self.gmhs.imin) * (mc_used - self.mcr[ji]) / (self.mcs[ji] - self.mcr[ji]))))
                    self.gmhs.a[ji, jsl] = self.gmhs.a_lin[i - 1, jsl, ji] * self.gmhs.kfact_root[ji, jsl, ins]
                    self.gmhs.b[ji, jsl] = self.gmhs.b_lin[i - 1, jsl, ji] * self.gmhs.kfact_root[ji, jsl, ins]
                    self.gmhs.d[ji, jsl] = self.gmhs.d_lin[i - 1, jsl, ji] * self.gmhs.kfact_root[ji, jsl, ins]
                    self.gmhs.k[ji, jsl] = self.gmhs.kfact_root[ji, jsl, ins] * np.maximum(self.gmhs.k_lin[self.gmhs.imin, jsl, ji], self.gmhs.a_lin[i - 1, jsl, ji] * mc_used + self.gmhs.b_lin[i - 1, jsl, ji])
        else:
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    mc_ratio = np.maximum(self.gmhs.mc[ji, jsl, ins] - self.mcr[ji], self.gmhs.zero) / (self.mcs[ji] - self.mcr[ji])
                    i = np.maximum(np.minimum(int((self.gmhs.imax - self.gmhs.imin) * mc_ratio) + self.gmhs.imin, self.gmhs.imax - 1), self.gmhs.imin)
                    self.gmhs.a[ji, jsl] = self.gmhs.a_lin[i - 1, jsl, ji] * self.gmhs.kfact_root[ji, jsl, ins]
                    self.gmhs.b[ji, jsl] = self.gmhs.b_lin[i - 1, jsl, ji] * self.gmhs.kfact_root[ji, jsl, ins]
                    self.gmhs.d[ji, jsl] = self.gmhs.d_lin[i - 1, jsl, ji] * self.gmhs.kfact_root[ji, jsl, ins]
                    self.gmhs.k[ji, jsl] = self.gmhs.kfact_root[ji, jsl, ins] * np.maximum(self.gmhs.k_lin[self.gmhs.imin, jsl, ji], self.gmhs.a_lin[i - 1, jsl, ji] * self.gmhs.mc[ji, jsl, ins] + self.gmhs.b_lin[i - 1, jsl, ji])

    def hydrol_soil_infilt(self, ins, flux_infilt, qinfilt_ns, ru_infilt, check):
        wat_inf_pot = np.zeros((self.kjpindex,), dtype=np.float64)
        wat_inf = np.zeros((self.kjpindex,), dtype=np.float64)
        dt_tmp = np.zeros((self.kjpindex,), dtype=np.float64)
        dt_inf = np.zeros((self.kjpindex,), dtype=np.float64)
        infilt_tmp = np.zeros((self.kjpindex,), dtype=np.float64)
        infilt_tot = np.zeros((self.kjpindex,), dtype=np.float64)
        flux_tmp = np.zeros((self.kjpindex,), dtype=np.float64)
        tmci = np.zeros((self.kjpindex,), dtype=np.float64)
        tmcf = np.zeros((self.kjpindex,), dtype=np.float64)
        if self.gmhs.check_cwrr:
            tmci[:] = self.gmhs.dz[1] * (self.gmhs.trois * self.gmhs.mc[:, 0, ins] + self.gmhs.mc[:, 1, ins]) / self.gmhs.huit
            for jsl in range(1, self.nslm - 1, 1):
                tmci[:] = tmci[:] + self.gmhs.dz[jsl] * (self.gmhs.trois * self.gmhs.mc[:, jsl, ins] + self.gmhs.mc[:, jsl - 1, ins]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * self.gmhs.mc[:, jsl, ins] + self.gmhs.mc[:, jsl + 1, ins]) / self.gmhs.huit
            tmci[:] = tmci[:] + self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * self.gmhs.mc[:, self.nslm - 1, ins] + self.gmhs.mc[:, self.nslm - 2, ins]) / self.gmhs.huit
        for ji in range(0, self.kjpindex, 1):
            wat_inf_pot[ji] = np.maximum((self.mcs[ji] - self.gmhs.mc[ji, 0, ins]) * self.gmhs.dz[1] / self.gmhs.deux, self.gmhs.zero)
            wat_inf[ji] = np.minimum(wat_inf_pot[ji], flux_infilt[ji])
            self.gmhs.mc[ji, 0, ins] = self.gmhs.mc[ji, 0, ins] + wat_inf[ji] * self.gmhs.deux / self.gmhs.dz[1]
        dt_tmp[:] = self.gmhs.dt_sechiba / self.gmhs.one_day
        infilt_tot[:] = wat_inf[:]
        flux_tmp[:] = (flux_infilt[:] - wat_inf[:]) / dt_tmp[:]
        for jsl in range(1, self.nslm - 1, 1):
            for ji in range(0, self.kjpindex, 1):
                k_m = (self.gmhs.k[ji, jsl] + self.ks[ji] * self.gmhs.kfact[jsl - 1, ji] * self.gmhs.kfact_root[ji, jsl, ins]) / self.gmhs.deux
                if self.gmhs.ok_freeze_cwrr:
                    if self.stempdiag[ji, jsl] < self.gmhs.ZeroCelsius:
                        k_m = self.gmhs.k[ji, jsl]
                infilt_tmp[ji] = k_m * (self.gmhs.un - np.exp(-(flux_tmp[ji] / k_m)))
                wat_inf_pot[ji] = np.maximum((self.mcs[ji] - self.gmhs.mc[ji, jsl, ins]) * (self.gmhs.dz[jsl] + self.gmhs.dz[jsl + 1]) / self.gmhs.deux, self.gmhs.zero)
                if infilt_tmp[ji] > self.gmhs.min_sechiba:
                    dt_inf[ji] = np.minimum(wat_inf_pot[ji] / infilt_tmp[ji], dt_tmp[ji])
                    if dt_inf[ji] * infilt_tmp[ji] > flux_infilt[ji] - infilt_tot[ji]:
                        dt_inf[ji] = np.maximum(flux_infilt[ji] - infilt_tot[ji], self.gmhs.zero) / infilt_tmp[ji]
                else:
                    dt_inf[ji] = dt_tmp[ji]
                wat_inf[ji] = dt_inf[ji] * infilt_tmp[ji]
                self.gmhs.mc[ji, jsl, ins] = self.gmhs.mc[ji, jsl, ins] + wat_inf[ji] * self.gmhs.deux / (self.gmhs.dz[jsl] + self.gmhs.dz[jsl + 1])
                dt_tmp[ji] = dt_tmp[ji] - dt_inf[ji]
                infilt_tot[ji] = infilt_tot[ji] + infilt_tmp[ji] * dt_inf[ji]
        ru_infilt[:, ins] = flux_infilt[:] - infilt_tot[:]
        qinfilt_ns[:, ins] = infilt_tot[:]
        if self.gmhs.check_cwrr:
            tmcf[:] = self.gmhs.dz[1] * (self.gmhs.trois * self.gmhs.mc[:, 0, ins] + self.gmhs.mc[:, 1, ins]) / self.gmhs.huit
            for jsl in range(1, self.nslm - 1, 1):
                tmcf[:] = tmcf[:] + self.gmhs.dz[jsl] * (self.gmhs.trois * self.gmhs.mc[:, jsl, ins] + self.gmhs.mc[:, jsl - 1, ins]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * self.gmhs.mc[:, jsl, ins] + self.gmhs.mc[:, jsl + 1, ins]) / self.gmhs.huit
            tmcf[:] = tmcf[:] + self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * self.gmhs.mc[:, self.nslm - 1, ins] + self.gmhs.mc[:, self.nslm - 2, ins]) / self.gmhs.huit
            check[:, ins] = tmcf[:] - (tmci[:] + infilt_tot[:])
        for ji in range(0, self.kjpindex, 1):
            if infilt_tot[ji] < -self.gmhs.min_sechiba or infilt_tot[ji] > flux_infilt[ji] + self.gmhs.min_sechiba:
                logging.info(f'Error in the calculation of infilt tot{infilt_tot[ji]}')
                logging.info(f'k, ji, jst, mc{self.gmhs.k[ji, :2]},{ji},{ins},{self.gmhs.mc[ji, 0, ins]}')
                logging.error('Exception: hydrol_soil_infilt We will STOP now. Error in calculation of infilt tot ')

    def hydrol_root_profile(self, sm, smw):
        count = np.zeros((self.kjpindex,), dtype=np.float64)
        root_profile_tmp = np.zeros((self.nslm,), dtype=np.float64)
        self.root_depth[:, :, :] = self.gmhs.zero
        for ji in range(0, self.kjpindex, 1):
            for jv in range(0, self.nvm, 1):
                root_depth_tmp = np.minimum(np.minimum(self.altmax[ji, jv], self.gmhs.maxaltmax), np.minimum(self.gmhs.zdr[self.nslm], self.gmhs.max_root_depth[jv]))
                self.root_depth[ji, jv, self.gmhs.inode - 1] = np.argmin(np.abs(root_depth_tmp - self.gmhs.znh[:]), axis=0) + 1
                self.root_depth[ji, jv, self.gmhs.iinterface - 1] = np.argmin(np.abs(root_depth_tmp - self.gmhs.zdr[:]), axis=0)
            if (self.root_depth[:, :, self.gmhs.inode - 2] <= 2).any():
                mask = (self.root_depth[:, :, self.gmhs.inode - 2] <= 2)
                self.root_depth[:, :, self.gmhs.inode - 1][mask] = 2
            if (self.root_depth[:, :, self.gmhs.iinterface - 2] <= 2).any():
                mask = (self.root_depth[:, :, self.gmhs.iinterface - 2] <= 2)
                self.root_depth[:, :, self.gmhs.iinterface - 1][mask] = 2
        self.root_profile[:, :, :, self.gmhs.istruc - 1] = self.gmhs.zero
        for jv in range(0, self.nvm, 1):
            for ji in range(0, self.kjpindex, 1):
                z_top = self.gmhs.zdr[1]
                z_bottom = self.gmhs.zdr[int(self.root_depth[ji, jv, self.gmhs.iinterface - 1])]
                rpc = self.gmhs.un / (np.exp(-(z_top * self.gmhs.humcste[jv])) - np.exp(-(z_bottom * self.gmhs.humcste[jv])))
                for jsl in range(1, int(self.root_depth[ji, jv, self.gmhs.iinterface - 1]), 1):
                    self.root_profile[ji, jv, jsl, self.gmhs.istruc - 1] = rpc * (np.exp(-(self.gmhs.zdr[jsl] * self.gmhs.humcste[jv])) - np.exp(-(self.gmhs.zdr[jsl + 1] * self.gmhs.humcste[jv])))
                self.root_profile[ji, jv, 0, self.gmhs.istruc - 1] = self.gmhs.zero
                if self.gmhs.err_act > 1:
                    if np.abs(np.sum(self.root_profile[ji, jv, :, self.gmhs.istruc - 1]) - self.gmhs.un) > 100 * np.finfo(np.float64).eps:
                        logging.info(f'pixel, PFT, sum of structural root_profile,{ji},{jv},{np.sum(self.root_profile[ji, jv, :, self.gmhs.istruc - 1])}')
                        logging.info(f'hydrol_root_profile, structural root_profile,{self.root_profile[ji, jv, :, self.gmhs.istruc - 1]}')
                        logging.error('Exception: hydrol.f90 structural root profile does not add up to 1 Check its calculation ')
        self.root_profile[:, :, :, self.gmhs.ifunc - 1] = self.gmhs.zero
        for jv in range(1, self.nvm, 1):
            for ji in range(0, self.kjpindex, 1):
                root_profile_tmp[:] = self.gmhs.zero
                for jsl in range(1, int(self.root_depth[ji, jv, self.gmhs.iinterface - 1]), 1):
                    root_profile_tmp[jsl] = np.maximum(self.gmhs.zero, sm[ji, jsl] - smw[ji, jsl])
                if np.sum(root_profile_tmp[:]) > self.gmhs.min_sechiba:
                    self.root_profile[ji, jv, :, self.gmhs.ifunc - 1] = root_profile_tmp[:] / np.sum(root_profile_tmp[:])
                else:
                    self.root_profile[ji, jv, :, self.gmhs.ifunc - 1] = self.gmhs.zero
                self.root_profile[ji, jv, 0, self.gmhs.ifunc - 1] = self.gmhs.zero
                if self.gmhs.err_act > 1:
                    if np.abs(np.sum(self.root_profile[ji, jv, :, self.gmhs.ifunc - 1]) - self.gmhs.un) > 100 * np.finfo(np.float64).eps and np.sum(root_profile_tmp[:]) > self.gmhs.min_sechiba:
                        logging.info(f'pixel, PFT, sum of functional root_profile,{ji},{jv},{np.sum(self.root_profile[ji, jv, :, self.gmhs.ifunc - 1])}')
                        logging.info(f'hydrol_root_profile, functional root_profile,{self.root_profile[ji, jv, :, self.gmhs.ifunc - 1]}')
                        logging.error('Exception: hydrol.f90 functional root profile does not add up to 1 Check its calculation ')

    def hydrol_soil(self):
        frac_mcs = np.float64(0.66)
        temp = np.zeros((self.kjpindex,), dtype=np.float64)
        tmcold = np.zeros((self.kjpindex,), dtype=np.float64)
        tmcint = np.zeros((self.kjpindex,), dtype=np.float64)
        mcint = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        mclint = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        is_under_mcr = np.zeros((self.kjpindex, self.nstm), dtype=np.bool)
        is_over_mcs = np.zeros((self.kjpindex,), dtype=np.bool)
        diff = np.zeros((self.kjpindex,), dtype=np.float64)
        deltahum = np.zeros((self.kjpindex,), dtype=np.float64)
        test = np.zeros((self.kjpindex,), dtype=np.bool)
        water2extract = np.zeros((self.kjpindex,), dtype=np.float64)
        returnflow_soil = np.zeros((self.kjpindex,), dtype=np.float64)
        reinfiltration_soil = np.zeros((self.kjpindex,), dtype=np.float64)
        irrigation_soil = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        flux_infilt = np.zeros((self.kjpindex,), dtype=np.float64)
        flux_bottom = np.zeros((self.kjpindex,), dtype=np.float64)
        flux_top = np.zeros((self.kjpindex,), dtype=np.float64)
        qinfilt_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        qinfilt = np.zeros((self.kjpindex,), dtype=np.float64)
        ru_infilt_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        ru_infilt = np.zeros((self.kjpindex,), dtype=np.float64)
        ru_corr_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        ru_corr = np.zeros((self.kjpindex,), dtype=np.float64)
        ru_corr2_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        ru_corr2 = np.zeros((self.kjpindex,), dtype=np.float64)
        dr_corr_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        dr_corrnum_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        dr_corr = np.zeros((self.kjpindex,), dtype=np.float64)
        dr_corrnum = np.zeros((self.kjpindex,), dtype=np.float64)
        dmc = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        dr_force_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        dr_force = np.zeros((self.kjpindex,), dtype=np.float64)
        wtd_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        wtd = np.zeros((self.kjpindex,), dtype=np.float64)
        sm = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        smt = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        smw = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        smf = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        sms = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        sm_nostress = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        check_infilt_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        check1_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        check_tr_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        check_over_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        check_under_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        tmci = np.zeros((self.kjpindex,), dtype=np.float64)
        tmcf = np.zeros((self.kjpindex,), dtype=np.float64)
        diag_tr = np.zeros((self.kjpindex,), dtype=np.float64)
        check_infilt = np.zeros((self.kjpindex,), dtype=np.float64)
        check1 = np.zeros((self.kjpindex,), dtype=np.float64)
        check_tr = np.zeros((self.kjpindex,), dtype=np.float64)
        check_over = np.zeros((self.kjpindex,), dtype=np.float64)
        check_under = np.zeros((self.kjpindex,), dtype=np.float64)
        lai_irrig_trig = np.zeros((self.kjpindex,), dtype=np.int32)
        qflux = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        check_top = np.zeros((self.kjpindex,), dtype=np.float64)
        mc_rel = np.zeros((self.kjpindex,), dtype=np.float64)
        evap_soil = np.zeros((self.kjpindex,), dtype=np.float64)
        r_soil_ns = np.zeros((self.kjpindex, self.nstm), dtype=np.float64)
        r_soil = np.zeros((self.kjpindex,), dtype=np.float64)
        tmcs_litter = np.zeros((self.kjpindex,), dtype=np.float64)
        root_profile_tmp = np.zeros((self.nslm,), dtype=np.float64)
        psi_moy = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        ksat = np.zeros((self.kjpindex, self.nslm), dtype=np.float64)
        tmp = np.zeros((self.kjpindex, self.nvm, self.nslm, self.gmhs.nroot_prof), dtype=np.float64)
        returnflow_soil[:] = self.gmhs.zero
        reinfiltration_soil[:] = self.gmhs.zero
        irrigation_soil[:, :] = self.gmhs.zero
        self.gmhs.qflux_ns[:, :, :] = self.gmhs.zero
        self.mc_layh[:, :] = self.gmhs.zero
        self.mcl_layh[:, :] = self.gmhs.zero
        self.gmhs.kk[:, :, :] = self.gmhs.zero
        self.gmhs.kk_moy[:, :] = self.gmhs.zero
        self.gmhs.undermcr[:] = self.gmhs.zero
        ksat[:, :] = self.gmhs.zero
        psi_moy[:, :] = self.gmhs.zero
        if self.gmhs.kfact_root_const:
            self.gmhs.kfact_root[:, :, :] = self.gmhs.un
        else:
            self.gmhs.kfact_root[:, :, :] = self.gmhs.un
            for jsl in range(0, self.nslm, 1):
                for jv in range(1, self.nvm, 1):
                    jst_kfact_root = self.gmhs.pref_soil_veg[jv]
                    for ji in range(0, self.kjpindex, 1):
                        if self.soiltile[ji, jst_kfact_root - 1] > self.gmhs.min_sechiba:
                            self.gmhs.kfact_root[ji, jsl, jst_kfact_root - 1] = self.gmhs.kfact_root[ji, jsl, jst_kfact_root - 1] * np.maximum((np.max(self.gmhs.ks_usda) / self.ks[ji]) ** (-(self.gmhs.vegetmax_soil[ji, jv, jst_kfact_root - 1] / 2 * (self.gmhs.humcste[jv] * self.gmhs.zz[jsl] / self.gmhs.mille - self.gmhs.un) / self.gmhs.deux)), self.gmhs.un)
        if self.gmhs.ok_freeze_cwrr:
            for jst in range(0, self.nstm, 1):
                self.hydrol_soil_froz(jst)
        else:
            self.gmhs.profil_froz_hydro_ns[:, :, :] = self.gmhs.zero
        self.hydrol_split_soil()
        if not self.gmhs.old_irrig_scheme:
            if not self.gmhs.irrigated_soiltile:
                for ji in range(0, self.kjpindex, 1):
                    if self.gmhs.vegtot[ji] > self.gmhs.min_sechiba:
                        returnflow_soil[ji] = self.gmhs.zero
                        reinfiltration_soil[ji] = (self.returnflow[ji] + self.reinfiltration[ji]) / self.gmhs.vegtot[ji]
                        if self.soiltile[ji, self.gmhs.irrig_st - 1] > self.gmhs.min_sechiba:
                            irrigation_soil[ji, self.gmhs.irrig_st - 1] = self.irrigation[ji] / (self.soiltile[ji, self.gmhs.irrig_st - 1] * self.gmhs.vegtot[ji])
        else:
            for ji in range(0, self.kjpindex, 1):
                if self.gmhs.vegtot[ji] > self.gmhs.min_sechiba:
                    returnflow_soil[ji] = self.gmhs.zero
                    reinfiltration_soil[ji] = (self.returnflow[ji] + self.reinfiltration[ji]) / self.gmhs.vegtot[ji]
                    irrigation_soil[ji, :] = self.irrigation[ji] / self.gmhs.vegtot[ji]
        for jst in range(0, self.nstm, 1):
            is_under_mcr[:, jst] = False
            is_over_mcs[:] = False
            tmcold[:] = self.gmhs.tmc[:, jst]
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    mcint[ji, jsl] = self.gmhs.mask_soiltile[ji, jst] * self.gmhs.mc[ji, jsl, jst]
            for ji in range(0, self.kjpindex, 1):
                tmcint[ji] = self.gmhs.dz[1] * (self.gmhs.trois * mcint[ji, 0] + mcint[ji, 1]) / self.gmhs.huit
            for jsl in range(1, self.nslm - 1, 1):
                for ji in range(0, self.kjpindex, 1):
                    tmcint[ji] = tmcint[ji] + self.gmhs.dz[jsl] * (self.gmhs.trois * mcint[ji, jsl] + mcint[ji, jsl - 1]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * mcint[ji, jsl] + mcint[ji, jsl + 1]) / self.gmhs.huit
            for ji in range(0, self.kjpindex, 1):
                tmcint[ji] = tmcint[ji] + self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * mcint[ji, self.nslm - 1] + mcint[ji, self.nslm - 2]) / self.gmhs.huit
            for ji in range(0, self.kjpindex, 1):
                temp[ji] = np.minimum(self.gmhs.water2infilt[ji, jst] + irrigation_soil[ji, jst] + reinfiltration_soil[ji] - np.minimum(self.gmhs.ae_ns[ji, jst], self.gmhs.zero) - np.minimum(self.gmhs.subsinksoil[ji], self.gmhs.zero) + self.gmhs.precisol_ns[ji, jst], np.maximum(self.gmhs.ae_ns[ji, jst], self.gmhs.zero) + np.maximum(self.gmhs.subsinksoil[ji], self.gmhs.zero))
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.water2infilt[ji, jst] = self.gmhs.water2infilt[ji, jst] + irrigation_soil[ji, jst] + reinfiltration_soil[ji] - np.minimum(self.gmhs.ae_ns[ji, jst], self.gmhs.zero) - np.minimum(self.gmhs.subsinksoil[ji], self.gmhs.zero) + self.gmhs.precisol_ns[ji, jst] - temp[ji]
            for ji in range(0, self.kjpindex, 1):
                water2extract[ji] = np.maximum(self.gmhs.ae_ns[ji, jst], self.gmhs.zero) + np.maximum(self.gmhs.subsinksoil[ji], self.gmhs.zero) - temp[ji]
            self.gmhs.ae_ns[:, jst] = self.gmhs.ae_ns[:, jst] + self.gmhs.subsinksoil[:]
            flux_top[:] = water2extract[:]
            for ji in range(0, self.kjpindex, 1):
                flux_infilt[ji] = self.gmhs.water2infilt[ji, jst]
            self.hydrol_soil_coef(jst)
            self.hydrol_soil_infilt(jst, flux_infilt, qinfilt_ns, ru_infilt_ns, check_infilt_ns)
            self.gmhs.ru_ns[:, jst] = ru_infilt_ns[:, jst]
            if not self.gmhs.doponds:
                for ji in range(0, self.kjpindex, 1):
                    self.gmhs.water2infilt[ji, jst] = self.reinf_slope_soil[ji, jst] * self.gmhs.ru_ns[ji, jst]
            else:
                for ji in range(0, self.kjpindex, 1):
                    self.gmhs.water2infilt[ji, jst] = self.gmhs.zero
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.ru_ns[ji, jst] = self.gmhs.ru_ns[ji, jst] - self.gmhs.water2infilt[ji, jst]
            self.hydrol_soil_coef(jst)
            self.hydrol_soil_setup(jst)
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.gmhs.mcl[ji, jsl, jst] = np.minimum(self.gmhs.mc[ji, jsl, jst], self.mcr[ji] + (self.gmhs.un - self.gmhs.profil_froz_hydro_ns[ji, jsl, jst]) * (self.gmhs.mc[ji, jsl, jst] - self.mcr[ji]))
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    mclint[ji, jsl] = self.gmhs.mask_soiltile[ji, jst] * self.gmhs.mcl[ji, jsl, jst]
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    if self.soiltile[ji, jst] > self.gmhs.zero:
                        mvg = self.gmhs.un - self.gmhs.un / self.gmhs.nvan_mod_tab[jsl, ji]
                        avg = self.gmhs.avan_mod_tab[jsl, ji] * 1000.0
                        mc_ratio = np.maximum(10.0 ** (-(14 * mvg)), (self.gmhs.mcl[ji, jsl, jst] - self.mcr[ji]) / (self.mcs[ji] - self.mcr[ji])) ** (-(self.gmhs.un / mvg))
                        psi = -(np.maximum(self.gmhs.zero, mc_ratio - self.gmhs.un) ** (self.gmhs.un / self.gmhs.nvan_mod_tab[jsl, ji]) / avg)
                        psi_moy[ji, jsl] = psi_moy[ji, jsl] + self.soiltile[ji, jst] * psi
            tmci[:] = self.gmhs.dz[1] * (self.gmhs.trois * self.gmhs.mcl[:, 0, jst] + self.gmhs.mcl[:, 1, jst]) / self.gmhs.huit
            for jsl in range(1, self.nslm - 1, 1):
                tmci[:] = tmci[:] + self.gmhs.dz[jsl] * (self.gmhs.trois * self.gmhs.mcl[:, jsl, jst] + self.gmhs.mcl[:, jsl - 1, jst]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * self.gmhs.mcl[:, jsl, jst] + self.gmhs.mcl[:, jsl + 1, jst]) / self.gmhs.huit
            tmci[:] = tmci[:] + self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * self.gmhs.mcl[:, self.nslm - 1, jst] + self.gmhs.mcl[:, self.nslm - 2, jst]) / self.gmhs.huit
            self.gmhs.resolv[:] = self.gmhs.mask_soiltile[:, jst] > 0
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.tmat[ji, 0, 0] = self.gmhs.zero
                self.gmhs.tmat[ji, 0, 1] = self.gmhs.f[ji, 0]
                self.gmhs.tmat[ji, 0, 2] = self.gmhs.g1[ji, 0]
                self.gmhs.rhs[ji, 0] = self.gmhs.fp[ji, 0] * self.gmhs.mcl[ji, 0, jst] + self.gmhs.gp[ji, 0] * self.gmhs.mcl[ji, 1, jst] - flux_top[ji] - (self.gmhs.b[ji, 0] + self.gmhs.b[ji, 1]) / self.gmhs.deux * (self.gmhs.dt_sechiba / self.gmhs.one_day) - self.gmhs.rootsink[ji, 0, jst]
            for jsl in range(1, self.nslm - 1, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.gmhs.tmat[ji, jsl, 0] = self.gmhs.e[ji, jsl]
                    self.gmhs.tmat[ji, jsl, 1] = self.gmhs.f[ji, jsl]
                    self.gmhs.tmat[ji, jsl, 2] = self.gmhs.g1[ji, jsl]
                    self.gmhs.rhs[ji, jsl] = self.gmhs.ep[ji, jsl] * self.gmhs.mcl[ji, jsl - 1, jst] + self.gmhs.fp[ji, jsl] * self.gmhs.mcl[ji, jsl, jst] + self.gmhs.gp[ji, jsl] * self.gmhs.mcl[ji, jsl + 1, jst] + (self.gmhs.b[ji, jsl - 1] - self.gmhs.b[ji, jsl + 1]) * (self.gmhs.dt_sechiba / self.gmhs.one_day) / self.gmhs.deux - self.gmhs.rootsink[ji, jsl, jst]
            for ji in range(0, self.kjpindex, 1):
                jsl = self.nslm - 1
                self.gmhs.tmat[ji, jsl, 0] = self.gmhs.e[ji, jsl]
                self.gmhs.tmat[ji, jsl, 1] = self.gmhs.f[ji, jsl]
                self.gmhs.tmat[ji, jsl, 2] = self.gmhs.zero
                self.gmhs.rhs[ji, jsl] = self.gmhs.ep[ji, jsl] * self.gmhs.mcl[ji, jsl - 1, jst] + self.gmhs.fp[ji, jsl] * self.gmhs.mcl[ji, jsl, jst] + (self.gmhs.b[ji, jsl - 1] + self.gmhs.b[ji, jsl] * (self.gmhs.un - self.gmhs.deux * self.gmhs.free_drain_coef[ji, jst])) * (self.gmhs.dt_sechiba / self.gmhs.one_day) / self.gmhs.deux - self.gmhs.rootsink[ji, jsl, jst]
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.gmhs.srhs[ji, jsl] = self.gmhs.rhs[ji, jsl]
                    self.gmhs.stmat[ji, jsl, 0] = self.gmhs.tmat[ji, jsl, 0]
                    self.gmhs.stmat[ji, jsl, 1] = self.gmhs.tmat[ji, jsl, 1]
                    self.gmhs.stmat[ji, jsl, 2] = self.gmhs.tmat[ji, jsl, 2]
            self.hydrol_soil_tridiag(jst)
            for ji in range(0, self.kjpindex, 1):
                if self.gmhs.resolv[ji]:
                    self.gmhs.dr_ns[ji, jst] = self.gmhs.mask_soiltile[ji, jst] * self.gmhs.k[ji, self.nslm - 1] * self.gmhs.free_drain_coef[ji, jst] * (self.gmhs.dt_sechiba / self.gmhs.one_day)
                else:
                    self.gmhs.dr_ns[ji, jst] = self.gmhs.zero
            tmcf[:] = self.gmhs.dz[1] * (self.gmhs.trois * self.gmhs.mcl[:, 0, jst] + self.gmhs.mcl[:, 1, jst]) / self.gmhs.huit
            for jsl in range(1, self.nslm - 1, 1):
                tmcf[:] = tmcf[:] + self.gmhs.dz[jsl] * (self.gmhs.trois * self.gmhs.mcl[:, jsl, jst] + self.gmhs.mcl[:, jsl - 1, jst]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * self.gmhs.mcl[:, jsl, jst] + self.gmhs.mcl[:, jsl + 1, jst]) / self.gmhs.huit
            tmcf[:] = tmcf[:] + self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * self.gmhs.mcl[:, self.nslm - 1, jst] + self.gmhs.mcl[:, self.nslm - 2, jst]) / self.gmhs.huit
            for ji in range(0, self.kjpindex, 1):
                diag_tr[ji] = np.sum(self.gmhs.rootsink[ji, :, jst])
            check_tr_ns[:, jst] = tmcf[:] - (tmci[:] - flux_top[:] - self.gmhs.dr_ns[:, jst] - diag_tr[:])
            for ji in range(0, self.kjpindex, 1):
                if check_tr_ns[ji, jst] < self.gmhs.zero:
                    dr_corrnum_ns[ji, jst] = -check_tr_ns[ji, jst]
                else:
                    dr_corrnum_ns[ji, jst] = -np.minimum(self.gmhs.dr_ns[ji, jst], check_tr_ns[ji, jst])
                self.gmhs.dr_ns[ji, jst] = self.gmhs.dr_ns[ji, jst] + dr_corrnum_ns[ji, jst]
            if self.gmhs.check_cwrr:
                check_tr_ns[:, jst] = tmcf[:] - (tmci[:] - flux_top[:] - self.gmhs.dr_ns[:, jst] - diag_tr[:])
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.gmhs.mc[ji, jsl, jst] = np.maximum(self.gmhs.mcl[ji, jsl, jst], self.gmhs.mcl[ji, jsl, jst] + self.gmhs.profil_froz_hydro_ns[ji, jsl, jst] * (self.gmhs.mc[ji, jsl, jst] - self.mcr[ji]))
            dr_corr_ns[:, jst] = self.gmhs.zero
            ru_corr_ns[:, jst] = self.gmhs.zero
            self.hydrol_soil_smooth_over_mcs2(jst, is_over_mcs, ru_corr_ns, check_over_ns)
            for ji in range(0, self.kjpindex, 1):
                if self.gmhs.free_drain_coef[ji, jst] >= 0.5 and (not self.gmhs.ok_freeze_cwrr):
                    dr_corr_ns[ji, jst] = ru_corr_ns[ji, jst]
                    ru_corr_ns[ji, jst] = self.gmhs.zero
            self.gmhs.dr_ns[:, jst] = self.gmhs.dr_ns[:, jst] + dr_corr_ns[:, jst]
            self.gmhs.ru_ns[:, jst] = self.gmhs.ru_ns[:, jst] + ru_corr_ns[:, jst]
            ru_corr2_ns[:, jst] = self.gmhs.zero
            for ji in range(0, self.kjpindex, 1):
                if self.gmhs.ru_ns[ji, jst] < self.gmhs.zero:
                    if self.gmhs.printlev >= 3:
                        logging.info(f'NEGATIVE RU_NS: runoff and drainage before correction{self.gmhs.ru_ns[ji, jst]},{self.gmhs.dr_ns[ji, jst]}')
                    self.gmhs.dr_ns[ji, jst] = self.gmhs.dr_ns[ji, jst] + self.gmhs.ru_ns[ji, jst]
                    ru_corr2_ns[ji, jst] = -self.gmhs.ru_ns[ji, jst]
                    self.gmhs.ru_ns[ji, jst] = 0.0
            if self.gmhs.ok_nudge_mc:
                self.hydrol_nudge_mc(jst, self.gmhs.mc)
            if self.gmhs.zwt_force[0, jst] <= self.gmhs.zmaxh:
                for jsl in range(0, self.nslm, 1):
                    for ji in range(0, self.kjpindex, 1):
                        dmc[ji, jsl] = self.gmhs.zero
                        if self.gmhs.zz[jsl] >= self.gmhs.zwt_force[ji, jst] * self.gmhs.mille:
                            dmc[ji, jsl] = self.mcs[ji] - self.gmhs.mc[ji, jsl, jst]
                            self.gmhs.mc[ji, jsl, jst] = self.mcs[ji]
                for ji in range(0, self.kjpindex, 1):
                    dr_force_ns[ji, jst] = self.gmhs.dz[1] * (self.gmhs.trois * dmc[ji, 0] + dmc[ji, 1]) / self.gmhs.huit
                for jsl in range(1, self.nslm - 1, 1):
                    for ji in range(0, self.kjpindex, 1):
                        dr_force_ns[ji, jst] = dr_force_ns[ji, jst] + self.gmhs.dz[jsl] * (self.gmhs.trois * dmc[ji, jsl] + dmc[ji, jsl - 1]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * dmc[ji, jsl] + dmc[ji, jsl + 1]) / self.gmhs.huit
                for ji in range(0, self.kjpindex, 1):
                    dr_force_ns[ji, jst] = dr_force_ns[ji, jst] + self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * dmc[ji, self.nslm - 1] + dmc[ji, self.nslm - 2]) / self.gmhs.huit
                    self.gmhs.dr_ns[ji, jst] = self.gmhs.dr_ns[ji, jst] - dr_force_ns[ji, jst]
            else:
                dr_force_ns[:, jst] = self.gmhs.zero
            for ji in range(0, self.kjpindex, 1):
                wtd_ns[ji, jst] = self.gmhs.undef_sechiba
                jsl = self.nslm - 1
                while self.gmhs.mc[ji, jsl, jst] == self.mcs[ji] and jsl > 1:
                    wtd_ns[ji, jst] = self.gmhs.zz[jsl] / self.gmhs.mille
                    jsl = jsl - 2
            self.hydrol_soil_smooth_under_mcr(jst, is_under_mcr, check_under_ns)
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.tmc[ji, jst] = self.gmhs.dz[1] * (self.gmhs.trois * self.gmhs.mc[ji, 0, jst] + self.gmhs.mc[ji, 1, jst]) / self.gmhs.huit
            for jsl in range(1, self.nslm - 1, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.gmhs.tmc[ji, jst] = self.gmhs.tmc[ji, jst] + self.gmhs.dz[jsl] * (self.gmhs.trois * self.gmhs.mc[ji, jsl, jst] + self.gmhs.mc[ji, jsl - 1, jst]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * self.gmhs.mc[ji, jsl, jst] + self.gmhs.mc[ji, jsl + 1, jst]) / self.gmhs.huit
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.tmc[ji, jst] = self.gmhs.tmc[ji, jst] + self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * self.gmhs.mc[ji, self.nslm - 1, jst] + self.gmhs.mc[ji, self.nslm - 2, jst]) / self.gmhs.huit
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.gmhs.mcl[ji, jsl, jst] = np.minimum(self.gmhs.mc[ji, jsl, jst], self.mcr[ji] + (self.gmhs.un - self.gmhs.profil_froz_hydro_ns[ji, jsl, jst]) * (self.gmhs.mc[ji, jsl, jst] - self.mcr[ji]))
            if self.gmhs.check_cwrr:
                self.hydrol_diag_soil_flux(jst, mclint, flux_top)
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.tmc[ji, jst] = self.gmhs.tmc[ji, jst] + self.gmhs.water2infilt[ji, jst]
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.tmc_litter[ji, jst] = self.gmhs.dz[1] * (self.gmhs.trois * self.gmhs.mcl[ji, 0, jst] + self.gmhs.mcl[ji, 1, jst]) / self.gmhs.huit
            for jsl in range(1, 4, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.gmhs.tmc_litter[ji, jst] = self.gmhs.tmc_litter[ji, jst] + self.gmhs.dz[jsl] * (self.gmhs.trois * self.gmhs.mcl[ji, jsl, jst] + self.gmhs.mcl[ji, jsl - 1, jst]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * self.gmhs.mcl[ji, jsl, jst] + self.gmhs.mcl[ji, jsl + 1, jst]) / self.gmhs.huit
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.soil_wet_litter[ji, jst] = np.minimum(self.gmhs.un, np.maximum(self.gmhs.zero, (self.gmhs.tmc_litter[ji, jst] - self.gmhs.tmc_litter_wilt[ji, jst]) / (self.gmhs.tmc_litter_field[ji, jst] - self.gmhs.tmc_litter_wilt[ji, jst])))
            sm[:, 0] = self.gmhs.dz[1] * (self.gmhs.trois * self.gmhs.mcl[:, 0, jst] + self.gmhs.mcl[:, 1, jst]) / self.gmhs.huit
            smt[:, 0] = self.gmhs.dz[1] * (self.gmhs.trois * self.gmhs.mc[:, 0, jst] + self.gmhs.mc[:, 1, jst]) / self.gmhs.huit
            smw[:, 0] = self.gmhs.dz[1] * (self.gmhs.quatre * self.mcw[:]) / self.gmhs.huit
            smf[:, 0] = self.gmhs.dz[1] * (self.gmhs.quatre * self.mcfc[:]) / self.gmhs.huit
            sms[:, 0] = self.gmhs.dz[1] * (self.gmhs.quatre * self.mcs[:]) / self.gmhs.huit
            for jsl in range(1, self.nslm - 1, 1):
                sm[:, jsl] = self.gmhs.dz[jsl] * (self.gmhs.trois * self.gmhs.mcl[:, jsl, jst] + self.gmhs.mcl[:, jsl - 1, jst]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * self.gmhs.mcl[:, jsl, jst] + self.gmhs.mcl[:, jsl + 1, jst]) / self.gmhs.huit
                smt[:, jsl] = self.gmhs.dz[jsl] * (self.gmhs.trois * self.gmhs.mc[:, jsl, jst] + self.gmhs.mc[:, jsl - 1, jst]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * self.gmhs.mc[:, jsl, jst] + self.gmhs.mc[:, jsl + 1, jst]) / self.gmhs.huit
                smw[:, jsl] = self.gmhs.dz[jsl] * (self.gmhs.quatre * self.mcw[:]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.quatre * self.mcw[:]) / self.gmhs.huit
                smf[:, jsl] = self.gmhs.dz[jsl] * (self.gmhs.quatre * self.mcfc[:]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.quatre * self.mcfc[:]) / self.gmhs.huit
                sms[:, jsl] = self.gmhs.dz[jsl] * (self.gmhs.quatre * self.mcs[:]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.quatre * self.mcs[:]) / self.gmhs.huit
            sm[:, self.nslm - 1] = self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * self.gmhs.mcl[:, self.nslm - 1, jst] + self.gmhs.mcl[:, self.nslm - 2, jst]) / self.gmhs.huit
            smt[:, self.nslm - 1] = self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * self.gmhs.mc[:, self.nslm - 1, jst] + self.gmhs.mc[:, self.nslm - 2, jst]) / self.gmhs.huit
            smw[:, self.nslm - 1] = self.gmhs.dz[self.nslm - 1] * (self.gmhs.quatre * self.mcw[:]) / self.gmhs.huit
            smf[:, self.nslm - 1] = self.gmhs.dz[self.nslm - 1] * (self.gmhs.quatre * self.mcfc[:]) / self.gmhs.huit
            sms[:, self.nslm - 1] = self.gmhs.dz[self.nslm - 1] * (self.gmhs.quatre * self.mcs[:]) / self.gmhs.huit
            for jsl in range(0, self.nslm, 1):
                sm_nostress[:, jsl] = smw[:, jsl] + self.gmhs.pcent[self.njsc[:]] * (smf[:, jsl] - smw[:, jsl])
            tmcs_litter[:] = self.gmhs.zero
            for jsl in range(0, 4, 1):
                tmcs_litter[:] = tmcs_litter[:] + sms[:, jsl]
            if jst == self.gmhs.irrig_st:
                for ji in range(0, self.kjpindex, 1):
                    self.root_deficit[ji] = np.sum(np.maximum(self.gmhs.zero, self.gmhs.beta_irrig * smf[ji, :nslm_root[ji]] - sm[ji, :nslm_root[ji]])) - self.gmhs.water2infilt[ji, jst]
                    self.root_deficit[ji] = np.maximum(self.root_deficit[ji], self.gmhs.zero)
                lai_irrig_trig[:] = 0
                for jv in range(0, self.nvm, 1):
                    if not self.gmhs.natural[jv]:
                        for ji in range(0, self.kjpindex, 1):
                            if self.veget[ji, jv] > self.veget_max[ji, jv] * (self.gmhs.un - np.exp(-(self.gmhs.lai_irrig_min * self.gmhs.ext_coeff_vegetfrac[jv]))):
                                lai_irrig_trig[ji] = lai_irrig_trig[ji] + 1
                for ji in range(0, self.kjpindex, 1):
                    if lai_irrig_trig[ji] < 1:
                        self.root_deficit[ji] = self.gmhs.zero
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.gmhs.soil_wet_ns[ji, jsl, jst] = np.minimum(self.gmhs.un, np.maximum(self.gmhs.zero, (sm[ji, jsl] - smw[ji, jsl]) / (sms[ji, jsl] - smw[ji, jsl])))
            self.gmhs.humrelv[:, 0, jst] = self.gmhs.zero
            for jv in range(1, self.nvm, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.us[ji, jv, jst, 0] = self.gmhs.zero
                    self.gmhs.humrelv[ji, jv, jst] = self.gmhs.zero
            self.hydrol_root_profile(sm, smw)
            tmp[:, :, :, :] = self.root_profile[:, :, :, :]
            for jv in range(0, self.nvm, 1):
                for jsl in range(0, self.nslm, 1):
                    if (np.sum(self.circ_class_biomass[:, jv, :, self.gmhs.iroot - 2, self.gmhs.icarbon - 2], axis=1) < self.gmhs.min_stomate).any():
                        mask = (np.sum(self.circ_class_biomass[:, jv, :, self.gmhs.iroot - 2, self.gmhs.icarbon - 2], axis=1) < self.gmhs.min_stomate)
                        tmp[:, jv, jsl, self.gmhs.istruc - 1][mask] = self.gmhs.xios_default_val
                        tmp[:, jv, jsl, self.gmhs.ifunc - 1][mask] = self.gmhs.xios_default_val
            for jsl in range(1, self.nslm, 1):
                for jv in range(1, self.nvm, 1):
                    for ji in range(0, self.kjpindex, 1):
                        if self.gmhs.new_watstress:
                            if sm[ji, jsl] - smw[ji, jsl] > self.gmhs.min_sechiba:
                                self.us[ji, jv, jst, jsl] = np.minimum(self.gmhs.un, np.maximum(self.gmhs.zero, np.exp(-(self.gmhs.alpha_watstress * ((smf[ji, jsl] - smw[ji, jsl]) / (sm_nostress[ji, jsl] - smw[ji, jsl])) * ((sm_nostress[ji, jsl] - sm[ji, jsl]) / (sm[ji, jsl] - smw[ji, jsl])))))) * self.root_profile[ji, jv, jsl, self.gmhs.ifunc - 1]
                            else:
                                self.us[ji, jv, jst, jsl] = 0.0
                        else:
                            self.us[ji, jv, jst, jsl] = np.minimum(self.gmhs.un, np.maximum(self.gmhs.zero, (sm[ji, jsl] - smw[ji, jsl]) / (sm_nostress[ji, jsl] - smw[ji, jsl]))) * self.root_profile[ji, jv, jsl, self.gmhs.ifunc - 1]
                        self.gmhs.humrelv[ji, jv, jst] = self.gmhs.humrelv[ji, jv, jst] + self.us[ji, jv, jst, jsl]
            self.gmhs.vegstressv[:, :, jst] = self.gmhs.zero
            for jv in range(1, self.nvm, 1):
                for ji in range(0, self.kjpindex, 1):
                    for jsl in range(0, self.nslm, 1):
                        self.gmhs.vegstressv[ji, jv, jst] = self.gmhs.vegstressv[ji, jv, jst] + np.minimum(self.gmhs.un, np.maximum(self.gmhs.zero, (sm[ji, jsl] - smw[ji, jsl]) / (smf[ji, jsl] - smw[ji, jsl]))) * self.root_profile[ji, jv, jsl, self.gmhs.ifunc - 1]
            for jv in range(1, self.nvm, 1):
                for ji in range(0, self.kjpindex, 1):
                    if self.gmhs.vegetmax_soil[ji, jv, jst] < self.gmhs.min_sechiba:
                        self.gmhs.humrelv[ji, jv, jst] = self.gmhs.zero
                        self.gmhs.vegstressv[ji, jv, jst] = self.gmhs.zero
                        self.us[ji, jv, jst, :] = self.gmhs.zero
            for jsl in range(0, self.nslm, 1):
                for jv in range(1, self.nvm, 1):
                    if np.any(is_under_mcr[:, jst]):
                        mask = is_under_mcr[:, jst]
                        self.us[:, jv, jst, jsl][mask] = self.gmhs.zero
            for jv in range(1, self.nvm, 1):
                if np.any(is_under_mcr[:, jst]):
                    mask = is_under_mcr[:, jst]
                    self.gmhs.humrelv[:, jv, jst][mask] = self.gmhs.zero
            for jsl in range(0, self.nslm, 1):
                if np.any(is_under_mcr[:, jst]):
                    mask = is_under_mcr[:, jst]
                    self.gmhs.soil_wet_ns[:, jsl, jst][mask] = self.gmhs.zero
            if np.any(is_under_mcr[:, jst]):
                mask = is_under_mcr[:, jst]
                self.gmhs.undermcr[mask] = self.gmhs.undermcr[mask] + self.gmhs.un
            self.mc_layh_s = self.gmhs.mc
            self.mcl_layh_s = self.gmhs.mc
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.mc_layh[ji, jsl] = self.mc_layh[ji, jsl] + self.gmhs.mc[ji, jsl, jst] * self.soiltile[ji, jst] * self.gmhs.vegtot[ji]
                    self.mcl_layh[ji, jsl] = self.mcl_layh[ji, jsl] + self.gmhs.mcl[ji, jsl, jst] * self.soiltile[ji, jst] * self.gmhs.vegtot[ji]
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.kk_moy[ji, :] = self.gmhs.kk_moy[ji, :] + self.soiltile[ji, jst] * self.gmhs.k[ji, :]
                self.gmhs.kk[ji, :, jst] = self.gmhs.k[ji, :]
            for jsl in range(0, self.nslm, 1):
                ksat[:, jsl] = ksat[:, jsl] + self.soiltile[:, jst] * (self.ks[:] * self.gmhs.kfact[jsl, :] * self.gmhs.kfact_root[:, jsl, jst])
            if self.gmhs.printlev >= 3:
                logging.info(f'prognostic/diagnostic part of hydrol_soil done for jst ={jst}')
        self.hydrol_diag_soil()
        wtd[:] = self.gmhs.zero
        ru_corr[:] = self.gmhs.zero
        ru_corr2[:] = self.gmhs.zero
        dr_corr[:] = self.gmhs.zero
        dr_corrnum[:] = self.gmhs.zero
        dr_force[:] = self.gmhs.zero
        for jst in range(0, self.nstm, 1):
            for ji in range(0, self.kjpindex, 1):
                wtd[ji] = wtd[ji] + self.soiltile[ji, jst] * wtd_ns[ji, jst]
                if self.gmhs.vegtot[ji] > self.gmhs.min_sechiba:
                    ru_corr[ji] = ru_corr[ji] + self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * ru_corr_ns[ji, jst]
                    ru_corr2[ji] = ru_corr2[ji] + self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * ru_corr2_ns[ji, jst]
                    dr_corr[ji] = dr_corr[ji] + self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * dr_corr_ns[ji, jst]
                    dr_corrnum[ji] = dr_corrnum[ji] + self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * dr_corrnum_ns[ji, jst]
                    dr_force[ji] = dr_force[ji] - self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * dr_force_ns[ji, jst]
        ru_infilt[:] = 0.0
        qinfilt[:] = 0.0
        check_infilt[:] = 0.0
        check_tr[:] = 0.0
        check_over[:] = 0.0
        check_under[:] = 0.0
        qflux[:, :] = 0.0
        check_top[:] = 0.0
        for jst in range(0, self.nstm, 1):
            for ji in range(0, self.kjpindex, 1):
                if self.gmhs.vegtot[ji] > self.gmhs.min_sechiba:
                    ru_infilt[ji] = ru_infilt[ji] + self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * ru_infilt_ns[ji, jst]
                    qinfilt[ji] = qinfilt[ji] + self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * qinfilt_ns[ji, jst]
        if self.gmhs.check_cwrr:
            for jst in range(0, self.nstm, 1):
                for ji in range(0, self.kjpindex, 1):
                    if self.gmhs.vegtot[ji] > self.gmhs.min_sechiba:
                        check_infilt[ji] = check_infilt[ji] + self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * check_infilt_ns[ji, jst]
                        check_tr[ji] = check_tr[ji] + self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * check_tr_ns[ji, jst]
                        check_over[ji] = check_over[ji] + self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * check_over_ns[ji, jst]
                        check_under[ji] = check_under[ji] + self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * check_under_ns[ji, jst]
                        qflux[ji, :] = qflux[ji, :] + self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * self.gmhs.qflux_ns[ji, :, jst]
                        check_top[ji] = check_top[ji] + self.gmhs.vegtot[ji] * self.soiltile[ji, jst] * self.gmhs.check_top_ns[ji, jst]
        self.evap_bare_lim[:] = self.gmhs.zero
        self.evap_bare_lim_ns[:, :] = self.gmhs.zero
        for jst in range(0, self.nstm, 1):
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    mcint[ji, jsl] = self.gmhs.mask_soiltile[ji, jst] * self.gmhs.mc[ji, jsl, jst]
                    mclint[ji, jsl] = self.gmhs.mask_soiltile[ji, jst] * self.gmhs.mcl[ji, jsl, jst]
            for ji in range(0, self.kjpindex, 1):
                temp[ji] = self.gmhs.tmc[ji, jst]
                tmcint[ji] = temp[ji] - self.gmhs.water2infilt[ji, jst]
            if self.gmhs.ok_freeze_cwrr:
                self.hydrol_soil_froz(jst)
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.gmhs.mcl[ji, jsl, jst] = np.minimum(self.gmhs.mc[ji, jsl, jst], self.mcr[ji] + (self.gmhs.un - self.gmhs.profil_froz_hydro_ns[ji, jsl, jst]) * (self.gmhs.mc[ji, jsl, jst] - self.mcr[ji]))
            self.hydrol_soil_coef(jst)
            self.ksoil[:, :, jst] = self.gmhs.k
            self.hydrol_soil_setup(jst)
            self.gmhs.resolv[:] = self.gmhs.mask_soiltile[:, jst] > 0
            for ji in range(0, self.kjpindex, 1):
                if self.gmhs.vegtot[ji] > self.gmhs.min_sechiba:
                    if self.gmhs.do_rsoil:
                        mc_rel[ji] = self.gmhs.tmc_litter[ji, jst] / tmcs_litter[ji]
                        r_soil_ns[ji, jst] = np.exp(8.206 - 4.255 * mc_rel[ji])
                    else:
                        r_soil_ns[ji, jst] = self.gmhs.zero
                    speed = np.maximum(self.gmhs.min_wind, np.sqrt(self.u[ji] * self.u[ji] + self.v[ji] * self.v[ji]))
                    if speed * self.tq_cdrag[ji] > self.gmhs.min_sechiba:
                        ra = self.gmhs.un / (speed * self.tq_cdrag[ji])
                        evap_soil[ji] = self.evapot_penm[ji] / (self.gmhs.un + r_soil_ns[ji, jst] / ra)
                    else:
                        evap_soil[ji] = self.evapot_penm[ji]
                    flux_top[ji] = evap_soil[ji] * np.trunc(self.gmhs.frac_bare_ns[ji, jst] + self.gmhs.un - self.gmhs.min_sechiba)
                else:
                    flux_top[ji] = self.gmhs.zero
                    r_soil_ns[ji, jst] = self.gmhs.zero
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.tmat[ji, 0, 0] = self.gmhs.zero
                self.gmhs.tmat[ji, 0, 1] = self.gmhs.f[ji, 0]
                self.gmhs.tmat[ji, 0, 2] = self.gmhs.g1[ji, 0]
                self.gmhs.rhs[ji, 0] = self.gmhs.fp[ji, 0] * self.gmhs.mcl[ji, 0, jst] + self.gmhs.gp[ji, 0] * self.gmhs.mcl[ji, 1, jst] - flux_top[ji] - (self.gmhs.b[ji, 0] + self.gmhs.b[ji, 1]) / self.gmhs.deux * (self.gmhs.dt_sechiba / self.gmhs.one_day)
            for jsl in range(1, self.nslm - 1, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.gmhs.tmat[ji, jsl, 0] = self.gmhs.e[ji, jsl]
                    self.gmhs.tmat[ji, jsl, 1] = self.gmhs.f[ji, jsl]
                    self.gmhs.tmat[ji, jsl, 2] = self.gmhs.g1[ji, jsl]
                    self.gmhs.rhs[ji, jsl] = self.gmhs.ep[ji, jsl] * self.gmhs.mcl[ji, jsl - 1, jst] + self.gmhs.fp[ji, jsl] * self.gmhs.mcl[ji, jsl, jst] + self.gmhs.gp[ji, jsl] * self.gmhs.mcl[ji, jsl + 1, jst] + (self.gmhs.b[ji, jsl - 1] - self.gmhs.b[ji, jsl + 1]) * (self.gmhs.dt_sechiba / self.gmhs.one_day) / self.gmhs.deux
            for ji in range(0, self.kjpindex, 1):
                jsl = self.nslm - 1
                self.gmhs.tmat[ji, jsl, 0] = self.gmhs.e[ji, jsl]
                self.gmhs.tmat[ji, jsl, 1] = self.gmhs.f[ji, jsl]
                self.gmhs.tmat[ji, jsl, 2] = self.gmhs.zero
                self.gmhs.rhs[ji, jsl] = self.gmhs.ep[ji, jsl] * self.gmhs.mcl[ji, jsl - 1, jst] + self.gmhs.fp[ji, jsl] * self.gmhs.mcl[ji, jsl, jst] + (self.gmhs.b[ji, jsl - 1] + self.gmhs.b[ji, jsl] * (self.gmhs.un - self.gmhs.deux * self.gmhs.free_drain_coef[ji, jst])) * (self.gmhs.dt_sechiba / self.gmhs.one_day) / self.gmhs.deux
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.gmhs.srhs[ji, jsl] = self.gmhs.rhs[ji, jsl]
                    self.gmhs.stmat[ji, jsl, 0] = self.gmhs.tmat[ji, jsl, 0]
                    self.gmhs.stmat[ji, jsl, 1] = self.gmhs.tmat[ji, jsl, 1]
                    self.gmhs.stmat[ji, jsl, 2] = self.gmhs.tmat[ji, jsl, 2]
            self.hydrol_soil_tridiag(jst)
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.resolv[ji] = self.gmhs.mcl[ji, 0, jst] < self.mcr[ji] and flux_top[ji] > self.gmhs.min_sechiba
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.gmhs.rhs[ji, jsl] = self.gmhs.srhs[ji, jsl]
                    self.gmhs.tmat[ji, jsl, 0] = self.gmhs.stmat[ji, jsl, 0]
                    self.gmhs.tmat[ji, jsl, 1] = self.gmhs.stmat[ji, jsl, 1]
                    self.gmhs.tmat[ji, jsl, 2] = self.gmhs.stmat[ji, jsl, 2]
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.tmat[ji, 0, 1] = self.gmhs.un
                self.gmhs.tmat[ji, 0, 2] = self.gmhs.zero
                self.gmhs.rhs[ji, 0] = self.mcr[ji]
            self.hydrol_soil_tridiag(jst)
            for ji in range(0, self.kjpindex, 1):
                flux_bottom[ji] = self.gmhs.mask_soiltile[ji, jst] * self.gmhs.k[ji, self.nslm - 1] * self.gmhs.free_drain_coef[ji, jst] * (self.gmhs.dt_sechiba / self.gmhs.one_day)
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.gmhs.mc[ji, jsl, jst] = np.maximum(self.gmhs.mcl[ji, jsl, jst], self.gmhs.mcl[ji, jsl, jst] + self.gmhs.profil_froz_hydro_ns[ji, jsl, jst] * (self.gmhs.mc[ji, jsl, jst] - self.mcr[ji]))
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.tmc[ji, jst] = self.gmhs.dz[1] * (self.gmhs.trois * self.gmhs.mc[ji, 0, jst] + self.gmhs.mc[ji, 1, jst]) / self.gmhs.huit
            for jsl in range(1, self.nslm - 1, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.gmhs.tmc[ji, jst] = self.gmhs.tmc[ji, jst] + self.gmhs.dz[jsl] * (self.gmhs.trois * self.gmhs.mc[ji, jsl, jst] + self.gmhs.mc[ji, jsl - 1, jst]) / self.gmhs.huit + self.gmhs.dz[jsl + 1] * (self.gmhs.trois * self.gmhs.mc[ji, jsl, jst] + self.gmhs.mc[ji, jsl + 1, jst]) / self.gmhs.huit
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.tmc[ji, jst] = self.gmhs.tmc[ji, jst] + self.gmhs.dz[self.nslm - 1] * (self.gmhs.trois * self.gmhs.mc[ji, self.nslm - 1, jst] + self.gmhs.mc[ji, self.nslm - 2, jst]) / self.gmhs.huit
            for ji in range(0, self.kjpindex, 1):
                self.evap_bare_lim_ns[ji, jst] = self.gmhs.mask_soiltile[ji, jst] * (tmcint[ji] - self.gmhs.tmc[ji, jst] - flux_bottom[ji] - np.sum(self.gmhs.rootsink[ji, :, jst]))
            for ji in range(0, self.kjpindex, 1):
                if self.gmhs.vegtot[ji] > self.gmhs.min_sechiba:
                    self.evap_bare_lim_ns[ji, jst] = self.evap_bare_lim_ns[ji, jst] * self.gmhs.frac_bare_ns[ji, jst]
                else:
                    self.evap_bare_lim_ns[ji, jst] = 0.0
            if self.gmhs.do_rsoil:
                for ji in range(0, self.kjpindex, 1):
                    if self.evapot[ji] > self.gmhs.min_sechiba:
                        self.evap_bare_lim_ns[ji, jst] = self.evap_bare_lim_ns[ji, jst] / self.evapot[ji]
                    else:
                        self.evap_bare_lim_ns[ji, jst] = self.gmhs.zero
                    self.evap_bare_lim_ns[ji, jst] = np.maximum(np.minimum(self.evap_bare_lim_ns[ji, jst], 1.0), 0.0)
            else:
                for ji in range(0, self.kjpindex, 1):
                    if self.evapot[ji] > self.gmhs.min_sechiba and self.gmhs.tmc_litter[ji, jst] > self.gmhs.tmc_litter_wilt[ji, jst]:
                        self.evap_bare_lim_ns[ji, jst] = self.evap_bare_lim_ns[ji, jst] / self.evapot[ji]
                    elif self.evapot[ji] > self.gmhs.min_sechiba and self.gmhs.tmc_litter[ji, jst] > self.gmhs.tmc_litter_res[ji, jst]:
                        self.evap_bare_lim_ns[ji, jst] = self.gmhs.un / self.gmhs.deux * self.evap_bare_lim_ns[ji, jst] / self.evapot[ji]
                    else:
                        self.evap_bare_lim_ns[ji, jst] = self.gmhs.zero
                    self.evap_bare_lim_ns[ji, jst] = np.maximum(np.minimum(self.evap_bare_lim_ns[ji, jst], 1.0), 0.0)
            if np.any(is_under_mcr[:, jst]):
                mask = is_under_mcr[:, jst]
                self.evap_bare_lim_ns[:, jst][mask] = self.gmhs.zero
            for jsl in range(0, self.nslm, 1):
                for ji in range(0, self.kjpindex, 1):
                    self.gmhs.mc[ji, jsl, jst] = self.gmhs.mask_soiltile[ji, jst] * mcint[ji, jsl]
                    self.gmhs.mcl[ji, jsl, jst] = self.gmhs.mask_soiltile[ji, jst] * mclint[ji, jsl]
            for ji in range(0, self.kjpindex, 1):
                self.gmhs.tmc[ji, jst] = temp[ji]
        for ji in range(0, self.kjpindex, 1):
            self.evap_bare_lim[ji] = np.sum(self.evap_bare_lim_ns[ji, :] * self.gmhs.vegtot[ji] * self.soiltile[ji, :])
            r_soil[ji] = np.sum(r_soil_ns[ji, :] * self.gmhs.vegtot[ji] * self.soiltile[ji, :])

    def read_dummy(self):
        print(f'--- inside the read dummy routine for hydrol_soil ---')
        path = f'/home/ssivanes/Fgpt/benchmark/hydrol_soil/dummy.bin'
        ffile = FortranFile(path, 'r')
        for var in [self.altmax, self.avan, self.circ_class_biomass, self.e_frac, self.evap_bare_lim, self.evap_bare_lim_ns, self.evapot, self.evapot_penm, self.F_absorption, self.frac_snow_nobio, self.humrel, self.irrigation, self.ks, self.mcfc, self.mcr, self.mcs, self.mcw, self.njsc, self.nvan, self.precip_rain, self.reinf_slope_soil, self.reinfiltration, self.returnflow, self.snow, self.snowdz, self.soiltile, self.stempdiag, self.tot_bare_soil, self.tot_melt, self.totfrac_nobio, self.tq_cdrag, self.transpir, self.u, self.us, self.v, self.veget, self.veget_max, self.vevapnu]:
            if isinstance(var, np.ndarray):
                arr_shape = var.shape
                if var.dtype == np.float64:
                    data = ffile.read_reals(np.float64)
                elif var.dtype == np.int32 or var.dtype == np.bool:
                    data = ffile.read_ints(np.int32)
                if data.size != np.prod(arr_shape):
                    continue
                var[:] = data.reshape(arr_shape, order='F')

    def test_hydrol_soil(self):
        print('--- inside the test function for hydrol_soil ---')
        path = f'/home/ssivanes/Fgpt/benchmark/hydrol_soil/output.bin'
        ffile = FortranFile(path, 'r')
        modif_var = ['ae_ns', 'dr_ns', 'evap_bare_lim', 'evap_bare_lim_ns', 'humrelv', 'kfact_root', 'kk', 'kk_moy', 'ksoil', 'mc', 'mc_layh', 'mc_layh_s', 'mcl', 'mcl_layh', 'mcl_layh_s', 'profil_froz_hydro_ns', 'qflux_ns', 'resolv', 'rhs', 'root_deficit', 'ru_ns', 'soil_wet_litter', 'soil_wet_ns', 'srhs', 'stmat', 'tmat', 'tmc', 'tmc_litter', 'undermcr', 'us', 'vegstressv', 'water2infilt']
        for (variable, value) in zip(modif_var, [self.gmhs.ae_ns, self.gmhs.dr_ns, self.evap_bare_lim, self.evap_bare_lim_ns, self.gmhs.humrelv, self.gmhs.kfact_root, self.gmhs.kk, self.gmhs.kk_moy, self.ksoil, self.gmhs.mc, self.mc_layh, self.mc_layh_s, self.gmhs.mcl, self.mcl_layh, self.mcl_layh_s, self.gmhs.profil_froz_hydro_ns, self.gmhs.qflux_ns, self.gmhs.resolv, self.gmhs.rhs, self.root_deficit, self.gmhs.ru_ns, self.gmhs.soil_wet_litter, self.gmhs.soil_wet_ns, self.gmhs.srhs, self.gmhs.stmat, self.gmhs.tmat, self.gmhs.tmc, self.gmhs.tmc_litter, self.gmhs.undermcr, self.us, self.gmhs.vegstressv, self.gmhs.water2infilt]):
            fortran_value = None
            if isinstance(value, np.ndarray):
                arr_shape = value.shape
                if value.dtype == np.float64:
                    fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
                elif value.dtype == np.int32 or value.dtype == np.bool:
                    fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
                if np.allclose(value, fortran_value):
                    print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
                else:
                    print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                    print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                    print(f' Min and max of python value: {value.min()}, {value.max()}')
                    print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
            elif not isinstance(value, np.ndarray):
                if value.dtype == np.bool or value.dtype == np.int32:
                    fortran_value = ffile.read_ints(np.int32)[-1]
                elif value.dtype == np.float64:
                    fortran_value = ffile.read_reals(np.float64)[-1]
                if np.isclose(value, fortran_value):
                    print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
                else:
                    print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    hs = Hydrol_soil()
    hs.read_dummy()
    hs.hydrol_soil()
    hs.test_hydrol_soil()
if __name__ == '__main__':
    main()
```

```python
import numpy as np
import os
from scipy.io import FortranFile

class Global_module_explicitsnow_fall:

    def __init__(self):
        self.nsnow = np.int32(3)
        self.nslm = np.int32(11)
        self.nvm = np.int32(15)
        self.nstm = np.int32(3)
        self.kjpindex = np.int32(4717)
        self.ier = np.int32(0)
        self.ic0 = np.int32(0)
        self.ic = np.int32(0)
        self.icr = np.float64(0.0)
        self.start_time = np.float64(0.0)
        self.stop_time = np.float64(0.0)
        self.zero = np.float64(0.0)
        self.snowfall_b_sn = np.float64(6.0)
        self.chalsu0 = np.float64(2834500.0)
        self.chalev0 = np.float64(2500800.0)
        self.xrhosmin = np.float64(50.0)
        self.tp_00 = np.float64(273.15)
        self.snowfall_c_sn = np.float64(26.0)
        self.min_wind = np.float64(0.1)
        self.dgrain_new_max = np.float64(0.0002)
        self.snowfall_a_sn = np.float64(109.0)
        self.xci = np.float64(2106.0)
        self.chalfu0 = np.float64(self.chalsu0 - self.chalev0)
```

```python
import os
import time
import functools
import numpy as np
import logging
from scipy.io import FortranFile
# from module_global import Global_module_explicitsnow_fall

def snow3lgrain_0d(PSNOWRHO):
    ZSNOWRAD_AGRAIN = np.float64(0.00016)
    ZSNOWRAD_BGRAIN = np.float64(1.1e-13)
    ZDSGRAIN_MAX = np.float64(0.002796)
    PDSGRAIN = ZSNOWRAD_AGRAIN + ZSNOWRAD_BGRAIN * PSNOWRHO ** 4
    PDSGRAIN = np.minimum(ZDSGRAIN_MAX, PDSGRAIN)
    return PDSGRAIN

def read_dummy(precip_snow, snowdz, snowgrain, snowheat, snowrho, snowtemp, temp_air, u, v):
    print(f'--- inside the read dummy routine for explicitsnow_fall ---')
    path = f'/home/ssivanes/Fgpt/benchmark/explicitsnow_fall/dummy.bin'
    ffile = FortranFile(path, 'r')
    for var in [precip_snow, snowdz, snowgrain, snowheat, snowrho, snowtemp, temp_air, u, v]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                data = ffile.read_reals(np.float64)
            elif var.dtype == np.int32 or var.dtype == np.bool:
                data = ffile.read_ints(np.int32)
            if data.size != np.prod(arr_shape):
                continue
            var[:] = data.reshape(arr_shape, order='F')

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        path = '/home/ssivanes/Fgpt/benchmark/explicitsnow_fall/time.txt'
        with open(path, 'a') as f:
            f.write(f"[TIMER] Python: '{func.__name__}' executed in {duration:.6f} seconds \n")
        return result
    return wrapper

@timer
def explicitsnow_fall(precip_snow, temp_air, u, v, snowrho, snowdz, snowheat, snowgrain, snowtemp, psnowhmass, gmef):
    precip_snow = np.zeros((gmef.kjpindex,), dtype=np.float64)
    temp_air = np.zeros((gmef.kjpindex,), dtype=np.float64)
    v = np.zeros((gmef.kjpindex,), dtype=np.float64)
    u = np.zeros((gmef.kjpindex,), dtype=np.float64)
    snowtemp = np.zeros((gmef.kjpindex, gmef.nsnow), dtype=np.float64)
    psnowhmass = np.zeros((gmef.kjpindex,), dtype=np.float64)
    snowrho = np.zeros((gmef.kjpindex, gmef.nsnow), dtype=np.float64)
    snowdz = np.zeros((gmef.kjpindex, gmef.nsnow), dtype=np.float64)
    snowheat = np.zeros((gmef.kjpindex, gmef.nsnow), dtype=np.float64)
    snowgrain = np.zeros((gmef.kjpindex, gmef.nsnow), dtype=np.float64)
    rhosnew = np.zeros((gmef.kjpindex,), dtype=np.float64)
    dsnowfall = np.zeros((gmef.kjpindex,), dtype=np.float64)
    snowdz_old = np.zeros((gmef.kjpindex, gmef.nsnow), dtype=np.float64)
    newgrain = np.zeros((gmef.kjpindex,), dtype=np.float64)
    snow_depth_old = np.zeros((gmef.kjpindex,), dtype=np.float64)
    snow_depth = np.zeros((gmef.kjpindex,), dtype=np.float64)
    snowdz_old = snowdz
    for ji in range(0, gmef.kjpindex, 1):
        snow_depth[ji] = np.sum(snowdz[ji, :])
    snow_depth_old = snow_depth
    snowfall_delta = 0.0
    for ji in range(0, gmef.kjpindex, 1):
        speed = np.maximum(gmef.min_wind, np.sqrt(u[ji] * u[ji] + v[ji] * v[ji]))
        psnowhmass[ji] = 0.0
        if precip_snow[ji] > 0.0:
            psnowhmass[ji] = precip_snow[ji] * (gmef.xci * (temp_air[ji] - gmef.tp_00) - gmef.chalfu0)
            rhosnew[ji] = np.maximum(gmef.xrhosmin, gmef.snowfall_a_sn + gmef.snowfall_b_sn * (temp_air[ji] - gmef.tp_00) + gmef.snowfall_c_sn * np.sqrt(speed))
            dsnowfall[ji] = precip_snow[ji] / rhosnew[ji]
            snow_depth[ji] = snow_depth[ji] + dsnowfall[ji]
            if dsnowfall[ji] > gmef.zero:
                snowrho[ji, 0] = (snowdz[ji, 0] * snowrho[ji, 0] + dsnowfall[ji] * rhosnew[ji]) / (snowdz[ji, 0] + dsnowfall[ji])
                snowdz[ji, 0] = snowdz[ji, 0] + dsnowfall[ji]
                snowheat[ji, 0] = snowheat[ji, 0] + psnowhmass[ji]
                newgrain[ji] = np.minimum(gmef.dgrain_new_max, snow3lgrain_0d(int(rhosnew[ji]) - 1))
                snowgrain[ji, 0] = (snowdz_old[ji, 0] * snowgrain[ji, 0] + dsnowfall[ji] * newgrain[ji]) / snowdz[ji, 0]
        else:
            dsnowfall[ji] = 0.0
        if dsnowfall[ji] > gmef.zero and snow_depth_old[ji] == gmef.zero:
            snowfall_delta = 1.0
            for jj in range(0, gmef.nsnow, 1):
                snowdz[ji, jj] = snowfall_delta * (dsnowfall[ji] / gmef.nsnow) + (1.0 - snowfall_delta) * snowdz[ji, jj]
                snowheat[ji, jj] = snowfall_delta * (psnowhmass[ji] / gmef.nsnow) + (1.0 - snowfall_delta) * snowheat[ji, jj]
                snowrho[ji, jj] = snowfall_delta * rhosnew[ji] + (1.0 - snowfall_delta) * snowrho[ji, jj]
                snowgrain[ji, jj] = snowfall_delta * newgrain[ji] + (1.0 - snowfall_delta) * snowgrain[ji, jj]

def test_explicitsnow_fall(snowdz, snowgrain, snowrho, snowheat, psnowhmass):
    print('--- inside the test function for explicitsnow_fall ---')
    path = f'/home/ssivanes/Fgpt/benchmark/explicitsnow_fall/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['psnowhmass', 'snowdz', 'snowgrain', 'snowheat', 'snowrho']
    for (variable, value) in zip(modif_var, [psnowhmass, snowdz, snowgrain, snowheat, snowrho]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32 or value.dtype == np.bool:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gmef = Global_module_explicitsnow_fall()
    # gmef.declaration_initialization()
    precip_snow = np.zeros((gmef.kjpindex,), dtype=np.float64)
    psnowhmass = np.zeros((gmef.kjpindex,), dtype=np.float64)
    snowdz = np.zeros((gmef.kjpindex, gmef.nsnow), dtype=np.float64)
    snowgrain = np.zeros((gmef.kjpindex, gmef.nsnow), dtype=np.float64)
    snowheat = np.zeros((gmef.kjpindex, gmef.nsnow), dtype=np.float64)
    snowrho = np.zeros((gmef.kjpindex, gmef.nsnow), dtype=np.float64)
    snowtemp = np.zeros((gmef.kjpindex, gmef.nsnow), dtype=np.float64)
    temp_air = np.zeros((gmef.kjpindex,), dtype=np.float64)
    u = np.zeros((gmef.kjpindex,), dtype=np.float64)
    v = np.zeros((gmef.kjpindex,), dtype=np.float64)
    read_dummy(precip_snow, snowdz, snowgrain, snowheat, snowrho, snowtemp, temp_air, u, v)
    explicitsnow_fall(precip_snow, temp_air, u, v, snowrho, snowdz, snowheat, snowgrain, snowtemp, psnowhmass, gmef)
    test_explicitsnow_fall(snowdz, snowgrain, snowrho, snowheat, psnowhmass)
if __name__ == '__main__':
    main()
```

```python
import numpy as np
import os
from scipy.io import FortranFile

class Global_module_snow3lhold_1d:

    def __init__(self):
        self.nsnow = np.int32(3)
        self.nslm = np.int32(11)
        self.nvm = np.int32(15)
        self.nstm = np.int32(3)
        self.kjpindex = np.int32(4717)
        self.ier = np.int32(0)
        self.ic0 = np.int32(0)
        self.ic = np.int32(0)
        self.icr = np.float64(0.0)
        self.start_time = np.float64(0.0)
        self.stop_time = np.float64(0.0)
        self.xwsnowholdmax1 = np.float64(0.03)
        self.xsnowrhohold = np.float64(200.0)
        self.ph2o = np.float64(1000.0)
        self.xrhosmax = np.float64(750.0)
        self.xwsnowholdmax2 = np.float64(0.1)
```

```python
import os
import time
import functools
import numpy as np
import logging
from scipy.io import FortranFile
# from module_global import Global_module_snow3lhold_1d

def read_dummy(PSNOWDZ, PSNOWRHO):
    print(f'--- inside the read dummy routine for snow3lhold_1d ---')
    path = f'/home/ssivanes/Fgpt/benchmark/snow3lhold_1d/dummy.bin'
    ffile = FortranFile(path, 'r')
    for var in [PSNOWDZ, PSNOWRHO]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                data = ffile.read_reals(np.float64)
            elif var.dtype == np.int32 or var.dtype == np.bool:
                data = ffile.read_ints(np.int32)
            if data.size != np.prod(arr_shape):
                continue
            var[:] = data.reshape(arr_shape, order='F')

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        path = '/home/ssivanes/Fgpt/benchmark/snow3lhold_1d/time.txt'
        with open(path, 'a') as f:
            f.write(f"[TIMER] Python: '{func.__name__}' executed in {duration:.6f} seconds \n")
        return result
    return wrapper

@timer
def snow3lhold_1d(PSNOWRHO, PSNOWDZ,gms1):
    PWHOLDMAX = np.zeros(((gms1.nsnow - 1 + 1) * 1,), dtype=np.float64)
    ZSNOWRHO = np.zeros(((gms1.nsnow - 1 + 1) * 1,), dtype=np.float64)
    ZHOLDMAXR = np.zeros(((gms1.nsnow - 1 + 1) * 1,), dtype=np.float64)
    ZSNOWRHO[:] = np.minimum(gms1.xrhosmax, PSNOWRHO[:])
    ZHOLDMAXR[:] = gms1.xwsnowholdmax1 + (gms1.xwsnowholdmax2 - gms1.xwsnowholdmax1) * np.maximum(0.0, gms1.xsnowrhohold - ZSNOWRHO[:]) / gms1.xsnowrhohold
    PWHOLDMAX[:] = ZHOLDMAXR[:] * PSNOWDZ[:] * ZSNOWRHO[:] / gms1.ph2o
    PWHOLDMAX[:][ZSNOWRHO[:] >= gms1.xrhosmax] = 0.0
    return PWHOLDMAX

def test_snow3lhold_1d(PWHOLDMAX):
    print('--- inside the test function for snow3lhold_1d ---')
    path = f'/home/ssivanes/Fgpt/benchmark/snow3lhold_1d/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['PWHOLDMAX']
    for (variable, value) in zip(modif_var, [PWHOLDMAX]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32 or value.dtype == np.bool:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gms1 = Global_module_snow3lhold_1d()
    PSNOWDZ = np.zeros((gms1.nsnow - 1 + 1,), dtype=np.float64)
    PSNOWRHO = np.zeros((gms1.nsnow - 1 + 1,), dtype=np.float64)
    PWHOLDMAX = np.zeros(((gms1.nsnow - 1 + 1) * 1,), dtype=np.float64)
    read_dummy(PSNOWDZ, PSNOWRHO)
    PWHOLDMAX = snow3lhold_1d(PSNOWRHO, PSNOWDZ,gms1)
    test_snow3lhold_1d(PWHOLDMAX)
if __name__ == '__main__':
    main()
```

```python
import numpy as np
import os
from scipy.io import FortranFile

class Global_module_explicitsnow_compactn:

    def __init__(self):
        self.nsnow = np.int32(3)
        self.nslm = np.int32(11)
        self.nvm = np.int32(15)
        self.nstm = np.int32(3)
        self.kjpindex = np.int32(4717)
        self.ier = np.int32(0)
        self.ic0 = np.int32(0)
        self.ic = np.int32(0)
        self.icr = np.float64(0.0)
        self.start_time = np.float64(0.0)
        self.stop_time = np.float64(0.0)
        self.ZSNOWCMPCT_V0 = np.float64(37000000.0)
        self.xrhosmax = np.float64(750.0)
        self.ZSNOWCMPCT_VR = np.float64(0.018)
        self.ZSNOWCMPCT_ACM = np.float64(2.8e-06)
        self.tp_00 = np.float64(273.15)
        self.ZSNOWCMPCT_CCM = np.float64(460.0)
        self.ZSNOWCMPCT_W0 = np.float64(20000000.0)
        self.ZSNOWCMPCT_WT = np.float64(0.35)
        self.cte_grav = np.float64(9.80665)
        self.ZSNOWCMPCT_VT = np.float64(0.081)
        self.ZSNOWCMPCT_BCM = np.float64(0.04)
        self.ZSNOWCMPCT_RHOD = np.float64(150.0)
        self.ZSNOWCMPCT_WR = np.float64(0.01)
        self.dt_sechiba = np.float64(0)

    def declaration_initialization(self):
        print(f'--- add the declaration and initialization in module global ---')
        path = f'/home/ssivanes/Fgpt/benchmark/explicitsnow_compactn/global.bin'
        ffile = FortranFile(path, 'r')
        self.dt_sechiba = ffile.read_reals(np.float64)[0]
        for attr_name in []:
            attribute = getattr(self, attr_name)
            if isinstance(attribute, np.ndarray):
                arr_shape = attribute.shape
                if attribute.dtype == np.float64:
                    data = ffile.read_reals(np.float64)
                elif attribute.dtype == np.int32 or attribute.dtype == np.bool:
                    data = ffile.read_ints(np.int32)
                if data.size != np.prod(arr_shape):
                    continue
                value = data.reshape(arr_shape, order='F')
                setattr(self, attr_name, value)
```

```python
import os
import time
import functools
import numpy as np
import logging
from scipy.io import FortranFile
# from module_global import Global_module_explicitsnow_compactn

def read_dummy(snowdz, snowrho, snowtemp):
    print(f'--- inside the read dummy routine for explicitsnow_compactn ---')
    path = f'/home/ssivanes/Fgpt/benchmark/explicitsnow_compactn/dummy.bin'
    ffile = FortranFile(path, 'r')
    for var in [snowdz, snowrho, snowtemp]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                data = ffile.read_reals(np.float64)
            elif var.dtype == np.int32 or var.dtype == np.bool:
                data = ffile.read_ints(np.int32)
            if data.size != np.prod(arr_shape):
                continue
            var[:] = data.reshape(arr_shape, order='F')

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        path = '/home/ssivanes/Fgpt/benchmark/explicitsnow_compactn/time.txt'
        with open(path, 'a') as f:
            f.write(f"[TIMER] Python: '{func.__name__}' executed in {duration:.6f} seconds \n")
        return result
    return wrapper

def explicitsnow_compactn(snowtemp, snowrho, snowdz, gmec):
    zsettle = np.zeros((gmec.kjpindex, gmec.nsnow), dtype=np.float64)
    zviscocity = np.zeros((gmec.kjpindex, gmec.nsnow), dtype=np.float64)
    zsnowrho2 = np.zeros((gmec.kjpindex, gmec.nsnow), dtype=np.float64)
    zsmass = np.zeros((gmec.kjpindex, gmec.nsnow), dtype=np.float64)
    zwsnowdz = np.zeros((gmec.kjpindex, gmec.nsnow), dtype=np.float64)
    zsmassc = np.zeros((gmec.kjpindex,), dtype=np.float64)
    snowdepth_crit = np.zeros((gmec.kjpindex,), dtype=np.float64)
    zsnowrho2[:,:] = snowrho
    zsettle[:, :] = gmec.ZSNOWCMPCT_ACM
    zviscocity[:, :] = gmec.ZSNOWCMPCT_V0
    for ji in range(0, gmec.kjpindex, 1):
        if np.sum(snowdz[ji, :]) > 0.0:
            zwsnowdz[ji, :] = snowdz[ji, :] * snowrho[ji, :]
            zsmassc[ji] = 0.0
            for jj in range(0, gmec.nsnow, 1):
                zsmass[ji, jj] = zsmassc[ji] + zwsnowdz[ji, jj]
                zsmassc[ji] = zsmassc[ji] + zwsnowdz[ji, jj]
            for jj in range(0, gmec.nsnow, 1):
                if snowrho[ji, jj] < gmec.xrhosmax:
                    zsettle[ji, jj] = gmec.ZSNOWCMPCT_ACM * np.exp(-(gmec.ZSNOWCMPCT_BCM * (gmec.tp_00 - np.minimum(gmec.tp_00, snowtemp[ji, jj]))) - gmec.ZSNOWCMPCT_CCM * np.maximum(0.0, snowrho[ji, jj] - gmec.ZSNOWCMPCT_RHOD))
                    if snowtemp[ji, jj] > 264.0:
                        zviscocity[ji, jj] = gmec.ZSNOWCMPCT_W0 * np.exp(gmec.ZSNOWCMPCT_WT * (gmec.tp_00 - np.minimum(gmec.tp_00, snowtemp[ji, jj])) + gmec.ZSNOWCMPCT_WR * snowrho[ji, jj])
                    else:
                        zviscocity[ji, jj] = gmec.ZSNOWCMPCT_V0 * np.exp(gmec.ZSNOWCMPCT_VT * (gmec.tp_00 - np.minimum(gmec.tp_00, snowtemp[ji, jj])) + gmec.ZSNOWCMPCT_VR * snowrho[ji, jj])
                    zsnowrho2[ji, jj] = snowrho[ji, jj] + snowrho[ji, jj] * gmec.dt_sechiba * (gmec.cte_grav * zsmass[ji, jj] / zviscocity[ji, jj] + zsettle[ji, jj])
                    snowdz[ji, jj] = snowdz[ji, jj] * (snowrho[ji, jj] / zsnowrho2[ji, jj])
            snowrho[ji, :] = zsnowrho2[ji, :]

def test_explicitsnow_compactn(snowrho, snowdz):
    print('--- inside the test function for explicitsnow_compactn ---')
    path = f'/home/ssivanes/Fgpt/benchmark/explicitsnow_compactn/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['snowdz', 'snowrho']
    for (variable, value) in zip(modif_var, [snowdz, snowrho]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32 or value.dtype == np.bool:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gmec = Global_module_explicitsnow_compactn()
    gmec.declaration_initialization()
    snowdz = np.zeros((gmec.kjpindex, gmec.nsnow), dtype=np.float64)
    snowrho = np.zeros((gmec.kjpindex, gmec.nsnow), dtype=np.float64)
    snowtemp = np.zeros((gmec.kjpindex, gmec.nsnow), dtype=np.float64)
    read_dummy(snowdz, snowrho, snowtemp)
    explicitsnow_compactn(snowtemp, snowrho, snowdz, gmec)
    test_explicitsnow_compactn(snowrho, snowdz)
if __name__ == '__main__':
    main()
```

```python
import numpy as np
import os
from scipy.io import FortranFile

class Global_module_explicitsnow_transf:

    def __init__(self):
        self.nsnow = np.int32(3)
        self.nslm = np.int32(11)
        self.nvm = np.int32(15)
        self.nstm = np.int32(3)
        self.kjpindex = np.int32(4717)
        self.ier = np.int32(0)
        self.ic0 = np.int32(0)
        self.ic = np.int32(0)
        self.icr = np.float64(0.0)
        self.start_time = np.float64(0.0)
        self.stop_time = np.float64(0.0)
        self.xsnowcritd = np.float64(0.03)
```

```python
import os
import time
import functools
import numpy as np
import logging
from scipy.io import FortranFile
# from module_global import Global_module_explicitsnow_transf

def read_dummy(snowdz, snowdz_old, snowgrain, snowheat, snowrho):
    print(f'--- inside the read dummy routine for explicitsnow_transf ---')
    path = f'/home/ssivanes/Fgpt/benchmark/explicitsnow_transf/dummy.bin'
    ffile = FortranFile(path, 'r')
    for var in [snowdz, snowdz_old, snowgrain, snowheat, snowrho]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                data = ffile.read_reals(np.float64)
            elif var.dtype == np.int32 or var.dtype == np.bool:
                data = ffile.read_ints(np.int32)
            if data.size != np.prod(arr_shape):
                continue
            var[:] = data.reshape(arr_shape, order='F')

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        path = '/home/ssivanes/Fgpt/benchmark/explicitsnow_transf/time.txt'
        with open(path, 'a') as f:
            f.write(f"[TIMER] Python: '{func.__name__}' executed in {duration:.6f} seconds \n")
        return result
    return wrapper

def explicitsnow_transf(snowdz_old, snowdz, snowrho, snowheat, snowgrain, gmet):
    zsnowzo = np.zeros((gmet.kjpindex, gmet.nsnow - 0 + 1), dtype=np.float64)
    zsnowzn = np.zeros((gmet.kjpindex, gmet.nsnow - 0 + 1), dtype=np.float64)
    zsnowddz = np.zeros((gmet.kjpindex, gmet.nsnow), dtype=np.float64)
    zdelta = np.zeros((gmet.kjpindex, gmet.nsnow), dtype=np.float64)
    zsnowgrainn = np.zeros((gmet.kjpindex, gmet.nsnow), dtype=np.float64)
    zsnowheatn = np.zeros((gmet.kjpindex, gmet.nsnow), dtype=np.float64)
    zsnowrhon = np.zeros((gmet.kjpindex, gmet.nsnow), dtype=np.float64)
    zsnowmix_delta = np.zeros((gmet.kjpindex,), dtype=np.float64)
    zsumgrain = np.zeros((gmet.kjpindex,), dtype=np.float64)
    zsumswe = np.zeros((gmet.kjpindex,), dtype=np.float64)
    zsumheat = np.zeros((gmet.kjpindex,), dtype=np.float64)
    locflag = np.zeros((gmet.nsnow, 2), dtype=np.int32)
    zsumheat[:] = 0.0
    zsumswe[:] = 0.0
    zsumgrain[:] = 0.0
    zsnowmix_delta[:] = 0.0
    locflag[:, :] = 0
    #zsnowrhon[:, :] = 0.0
    #zsnowheatn[:, :] = 0.0
    #zsnowgrainn[:, :] = 0.0
    for ji in range(13, 14, 1): # 13, 14, 1 0, gmet.kjpindex, 1
        psnow = np.sum(snowdz[ji, :])
        print(f'psnow ave :{psnow}')
        if ((psnow >= gmet.xsnowcritd and snowdz_old[ji, 0] != 0) and snowdz_old[ji, 1] != 0) and snowdz_old[ji, 2] != 0:
            zsnowzo[ji, 0] = 0.0
            zsnowzn[ji, 0] = 0.0
            zsnowzo[ji, 1] = snowdz_old[ji, 0]
            zsnowzn[ji, 1] = snowdz[ji, 0]
            for jj in range(1, gmet.nsnow, 1):
                zsnowzo[ji, jj + 1] = zsnowzo[ji, jj] + snowdz_old[ji, jj]
                zsnowzn[ji, jj + 1] = zsnowzn[ji, jj] + snowdz[ji, jj]
            for jj in range(0, gmet.nsnow, 1):
                if jj == 0:
                    locflag[jj, 0] = 0
                else:
                    for jjj in range(gmet.nsnow - 1, -1, -1):
                        if zsnowzn[ji, jj] <= zsnowzo[ji, jjj + 1]:
                            locflag[jj, 0] = jjj
                if jj == gmet.nsnow - 1:
                    locflag[jj, 1] = gmet.nsnow - 1
                else:
                    for jjj in range(gmet.nsnow - 1, -1, -1):
                        if zsnowzn[ji, jj + 1] <= zsnowzo[ji, jjj + 1]:
                            locflag[jj, 1] = jjj
                #print(f'locflag[{jj},:] = {locflag[jj,:]}')
                #print(locflag[jj, 0],locflag[jj, 1])
                if locflag[jj, 0] == locflag[jj, 1]:
                    zsnowrhon[ji, jj] = snowrho[ji, int(locflag[jj, 0])]
                    zsnowheatn[ji, jj] = snowheat[ji, int(locflag[jj, 0])] * snowdz[ji, jj] / snowdz_old[ji, int(locflag[jj, 0])]
                    #print(f" If zsnowheatn[{ji}, {jj}] = {zsnowheatn[ji, jj]}")
                    zsnowgrainn[ji, jj] = snowgrain[ji, int(locflag[jj, 0])]
                else:
                    zsnowrhon[ji, jj] = snowrho[ji, int(locflag[jj, 0])] * (zsnowzo[ji, int(locflag[jj, 0]) + 1] - zsnowzn[ji, jj]) + snowrho[ji, int(locflag[jj, 1])] * (zsnowzn[ji, jj + 1] - zsnowzo[ji, int(locflag[jj, 1]) + 1 - 1])
                    for jjj in range(locflag[jj, 0], locflag[jj, 1] , 1):
                        zsnowrhon[ji, jj] = zsnowrhon[ji, jj] + (jjj - locflag[jj, 0]) * snowrho[ji, jjj] * snowdz_old[ji, jjj]
                    zsnowrhon[ji, jj] = zsnowrhon[ji, jj] / snowdz[ji, jj]
                    if snowdz_old[ji, int(locflag[jj, 0])] > 0.0:
                        zsnowheatn[ji, jj] = snowheat[ji, int(locflag[jj, 0])] * (zsnowzo[ji, int(locflag[jj, 0]) + 1] - zsnowzn[ji, jj]) / snowdz_old[ji, int(locflag[jj, 0])] + snowheat[ji, int(locflag[jj, 1])] * (zsnowzn[ji, jj + 1] - zsnowzo[ji, int(locflag[jj, 1]) + 1 - 1]) / snowdz_old[ji, int(locflag[jj, 1])]
                        # print(f"znowwheatn[{ji},{jj}] = {zsnowheatn[ji, jj]}")
                    else:
                        zsnowheatn[ji, jj] = snowheat[ji, int(locflag[jj, 1])] / snowdz_old[ji, int(locflag[jj, 1])] * (zsnowzn[ji, int(locflag[jj, 0]) + 1] - zsnowzo[ji, int(locflag[jj, 0]) + 1])
                    for jjj in range(locflag[jj, 0], locflag[jj, 1], 1):
                        zsnowheatn[ji, jj] = zsnowheatn[ji, jj] + (jjj - locflag[jj, 0]) * snowheat[ji, jjj]
                    zsnowgrainn[ji, jj] = snowgrain[ji, int(locflag[jj, 0])] * (zsnowzo[ji, int(locflag[jj, 0]) + 1] - zsnowzn[ji, jj]) + snowgrain[ji, int(locflag[jj, 1])] * (zsnowzn[ji, jj + 1] - zsnowzo[ji, locflag[jj, 1] - 1 + 1])
                    for jjj in range(locflag[jj, 0], locflag[jj, 1], 1):
                        zsnowgrainn[ji, jj] = zsnowgrainn[ji, jj] + (jjj - locflag[jj, 0]) * snowgrain[ji, jjj] * snowdz_old[ji, jjj]
                    zsnowgrainn[ji, jj] = zsnowgrainn[ji, jj] / snowdz[ji, jj]
            snowrho[ji, :] = zsnowrhon[ji, :]
            snowheat[ji, :] = zsnowheatn[ji, :]
            #print(f"snowheat[{ji},:] = {snowheat[ji,:]}")
            #print(f"znowwheatn[{ji},:] = {zsnowheatn[ji, :]}")
            snowgrain[ji, :] = zsnowgrainn[ji, :]

def test_explicitsnow_transf(snowdz, snowgrain, snowheat, snowrho):
    print('--- inside the test function for explicitsnow_transf ---')
    path = f'/home/ssivanes/Fgpt/benchmark/explicitsnow_transf/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['snowdz', 'snowgrain', 'snowheat', 'snowrho']
    for (variable, value) in zip(modif_var, [snowdz, snowgrain, snowheat, snowrho]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32 or value.dtype == np.bool:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gmet = Global_module_explicitsnow_transf()
    snowdz = np.zeros((gmet.kjpindex, gmet.nsnow), dtype=np.float64)
    snowdz_old = np.zeros((gmet.kjpindex, gmet.nsnow), dtype=np.float64)
    snowgrain = np.zeros((gmet.kjpindex, gmet.nsnow), dtype=np.float64)
    snowheat = np.zeros((gmet.kjpindex, gmet.nsnow), dtype=np.float64)
    snowrho = np.zeros((gmet.kjpindex, gmet.nsnow), dtype=np.float64)
    read_dummy(snowdz, snowdz_old, snowgrain, snowheat, snowrho)
    explicitsnow_transf(snowdz_old, snowdz, snowrho, snowheat, snowgrain, gmet)
    test_explicitsnow_transf(snowdz, snowgrain, snowheat, snowrho)
if __name__ == '__main__':
    main()
```

```python
import numpy as np
import os
from scipy.io import FortranFile

class Global_module_explicitsnow_subli:

    def __init__(self):
        self.nsnow = np.int32(3)
        self.nslm = np.int32(11)
        self.nvm = np.int32(15)
        self.nstm = np.int32(3)
        self.kjpindex = np.int32(4717)
        self.ier = np.int32(0)
        self.ic0 = np.int32(0)
        self.ic = np.int32(0)
        self.icr = np.float64(0.0)
        self.start_time = np.float64(0.0)
        self.stop_time = np.float64(0.0)
        self.min_sechiba = np.float64(1e-08)
        self.zero = np.float64(0.0)
        self.nnobio = np.int32(1)
        self.tp_00 = np.float64(273.15)
        self.iice = np.int32(1)
```

```python
import os
import time
import functools
import numpy as np
import logging
from scipy.io import FortranFile
#from module_global import Global_module_explicitsnow_subli

def read_dummy(frac_nobio, snowdz, snowrho, vevapsno):
    print(f'--- inside the read dummy routine for explicitsnow_subli ---')
    path = f'/home/ssivanes/Fgpt/benchmark/explicitsnow_subli/dummy.bin'
    ffile = FortranFile(path, 'r')
    for var in [frac_nobio, snowdz, snowrho, vevapsno]:
        if isinstance(var, np.ndarray):
            arr_shape = var.shape
            if var.dtype == np.float64:
                data = ffile.read_reals(np.float64)
            elif var.dtype == np.int32 or var.dtype == np.bool:
                data = ffile.read_ints(np.int32)
            if data.size != np.prod(arr_shape):
                continue
            var[:] = data.reshape(arr_shape, order='F')

def timer(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"[TIMER] '{func.__name__}' executed in {duration:.6f} seconds")
        path = '/home/ssivanes/Fgpt/benchmark/explicitsnow_subli/time.txt'
        with open(path, 'a') as f:
            f.write(f"[TIMER] Python: '{func.__name__}' executed in {duration:.6f} seconds \n")
        return result
    return wrapper

def explicitsnow_subli(snowrho, frac_nobio, snowdz, vevapsno, snowliq, snowtemp, snow, subsnownobio, subsinksoil, gmes):
    WSNOWDZ = np.zeros((gmes.kjpindex, gmes.nsnow), dtype=np.float64)
    SMASSC = np.zeros((gmes.kjpindex,), dtype=np.float64)
    SMASS = np.zeros((gmes.kjpindex, gmes.nsnow), dtype=np.float64)
    ZSNOWEVAPS = np.zeros((gmes.kjpindex,), dtype=np.float64)
    ZSNOWDZ = np.zeros((gmes.kjpindex,), dtype=np.float64)
    snowacc = np.zeros((gmes.kjpindex,), dtype=np.float64)
    subsnowveg = np.zeros((gmes.kjpindex,), dtype=np.float64)
    snow[:] = 0.0
    for ji in range(0, gmes.kjpindex, 1):
        snow[ji] = np.sum(snowrho[ji, :] * snowdz[ji, :])
    subsnownobio[:, :] = gmes.zero
    subsinksoil[:] = gmes.zero
    for ji in range(0,gmes.kjpindex, 1): # 6,10, 1
        if vevapsno[ji] > snow[ji]:
            subsinksoil[ji] = vevapsno[ji] - snow[ji]
            vevapsno[ji] = snow[ji]
            snow[ji] = gmes.zero
            snowdz[ji, :] = 0
            snowliq[ji, :] = 0
            snowtemp[ji, :] = gmes.tp_00
            if frac_nobio[ji, gmes.iice - 1] > gmes.min_sechiba:
                subsnownobio[ji, gmes.iice - 1] = frac_nobio[ji, gmes.iice - 1] * vevapsno[ji]
                subsnowveg[ji] = vevapsno[ji] - subsnownobio[ji, gmes.iice - 1]
            else:
                subsnownobio[ji, gmes.iice - 1] = gmes.zero
                subsnowveg[ji] = vevapsno[ji]
        else:
            WSNOWDZ[ji, :] = snowdz[ji, :] * snowrho[ji, :]
            SMASSC[ji] = 0.0
            for jj in range(0, gmes.nsnow, 1):
                SMASS[ji, jj] = SMASSC[ji] + WSNOWDZ[ji, jj]
                SMASSC[ji] = SMASSC[ji] + WSNOWDZ[ji, jj]
            
            locjj = 0
            for jj in range(0, gmes.nsnow - 1, 1):
                #print(f'jj={jj}, SMASS={SMASS[ji, jj]}, next={SMASS[ji, jj + 1]}, vevapsno={vevapsno[ji]}')
                if SMASS[ji, jj] <= vevapsno[ji] and SMASS[ji, jj + 1] >= vevapsno[ji]:
                    locjj = jj + 1
                    #print(f' For loop : locjj = {locjj}')
            #print(f'locjj = {locjj}, ji = {ji}')
            if locjj == 0:
                ZSNOWEVAPS[ji] = vevapsno[ji] / snowrho[ji, 0]
                ZSNOWDZ[ji] = snowdz[ji, 0] - ZSNOWEVAPS[ji]
                snowdz[ji, 0] = np.maximum(0.0, ZSNOWDZ[ji])
            
            elif locjj > 0:
                snowacc[ji] = 0
                for jj in range(0, locjj, 1):
                    snowacc[ji] = snowacc[ji] + snowdz[ji, jj] * snowrho[ji, jj]
                    snowdz[ji, jj] = 0
                #print(f'snowdz[{ji},0] = {snowdz[ji,0]}')
                ZSNOWEVAPS[ji] = (vevapsno[ji] - snowacc[ji]) / snowrho[ji, locjj]
                ZSNOWDZ[ji] = snowdz[ji, locjj] - ZSNOWEVAPS[ji]
                snowdz[ji, locjj] = np.maximum(0.0, ZSNOWDZ[ji])
            
            
def test_explicitsnow_subli(snowtemp, snowliq, subsinksoil, subsnownobio, vevapsno, snowdz, snow):
    print('--- inside the test function for explicitsnow_subli ---')
    path = f'/home/ssivanes/Fgpt/benchmark/explicitsnow_subli/output.bin'
    ffile = FortranFile(path, 'r')
    modif_var = ['snow', 'snowdz', 'snowliq', 'snowtemp', 'subsinksoil', 'subsnownobio', 'vevapsno']
    for (variable, value) in zip(modif_var, [snow, snowdz, snowliq, snowtemp, subsinksoil, subsnownobio, vevapsno]):
        fortran_value = None
        if isinstance(value, np.ndarray):
            arr_shape = value.shape
            if value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64).reshape(arr_shape, order='F')
            elif value.dtype == np.int32 or value.dtype == np.bool:
                fortran_value = ffile.read_ints(np.int32).reshape(arr_shape, order='F')
            if np.allclose(value, fortran_value):
                print(f'Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: All elements in {variable} do not match to FORTRAN variable: {variable}')
                print(f' Maximum absolute error: {np.max(abs(value - fortran_value))}')
                print(f' Min and max of python value: {value.min()}, {value.max()}')
                print(f' Min and max of FORTRAN value: {fortran_value.min()}, {fortran_value.max()}')
        elif not isinstance(value, np.ndarray):
            if value.dtype == np.bool or value.dtype == np.int32:
                fortran_value = ffile.read_ints(np.int32)[0]
            elif value.dtype == np.float64:
                fortran_value = ffile.read_reals(np.float64)[0]
            if np.isclose(value, fortran_value):
                print(f' Test passed: Variable:{variable} is equal to FORTRAN variable: {variable}')
            else:
                print(f'Test failed: Python variable:{variable} do not match to FORTRAN variable: {variable}')

def main():
    print('--- inside the main program ---')
    gmes = Global_module_explicitsnow_subli()
    frac_nobio = np.zeros((gmes.kjpindex, gmes.nnobio), dtype=np.float64)
    snow = np.zeros((gmes.kjpindex,), dtype=np.float64)
    snowdz = np.zeros((gmes.kjpindex, gmes.nsnow), dtype=np.float64)
    snowliq = np.zeros((gmes.kjpindex, gmes.nsnow), dtype=np.float64)
    snowrho = np.zeros((gmes.kjpindex, gmes.nsnow), dtype=np.float64)
    snowtemp = np.zeros((gmes.kjpindex, gmes.nsnow), dtype=np.float64)
    subsinksoil = np.zeros((gmes.kjpindex,), dtype=np.float64)
    subsnownobio = np.zeros((gmes.kjpindex, gmes.nnobio), dtype=np.float64)
    vevapsno = np.zeros((gmes.kjpindex,), dtype=np.float64)
    read_dummy(frac_nobio, snowdz, snowrho, vevapsno)
    explicitsnow_subli(snowrho, frac_nobio, snowdz, vevapsno, snowliq, snowtemp, snow, subsnownobio, subsinksoil, gmes)
    test_explicitsnow_subli(snowtemp, snowliq, subsinksoil, subsnownobio, vevapsno, snowdz, snow)
if __name__ == '__main__':
    main()
```

```python
def hydrol_hydraulic_arch_tuzet_calc(kjit, kjpindex, ks, nvan, avan, transpir, mc_out, veget, veget_max, njsc, soiltile, circ_class_n, circ_class_biomass, u, v, tq_cdrag, gsmean, pb, temp_air, lalo, psi_leaf, psi_leaf_next, psi_sto_leaf_save, psi_sto_wood_save, e_frac, psi_root_sup, psi_root_inf, psi_xylem_trunk, psi_xylem_leaf, psi_xylem_collar, psi_sto_wood, psi_sto_leaf, mc_i_sup, mc_i_inf, F_absorption):
    mc_sup = np.zeros((kjpindex, nstm), dtype=np.float64)
    mc_inf = np.zeros((kjpindex, nstm), dtype=np.float64)
    mc_sup_save = np.zeros((kjpindex, nstm), dtype=np.float64)
    mc_inf_save = np.zeros((kjpindex, nstm), dtype=np.float64)
    Fi = np.zeros((kjpindex, nvm, nslm, nstm), dtype=np.float64)
    Fxylem_up = np.zeros((kjpindex, nvm), dtype=np.float64)
    Fxylem_low = np.zeros((kjpindex, nvm), dtype=np.float64)
    Fsto_leaf = np.zeros((kjpindex, nvm), dtype=np.float64)
    Fsto_wood = np.zeros((kjpindex, nvm), dtype=np.float64)
    Fsup = np.zeros((kjpindex, nvm), dtype=np.float64)
    Finf = np.zeros((kjpindex, nvm), dtype=np.float64)
    V_wood = np.zeros((kjpindex, nvm), dtype=np.float64)
    V_wood_save = np.zeros((kjpindex, nvm), dtype=np.float64)
    V_leaf = np.zeros((kjpindex, nvm), dtype=np.float64)
    V_leaf_save = np.zeros((kjpindex, nvm), dtype=np.float64)
    C_wood = np.zeros((kjpindex, nvm), dtype=np.float64)
    C_leaf = np.zeros((kjpindex, nvm), dtype=np.float64)
    tau_wood = np.zeros((kjpindex, nvm), dtype=np.float64)
    tau_leaf = np.zeros((kjpindex, nvm), dtype=np.float64)
    Vr_wood = np.zeros((kjpindex, nvm), dtype=np.float64)
    Vr_leaf = np.zeros((kjpindex, nvm), dtype=np.float64)
    Vmax_wood = np.zeros((kjpindex, nvm), dtype=np.float64)
    Vmax_leaf = np.zeros((kjpindex, nvm), dtype=np.float64)
    Res_sto_leaf = np.zeros((kjpindex, nvm), dtype=np.float64)
    Res_sto_wood = np.zeros((kjpindex, nvm), dtype=np.float64)
    Res_xylem_up = np.zeros((kjpindex, nvm), dtype=np.float64)
    Res_xylem_low = np.zeros((kjpindex, nvm), dtype=np.float64)
    Res_mesophyll = np.zeros((kjpindex, nvm), dtype=np.float64)
    Res_root_sup = np.zeros((kjpindex, nvm), dtype=np.float64)
    Res_root_inf = np.zeros((kjpindex, nvm), dtype=np.float64)
    psi_sto_wood_predict = np.zeros((kjpindex, nvm), dtype=np.float64)
    psi_sto_leaf_predict = np.zeros((kjpindex, nvm), dtype=np.float64)
    psi_xylem_leaf_temp = np.zeros((kjpindex, nvm), dtype=np.float64)
    psi_xylem_trunk_temp = np.zeros((kjpindex, nvm), dtype=np.float64)
    psi_xylem_collar_temp = np.zeros((kjpindex, nvm), dtype=np.float64)
    psi_root_sup_temp = np.zeros((kjpindex, nvm), dtype=np.float64)
    psi_root_inf_temp = np.zeros((kjpindex, nvm), dtype=np.float64)
    psi_leaf_temp = np.zeros((kjpindex, nvm), dtype=np.float64)
    psi_leaf_previous = np.zeros((kjpindex, nvm), dtype=np.float64)
    psi_leaf_temp_save = np.zeros((kjpindex, nvm), dtype=np.float64)
    psi_sto_wood_temp = np.zeros((kjpindex, nvm), dtype=np.float64)
    psi_sto_leaf_temp = np.zeros((kjpindex, nvm), dtype=np.float64)
    Fi_temp = np.zeros((kjpindex, nvm, nslm, nstm), dtype=np.float64)
    Fsup_temp = np.zeros((kjpindex, nvm), dtype=np.float64)
    Finf_temp = np.zeros((kjpindex, nvm), dtype=np.float64)
    F_absorption_temp = np.zeros((kjpindex, nvm), dtype=np.float64)
    mc_out_temp = np.zeros((kjpindex, nslm, nstm), dtype=np.float64)
    mc_sup_temp = np.zeros((kjpindex, nstm), dtype=np.float64)
    mc_inf_temp = np.zeros((kjpindex, nstm), dtype=np.float64)
    mc_i_sup_temp = np.zeros((kjpindex, nvm, nrp), dtype=np.float64)
    mc_i_inf_temp = np.zeros((kjpindex, nvm, nrp), dtype=np.float64)
    func_predict = np.zeros((kjpindex, nvm), dtype=np.float64)
    func_n = np.zeros((kjpindex, nvm), dtype=np.float64)
    func_n_1 = np.zeros((kjpindex, nvm), dtype=np.float64)
    f_psi_1 = np.zeros((kjpindex, nvm), dtype=np.float64)
    f_psi_2 = np.zeros((kjpindex, nvm), dtype=np.float64)
    speed = np.zeros((kjpindex,), dtype=np.float64)
    nsub_step = np.zeros((kjpindex, nvm), dtype=np.int32)
    launch_next_calc = np.zeros((kjpindex, nvm), dtype=np.bool)
    this_tstep_calc = np.zeros((kjpindex, nvm), dtype=np.bool)
    transpir_test = np.zeros((kjpindex, nvm), dtype=np.float64)
    g_stom = np.zeros((kjpindex, nvm), dtype=np.float64)
    delta = np.zeros((kjpindex,), dtype=np.float64)
    h0 = np.zeros((kjpindex,), dtype=np.float64)
    sin_t1 = np.zeros((kjpindex,), dtype=np.float64)
    sin_t2 = np.zeros((kjpindex,), dtype=np.float64)
    kjit_sunrise = np.zeros((kjpindex,), dtype=np.int32)
    kjit_delta = np.zeros((kjpindex,), dtype=np.int32)
    nsub_step[:, :] = 1
    launch_next_calc[:, :] = False
    this_tstep_calc[:, :] = True
    for ji in range(0, kjpindex, 1):
        for jv in range(0, nvm, 1):
            if transpir[ji, jv] > min_sechiba:
                if psi_leaf[ji, jv] < psi_ref_g0[jv] - psi_ref_g0[jv] / 2.0 and gsmean[ji, jv] != 0:
                    nsub_step[ji, jv] = 25
                    launch_next_calc = True
                    psi_leaf_previous[ji, jv] = psi_leaf[ji, jv]
    for ji in range(0, kjpindex, 1):
        for jv in range(0, nvm, 1):
            Fi[ji, jv, :, :] = 0.0
            Fi_temp[ji, jv, :, :] = 0.0
            F_absorption[ji, jv] = 0.0
            mc_i_sup_temp[ji, jv, :] = mc_i_sup[ji, jv, :]
            mc_i_inf_temp[ji, jv, :] = mc_i_inf[ji, jv, :]
            mc_out_temp[ji, :, :] = mc_out[ji, :, :]
            jst = pref_soil_veg[jv]
            for i in range(0, int(nsub_step[ji, jv]), 1):
                Fxylem_up[ji, jv] = 0.0
                Fxylem_low[ji, jv] = 0.0
                mc_sup_temp[ji, :] = 0.0
                mc_inf_temp[ji, :] = 0.0
                mc_i_sup_temp[ji, jv, :] = mc_i_sup[ji, jv, :]
                mc_i_inf_temp[ji, jv, :] = mc_i_inf[ji, jv, :]
                psi_root_sup_temp[ji, jv] = psi_root_sup[ji, jv]
                psi_root_inf_temp[ji, jv] = psi_root_inf[ji, jv]
                mc_out_temp[ji, :, :] = mc_out[ji, :, :]
                if this_tstep_calc[ji, jv]:
                    psi_xylem_leaf_temp[ji, jv] = psi_xylem_leaf[ji, jv]
                    psi_xylem_trunk_temp[ji, jv] = psi_xylem_trunk[ji, jv]
                    psi_xylem_collar_temp[ji, jv] = psi_xylem_collar[ji, jv]
                    psi_sto_leaf_temp[ji, jv] = psi_sto_leaf[ji, jv]
                    psi_sto_wood_temp[ji, jv] = psi_sto_wood[ji, jv]
                if this_tstep_calc[ji, jv]:
                    if soiltile[ji, int(pref_soil_veg[jv])] * vegtot[ji] > min_sechiba:
                        transpir_test[ji, jv] = transpir[ji, jv] / (dt_sechiba * kilo_to_unit * soiltile[ji, int(pref_soil_veg[jv])] * vegtot[ji])
                    else:
                        transpir_test[ji, jv] = 0.0
                elif soiltile[ji, int(pref_soil_veg[jv])] * vegtot[ji] > min_sechiba:
                    gamma = 2.0 * pi * julian_diff / one_year
                    dec = 0.006918 - 0.399912 * np.cos(gamma) + 0.070257 * np.sin(gamma) - 0.006758 * np.cos(2 * gamma) + 0.000907 * np.sin(2 * gamma) - 0.002697 * np.cos(3 * gamma) + 0.00148 * np.sin(3 * gamma)
                    kjit_sunrise[ji] = int(6.5 * 3600 / dt_sechiba)
                    kjit_delta[ji] = int(21 * 3600 / dt_sechiba) - kjit_sunrise[ji]
                    kjit_temp = np.mod(kjit, int(one_day / dt_sechiba))
                    if ((kjit_temp - kjit_sunrise[ji] < kjit_delta[ji] and kjit_temp + 1 - kjit_sunrise[ji] < kjit_delta[ji]) and kjit_temp > kjit_sunrise[ji]) and kjit_temp + 1 > kjit_sunrise[ji]:
                        sin_t1[ji] = np.abs(np.sin((kjit_temp - kjit_sunrise[ji]) * pi / kjit_delta[ji]))
                        sin_t2[ji] = np.abs(np.sin((kjit_temp + 1 - kjit_sunrise[ji]) * pi / kjit_delta[ji]))
                        delta[ji] = sin_t2[ji] / sin_t1[ji]
                    else:
                        delta[ji] = 1
                    g_stom[ji, jv] = mol_to_m_1 * (temp_air[ji] / tp_00) * (pb_std / pb[ji]) * gsmean[ji, jv] * ratio_H2O_to_CO2
                    f_psi_1[ji, jv] = (1.0 + np.exp(sf[jv] * psi_ref_g0[jv])) / (1.0 + np.exp(sf[jv] * (psi_ref_g0[jv] - psi_leaf_previous[ji, jv])))
                    f_psi_2[ji, jv] = (1.0 + np.exp(sf[jv] * psi_ref_g0[jv])) / (1.0 + np.exp(sf[jv] * (psi_ref_g0[jv] - psi_leaf_temp[ji, jv])))
                    speed[ji] = np.minimum(min_wind, np.sqrt(u[ji] * u[ji] + v[ji] * v[ji]))
                    transpir_test[ji, jv] = transpir[ji, jv] / (dt_sechiba * kilo_to_unit * soiltile[ji, int(pref_soil_veg[jv])] * vegtot[ji]) * delta[ji] * ((1 + veget[ji, jv] / veget_max[ji, jv] * (tq_cdrag[ji] * speed[ji]) / g_stom[ji, jv]) / (1 + veget[ji, jv] / veget_max[ji, jv] * (tq_cdrag[ji] * speed[ji] * f_psi_1[ji, jv]) / (delta[ji] * g_stom[ji, jv] * f_psi_2[ji, jv])))
                else:
                    transpir_test[ji, jv] = 0.0
                for jsl in range(0, nslm, 1):
                    if np.sum(Fi_temp[ji, jv, :, jst]) > min_sechiba:
                        if isl <= lim_layer:
                            mc_out_temp[ji, jsl, jst] = mc_out[ji, jsl, jst] - Fsup_temp[ji, jv] * Fi_temp[ji, jv, jsl, jst] / np.sum(Fi_temp[ji, jv, :, jst]) * dt_sechiba * kilo_to_unit / np.sum(dh[:lim_layer])
                        else:
                            mc_out_temp[ji, jsl, jst] = mc_out[ji, jsl, jst] - Finf_temp[ji, jv] * Fi_temp[ji, jv, jsl, jst] / np.sum(Fi_temp[ji, jv, :, jst]) * dt_sechiba * kilo_to_unit / np.sum(dh[lim_layer + 1:])
                    else:
                        mc_out_temp[ji, jsl, :] = mc_out[ji, jsl, :]
                if ok_tuzet_hydrol_arch_dyn_resist:
                    Res_mesophyll[ji, jv] = (1 + np.exp(a_leaf[jv] * (psi_xylem_leaf_temp[ji, jv] - psi_50_res_leaf[jv]))) / k_leaf_max[jv]
                    Res_sto_leaf[ji, jv] = (1 + np.exp(a_leaf[jv] * (psi_sto_leaf_temp[ji, jv] - psi_50_res_leaf[jv]))) / k_leaf_max[jv]
                    Res_sto_wood[ji, jv] = (1 + np.exp(a_wood[jv] * (psi_sto_wood_temp[ji, jv] - psi_50_res_wood[jv]))) / k_wood_max[jv]
                    Res_xylem_up[ji, jv] = (1 + np.exp(a_wood[jv] * (psi_xylem_trunk_temp[ji, jv] - psi_50_res_wood[jv]))) / k_wood_max[jv]
                    Res_xylem_low[ji, jv] = (1 + np.exp(a_wood[jv] * (psi_xylem_collar_temp[ji, jv] - psi_50_res_wood[jv]))) / k_wood_max[jv]
                    Res_root_sup[ji, jv] = (1 + np.exp(a_root[jv] * (psi_root_sup_temp[ji, jv] - psi_50_res_root[jv]))) / k_root_max[jv]
                    Res_root_inf[ji, jv] = (1 + np.exp(a_root[jv] * (psi_root_inf_temp[ji, jv] - psi_50_res_root[jv]))) / k_root_max[jv]
                else:
                    Res_sto_leaf[ji, jv] = Rsto_leaf[jv]
                    Res_sto_wood[ji, jv] = Rsto_wood[jv]
                    Res_mesophyll[ji, jv] = Rleaf[jv]
                    Res_xylem_up[ji, jv] = Rxylem[jv]
                    Res_xylem_low[ji, jv] = Rxylem[jv]
                    Res_root_sup[ji, jv] = Rroot_sup[jv]
                    Res_root_inf[ji, jv] = Rroot_inf[jv]
                for jsl in range(0, nslm, 1):
                    for jst in range(0, nstm, 1):
                        if isl <= lim_layer:
                            mc_sup_temp[ji, jst] = mc_sup_temp[ji, jst] + mc_out_temp[ji, jsl, jst] * dh[jsl] / np.sum(dh[:lim_layer])
                        else:
                            mc_inf_temp[ji, jst] = mc_inf_temp[ji, jst] + mc_out_temp[ji, jsl, jst] * dh[jsl] / np.sum(dh[lim_layer + 1:])
                if ok_tuzet_hydrol_arch_storage:
                    if soiltile[ji, jst] * vegtot[ji] > min_sechiba:
                        if np.sum(circ_class_biomass[ji, jv, :, ileaf, icarbon] * circ_class_n[ji, jv, :]) > min_sechiba:
                            Vmax_leaf[ji, jv] = deux * np.sum(circ_class_biomass[ji, jv, :, ileaf, icarbon] * circ_class_n[ji, jv, :]) * 0.8 / (kilo_to_unit * ph2o) * 6.0
                            Vr_leaf[ji, jv] = 0.4 * Vmax_leaf[ji, jv]
                            V_leaf_save[ji, jv] = (1 + np.exp(lambda_leaf[jv] * psi_ref_sto_leaf[jv])) / (1 + np.exp(lambda_leaf[jv] * (-psi_sto_leaf[ji, jv] + psi_ref_sto_leaf[jv]))) * (Vmax_leaf[ji, jv] - Vr_leaf[ji, jv]) + Vr_leaf[ji, jv]
                            C_leaf[ji, jv] = lambda_leaf[jv] * ((1 + np.exp(lambda_leaf[jv] * psi_ref_sto_leaf[jv])) * Vmax_leaf[ji, jv] + Vr_leaf[ji, jv] * np.exp(lambda_leaf[jv] * psi_ref_sto_leaf[jv]) - V_leaf_save[ji, jv]) / (1 + np.exp(lambda_leaf[jv] * psi_ref_sto_leaf[jv])) * ((1 + np.exp(lambda_leaf[jv] * psi_ref_sto_leaf[jv])) / (1 + np.exp(lambda_leaf[jv] * (-psi_sto_leaf[ji, jv] + psi_ref_sto_leaf[jv]))))
                            tau_leaf[ji, jv] = C_leaf[ji, jv] * (Res_xylem_up[ji, jv] + Res_sto_leaf[ji, jv])
                            psi_sto_leaf_predict[ji, jv] = psi_sto_leaf[ji, jv] + dt_sechiba * (1 / tau_leaf[ji, jv]) * (psi_xylem_trunk[ji, jv] - Res_xylem_up[ji, jv] * transpir_test[ji, jv] - psi_sto_leaf[ji, jv])
                            func_n[ji, jv] = 1 / tau_leaf[ji, jv] * (psi_xylem_trunk[ji, jv] - Res_xylem_up[ji, jv] * transpir_test[ji, jv] - psi_sto_leaf[ji, jv])
                            func_predict[ji, jv] = 1 / tau_leaf[ji, jv] * (psi_xylem_trunk[ji, jv] - Res_xylem_up[ji, jv] * transpir_test[ji, jv] - psi_sto_leaf_predict[ji, jv])
                            func_n_1[ji, jv] = 1 / tau_leaf[ji, jv] * (psi_xylem_trunk[ji, jv] - Res_xylem_up[ji, jv] * transpir_test[ji, jv] - psi_sto_leaf_save[ji, jv])
                            psi_sto_leaf_save[ji, jv] = psi_sto_leaf[ji, jv]
                            psi_sto_leaf_temp[ji, jv] = np.maximum(psi_xylem_leaf[ji, jv], np.minimum(0.0, psi_sto_leaf_save[ji, jv] + dt_sechiba / 12.0 * (5.0 * func_predict[ji, jv] + 8.0 * func_n[ji, jv] - func_n_1[ji, jv])))
                            V_leaf[ji, jv] = np.minimum(Vmax_leaf[ji, jv], (1 + np.exp(lambda_leaf[jv] * psi_ref_sto_leaf[jv])) / (1 + np.exp(lambda_leaf[jv] * (-psi_sto_leaf_temp[ji, jv] + psi_ref_sto_leaf[jv]))) * (Vmax_leaf[ji, jv] - Vr_leaf[ji, jv]) + Vr_leaf[ji, jv])
                            Fsto_leaf[ji, jv] = (V_leaf[ji, jv] - V_leaf_save[ji, jv]) / dt_sechiba
                            if Fsto_leaf[ji, jv] > 0.0 and transpir[ji, jv] > min_sechiba:
                                Fsto_leaf[ji, jv] = 0.0
                                psi_sto_leaf_save[ji, jv] = psi_sto_leaf[ji, jv]
                                psi_sto_leaf_temp[ji, jv] = psi_sto_leaf[ji, jv]
                            if Fsto_leaf[ji, jv] < 0.0 and transpir[ji, jv] < min_sechiba:
                                Fsto_leaf[ji, jv] = 0.0
                                psi_sto_leaf_save[ji, jv] = psi_sto_leaf[ji, jv]
                                psi_sto_leaf_temp[ji, jv] = psi_sto_leaf[ji, jv]
                        else:
                            psi_sto_leaf_save[ji, jv] = psi_sto_leaf[ji, jv]
                            psi_sto_leaf_temp[ji, jv] = psi_xylem_leaf[ji, jv]
                            Fsto_leaf[ji, jv] = 0.0
                    else:
                        psi_sto_leaf_save[ji, jv] = psi_sto_leaf[ji, jv]
                        psi_sto_leaf_temp[ji, jv] = psi_xylem_leaf[ji, jv]
                        Fsto_leaf[ji, jv] = 0.0
                    if soiltile[ji, jst] * vegtot[ji] > min_sechiba:
                        if np.sum(circ_class_biomass[ji, jv, :, ileaf, icarbon] * circ_class_n[ji, jv, :]) > min_sechiba:
                            Fxylem_up[ji, jv] = transpir_test[ji, jv] + Fsto_leaf[ji, jv]
                        else:
                            Fxylem_up[ji, jv] = transpir_test[ji, jv]
                    else:
                        Fxylem_up[ji, jv] = 0.0
                    if np.sum((circ_class_biomass[ji, jv, :, isapabove, icarbon] + circ_class_biomass[ji, jv, :, isapbelow, icarbon]) * circ_class_n[ji, jv, :]) > min_sechiba:
                        Vmax_wood[ji, jv] = deux * np.sum((circ_class_biomass[ji, jv, :, isapabove, icarbon] + circ_class_biomass[ji, jv, :, isapbelow, icarbon]) * circ_class_n[ji, jv, :]) * 0.8 / (kilo_to_unit * ph2o) * 5.0
                        Vr_wood[ji, jv] = 0.4 * Vmax_wood[ji, jv]
                        V_wood_save[ji, jv] = (1 + np.exp(lambda_wood[jv] * psi_ref_sto_wood[jv])) / (1 + np.exp(lambda_wood[jv] * (-psi_sto_wood[ji, jv] + psi_ref_sto_wood[jv]))) * (Vmax_wood[ji, jv] - Vr_wood[ji, jv]) + Vr_wood[ji, jv]
                        C_wood[ji, jv] = lambda_wood[jv] * ((1 + np.exp(lambda_wood[jv] * psi_ref_sto_wood[jv])) * Vmax_wood[ji, jv] + Vr_wood[ji, jv] * np.exp(lambda_wood[jv] * psi_ref_sto_wood[jv]) - V_wood_save[ji, jv]) / (1 + np.exp(lambda_wood[jv] * psi_ref_sto_wood[jv])) * ((1 + np.exp(lambda_wood[jv] * psi_ref_sto_wood[jv])) / (1 + np.exp(lambda_wood[jv] * (-psi_sto_wood[ji, jv] + psi_ref_sto_wood[jv]))))
                        tau_wood[ji, jv] = C_wood[ji, jv] * (Res_xylem_low[ji, jv] + Res_sto_wood[ji, jv])
                        psi_sto_wood_predict[ji, jv] = psi_sto_wood[ji, jv] + dt_sechiba * (1 / tau_wood[ji, jv]) * (psi_xylem_collar[ji, jv] - Res_xylem_low[ji, jv] * Fxylem_up[ji, jv] - psi_sto_wood[ji, jv])
                        func_n[ji, jv] = 1 / tau_wood[ji, jv] * (psi_xylem_collar[ji, jv] - Res_xylem_low[ji, jv] * Fxylem_up[ji, jv] - psi_sto_wood[ji, jv])
                        func_predict[ji, jv] = 1 / tau_wood[ji, jv] * (psi_xylem_collar[ji, jv] - Res_xylem_low[ji, jv] * Fxylem_up[ji, jv] - psi_sto_wood_predict[ji, jv])
                        func_n_1[ji, jv] = 1 / tau_wood[ji, jv] * (psi_xylem_collar[ji, jv] - Res_xylem_low[ji, jv] * Fxylem_up[ji, jv] - psi_sto_wood_save[ji, jv])
                        psi_sto_wood_save[ji, jv] = psi_sto_wood[ji, jv]
                        psi_sto_wood_temp[ji, jv] = np.maximum(psi_xylem_trunk[ji, jv], np.minimum(0.0, psi_sto_wood_save[ji, jv] + dt_sechiba / 12.0 * (5.0 * func_predict[ji, jv] + 8.0 * func_n[ji, jv] - func_n_1[ji, jv])))
                        V_wood[ji, jv] = np.minimum(Vmax_wood[ji, jv], (1 + np.exp(lambda_wood[jv] * psi_ref_sto_wood[jv])) / (1 + np.exp(lambda_wood[jv] * (-psi_sto_wood_temp[ji, jv] + psi_ref_sto_wood[jv]))) * (Vmax_wood[ji, jv] - Vr_wood[ji, jv]) + Vr_wood[ji, jv])
                        Fsto_wood[ji, jv] = (V_wood[ji, jv] - V_wood_save[ji, jv]) / dt_sechiba
                        if Fsto_wood[ji, jv] > 0.0 and transpir[ji, jv] > min_sechiba:
                            Fsto_wood[ji, jv] = 0.0
                            psi_sto_wood_save[ji, jv] = psi_sto_wood[ji, jv]
                            psi_sto_wood_temp[ji, jv] = psi_sto_wood[ji, jv]
                        if Fsto_wood[ji, jv] < 0.0 and transpir[ji, jv] < min_sechiba:
                            Fsto_wood[ji, jv] = 0.0
                            psi_sto_wood_save[ji, jv] = psi_sto_wood[ji, jv]
                            psi_sto_wood_temp[ji, jv] = psi_sto_wood[ji, jv]
                    else:
                        psi_sto_wood_save[ji, jv] = psi_sto_wood[ji, jv]
                        psi_sto_wood_temp[ji, jv] = psi_xylem_trunk[ji, jv]
                        Fsto_wood[ji, jv] = 0.0
                    if np.sum((circ_class_biomass[ji, jv, :, isapabove, icarbon] + circ_class_biomass[ji, jv, :, isapbelow, icarbon]) * circ_class_n[ji, jv, :]) > min_sechiba:
                        Fxylem_low[ji, jv] = Fxylem_up[ji, jv] + Fsto_wood[ji, jv]
                    else:
                        Fxylem_low[ji, jv] = Fxylem_up[ji, jv]
                    F_absorption_temp[ji, jv] = Fxylem_low[ji, jv]
                else:
                    jst = pref_soil_veg[jv]
                    if soiltile[ji, jst] * vegtot[ji] > min_sechiba:
                        Fxylem_up[ji, jv] = transpir_test[ji, jv]
                        Fxylem_low[ji, jv] = Fxylem_up[ji, jv]
                        F_absorption_temp[ji, jv] = Fxylem_low[ji, jv]
                    else:
                        F_absorption_temp[ji, jv] = 0.0
                if ok_tuzet_hydrol_arch_muff:
                    hydrol_hydraulic_arch_tuzet_muff(kjit, kjpindex, ji, jv, soiltile, veget_max, njsc, ks, nvan, avan, F_absorption_temp, circ_class_biomass, circ_class_n, Res_root_sup, Res_root_inf, Fsup_temp, Finf_temp, psi_root_sup_temp, psi_root_inf_temp, mc_sup_temp, mc_inf_temp, mc_i_sup_temp, mc_i_inf_temp)
                else:
                    hydrol_hydraulic_arch_tuzet_resist(kjpindex, ji, jv, soiltile, njsc, ks, nvan, avan, F_absorption_temp, circ_class_biomass, circ_class_n, mc_sup_temp, mc_inf_temp, Res_root_sup, Res_root_inf, Fsup_temp, Finf_temp, psi_root_sup_temp, psi_root_inf_temp)
                for jsl in range(0, nslm, 1):
                    if isl <= lim_layer:
                        Fi_temp[ji, jv, jsl, jst] = Fsup_temp[ji, jv] * (mc_out_temp[ji, jsl, jst] - mcr_sup[ji]) * dh[jsl] / (np.maximum(0.001, mc_sup_temp[ji, jst] - mcr_sup[ji]) * np.sum(dh[:lim_layer]))
                    else:
                        Fi_temp[ji, jv, jsl, jst] = Finf_temp[ji, jv] * (mc_out_temp[ji, jsl, jst] - mcr_inf[ji]) * dh[jsl] / (np.maximum(0.001, mc_inf_temp[ji, jst] - mcr_inf[ji]) * np.sum(dh[lim_layer + 1:]))
                psi_xylem_collar_temp[ji, jv] = -(Fsup_temp[ji, jv] * Res_root_sup[ji, jv]) + psi_root_sup_temp[ji, jv]
                psi_leaf_temp_save[ji, jv] = psi_leaf_temp[ji, jv]
                if ok_tuzet_hydrol_arch_storage:
                    psi_xylem_trunk_temp[ji, jv] = -(Fxylem_low[ji, jv] * Res_xylem_low[ji, jv]) + psi_xylem_collar_temp[ji, jv]
                    psi_xylem_leaf_temp[ji, jv] = -(Fxylem_up[ji, jv] * Res_xylem_up[ji, jv]) + psi_xylem_trunk_temp[ji, jv]
                    if soiltile[ji, jst] * vegtot[ji] > min_sechiba:
                        psi_leaf_temp[ji, jv] = -(transpir_test[ji, jv] * Res_mesophyll[ji, jv]) + psi_xylem_leaf_temp[ji, jv]
                    else:
                        psi_leaf_temp[ji, jv] = psi_xylem_leaf_temp[ji, jv]
                elif np.sum(circ_class_biomass[ji, jv, :, iroot, icarbon]) >= min_sechiba:
                    psi_xylem_trunk_temp[ji, jv] = -(Fxylem_low[ji, jv] * Res_xylem_low[ji, jv]) + psi_xylem_collar_temp[ji, jv]
                    psi_xylem_leaf_temp[ji, jv] = -(Fxylem_up[ji, jv] * Res_xylem_up[ji, jv]) + psi_xylem_trunk_temp[ji, jv]
                    psi_leaf_temp[ji, jv] = -(transpir_test[ji, jv] * Res_mesophyll[ji, jv]) + psi_xylem_leaf_temp[ji, jv]
                else:
                    psi_xylem_trunk_temp[ji, jv] = psi_xylem_collar_temp[ji, jv]
                    psi_xylem_leaf_temp[ji, jv] = psi_xylem_trunk_temp[ji, jv]
                    psi_leaf_temp[ji, jv] = psi_xylem_leaf_temp[ji, jv]
                psi_root_sup_temp[ji, jv] = np.minimum(psi_root_sup_temp[ji, jv], 0.0)
                psi_root_inf_temp[ji, jv] = np.minimum(psi_root_inf_temp[ji, jv], 0.0)
                psi_xylem_collar_temp[ji, jv] = np.minimum(psi_xylem_collar_temp[ji, jv], 0.0)
                psi_leaf_temp[ji, jv] = np.minimum(psi_leaf_temp[ji, jv], 0.0)
                if this_tstep_calc[ji, jv]:
                    psi_leaf[ji, jv] = psi_leaf_temp[ji, jv]
                    psi_xylem_leaf[ji, jv] = psi_xylem_leaf_temp[ji, jv]
                    psi_xylem_trunk[ji, jv] = psi_xylem_trunk_temp[ji, jv]
                    psi_xylem_collar[ji, jv] = psi_xylem_collar_temp[ji, jv]
                    psi_sto_leaf[ji, jv] = psi_sto_leaf_temp[ji, jv]
                    psi_sto_wood[ji, jv] = psi_sto_wood_temp[ji, jv]
                    psi_root_sup[ji, jv] = psi_root_sup_temp[ji, jv]
                    psi_root_inf[ji, jv] = psi_root_inf_temp[ji, jv]
                    Finf[ji, jv] = Finf_temp[ji, jv]
                    Fsup[ji, jv] = Fsup_temp[ji, jv]
                    mc_sup[ji, int(pref_soil_veg[jv])] = mc_sup_temp[ji, int(pref_soil_veg[jv])]
                    mc_inf[ji, int(pref_soil_veg[jv])] = mc_inf_temp[ji, int(pref_soil_veg[jv])]
                    mc_i_sup[ji, jv, :] = mc_i_sup_temp[ji, jv, :]
                    mc_i_inf[ji, jv, :] = mc_i_inf_temp[ji, jv, :]
                    Fi[ji, jv, :, :] = Fi_temp[ji, jv, :, :]
                    F_absorption[ji, jv] = F_absorption_temp[ji, jv]
                    if launch_next_calc[ji, jv]:
                        this_tstep_calc[ji, jv] = False
                        continue
                if not this_tstep_calc[ji, jv]:
                    if printlev > 4:
                        logging.info('"This Time step values"')
                        logging.info(f'"psi_leaf_temp = "{psi_leaf_temp[ji, jv]}')
                        logging.info(f'"transpir_test = "{transpir_test[ji, jv]}')
                        logging.info(f'"transpir_fsupinf ="{transpir[test_grid, test_pft] / (dt_sechiba * kilo_to_unit * soiltile[test_grid, int(pref_soil_veg[test_pft])] * vegtot[test_grid])}')
                        logging.info(f'"Fxylem_up ="{Fxylem_up[test_grid, test_pft]}')
                        logging.info(f'"Fsto_leaf ="{Fsto_leaf[test_grid, test_pft]}')
                        logging.info(f'"Fxylem_low ="{Fxylem_low[test_grid, test_pft]}')
                        logging.info(f'"Fsto_wood ="{Fsto_wood[test_grid, test_pft]}')
                        logging.info(f'"F_absorption_temp ="{F_absorption_temp[test_grid, test_pft]}')
                        logging.info(f'"Fsup_temp ="{Fsup_temp[test_grid, test_pft]}')
                        logging.info(f'"Finf_temp ="{Finf_temp[test_grid, test_pft]}')
                        logging.info(f'"psi_root_sup_temp ="{psi_root_sup_temp[test_grid, test_pft]}')
                        logging.info(f'"psi_root_inf_temp ="{psi_root_inf_temp[test_grid, test_pft]}')
                    if np.abs(psi_leaf_temp[ji, jv] - psi_leaf_temp_save[ji, jv]) > 0.005:
                        psi_leaf_temp[ji, jv] = (psi_leaf_temp_save[ji, jv] + psi_leaf_temp[ji, jv]) / 2.0
                    else:
                        break
            psi_leaf_next[ji, jv] = psi_leaf_temp[ji, jv]
            for jsl in range(0, nslm, 1):
                if np.sum(Fi[ji, jv, :, jst]) > 0.0:
                    e_frac[ji, jv, jsl, jst] = Fi[ji, jv, jsl, jst] / np.sum(Fi[ji, jv, :, jst])
                else:
                    e_frac[ji, jv, jsl, jst] = 0.0
    if printlev > 4:
        logging.info('"Next time step values"')
        logging.info(f'"dh ="{dh[:]}')
        logging.info(f'"SUM dh ="{np.sum(dh[:])}')
        logging.info(f'"zmaxh ="{zmaxh}')
        logging.info(f'"veget_max"{veget_max[:, :]}')
        logging.info(f'"psi_sto_leaf_save ="{psi_sto_leaf_save[test_grid, test_pft]}')
        logging.info(f'"psi_xylem_trunk ="{psi_xylem_trunk[test_grid, test_pft]}')
        logging.info(f'"Res_sto_leaf ="{Res_sto_leaf[test_grid, test_pft]}')
        logging.info(f'"Res_xylem_up ="{Res_xylem_up[test_grid, test_pft]}')
        logging.info(f'"transpir_fsupinf ="{transpir[test_grid, test_pft] / (dt_sechiba * kilo_to_unit * soiltile[test_grid, int(pref_soil_veg[test_pft])] * vegtot[test_grid])}')
        logging.info(f'"psi_sto_wood_save ="{psi_sto_wood_save[test_grid, test_pft]}')
        logging.info(f'"psi_xylem_collar ="{psi_xylem_collar[test_grid, test_pft]}')
        logging.info(f'"Res_xylem_low ="{Res_xylem_low[test_grid, test_pft]}')
        logging.info(f'"Fxylem_up ="{Fxylem_up[test_grid, test_pft]}')
        logging.info(f'"test_grid ="{test_grid}')
        logging.info(f'"test_pft ="{test_pft}')
        logging.info(f'"transpir ="{transpir[test_grid, test_pft]}')
        logging.info(f'"transpir_fsupinf ="{transpir[test_grid, test_pft] / (dt_sechiba * kilo_to_unit * soiltile[test_grid, int(pref_soil_veg[test_pft])] * vegtot[test_grid])}')
        logging.info(f'"Fxylem_up ="{Fxylem_up[test_grid, test_pft]}')
        logging.info(f'"Fsto_leaf ="{Fsto_leaf[test_grid, test_pft]}')
        logging.info(f'"Fxylem_low ="{Fxylem_low[test_grid, test_pft]}')
        logging.info(f'"Fsto_wood ="{Fsto_wood[test_grid, test_pft]}')
        logging.info(f'"F_absorption ="{F_absorption[test_grid, test_pft]}')
        logging.info(f'"mc_out ="{mc_out[test_grid, :, int(pref_soil_veg[test_pft])]}')
        logging.info(f'"mc_sup ="{mc_sup[test_grid, int(pref_soil_veg[test_pft])]}')
        logging.info(f'"mc_inf ="{mc_inf[test_grid, int(pref_soil_veg[test_pft])]}')
        logging.info(f'"Fsup ="{Fsup[test_grid, test_pft]}')
        logging.info(f'"Finf ="{Finf[test_grid, test_pft]}')
        logging.info(f'"Fi ="{Fi[test_grid, test_pft, :, int(pref_soil_veg[test_pft])]}')
        logging.info(f'"SUM(Fi) ="{np.sum(Fi[test_grid, test_pft, :, int(pref_soil_veg[test_pft])])}')
        logging.info(f'"e_frac ="{e_frac[test_grid, test_pft, :, int(pref_soil_veg[test_pft])]}')
        logging.info(f'"SUM(e_frac) ="{np.sum(e_frac[test_grid, test_pft, :, int(pref_soil_veg[test_pft])])}')
        logging.info(f'"psi_root_sup ="{psi_root_sup[test_grid, test_pft]}')
        logging.info(f'"psi_root_inf ="{psi_root_inf[test_grid, test_pft]}')
        logging.info(f'"psi_ss_inf ="{-(Finf[test_grid, test_pft] * Res_root_inf[test_grid, test_pft]) + psi_root_inf[test_grid, test_pft]}')
        logging.info(f'"psi_xylem_collar ="{psi_xylem_collar[test_grid, test_pft]}')
        logging.info(f'"psi_xylem_trunk ="{psi_xylem_trunk[test_grid, test_pft]}')
        logging.info(f'"psi_xylem_leaf ="{psi_xylem_leaf[test_grid, test_pft]}')
        logging.info(f'"psi_sto_wood ="{psi_sto_wood[test_grid, test_pft]}')
        logging.info(f'"psi_sto_leaf ="{psi_sto_leaf[test_grid, test_pft]}')
        logging.info(f'"psi_leaf ="{psi_leaf[test_grid, test_pft]}')
        logging.info(f'"psi_leaf_next ="{psi_leaf_next[test_grid, test_pft]}')
        logging.info(f'"Vr_leaf ="{Vr_leaf[test_grid, test_pft]}')
        logging.info(f'"Vmax_leaf ="{Vmax_leaf[test_grid, test_pft]}')
        logging.info(f'"V_leaf ="{V_leaf[test_grid, test_pft]}')
        logging.info(f'"C_leaf ="{C_leaf[test_grid, test_pft]}')
        logging.info(f'"tau_leaf ="{tau_leaf[test_grid, test_pft]}')
        logging.info(f'"Vr_wood ="{Vr_wood[test_grid, test_pft]}')
        logging.info(f'"Vmax_wood ="{Vmax_wood[test_grid, test_pft]}')
        logging.info(f'"V_wood ="{V_wood[test_grid, test_pft]}')
        logging.info(f'"C_wood ="{C_wood[test_grid, test_pft]}')
        logging.info(f'"tau_wood ="{tau_wood[test_grid, test_pft]}')
    logging.info('INFO: xios_orchidee_send_field: "Fsup"')
    logging.info('INFO: xios_orchidee_send_field: "Finf"')
    logging.info('INFO: xios_orchidee_send_field: "mc_sup"')
    logging.info('INFO: xios_orchidee_send_field: "mc_inf"')
    logging.info('INFO: xios_orchidee_send_field: "psi_leaf"')
    logging.info('INFO: xios_orchidee_send_field: "psi_xylem_trunk"')
    logging.info('INFO: xios_orchidee_send_field: "psi_xylem_leaf"')
    logging.info('INFO: xios_orchidee_send_field: "psi_xylem_collar"')
    logging.info('INFO: xios_orchidee_send_field: "psi_sto_wood"')
    logging.info('INFO: xios_orchidee_send_field: "psi_sto_leaf"')
    logging.info('INFO: xios_orchidee_send_field: "psi_root_sup"')
    logging.info('INFO: xios_orchidee_send_field: "psi_root_inf"')
```
