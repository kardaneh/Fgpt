```python
# Notebook Setup

%load_ext autoreload
%autoreload 2

import os

# Change to the Fgpt directory
%cd /home/kardaneh/Fgpt

# Fortran Parser
from fparser.two import Fortran2003 as F23
from fparser.two import Fortran2008 as F28
from fparser.two.utils import walk

# Fgpt Imports
from fgpt.core.frontend import Processor
from fgpt.core.frontend import Extractor
from fgpt.core.common import Logger

# Initialize Fgpt
logger = Logger()
processor = Processor(logger=logger)
```

```python
source = """
    a(j) = a(j) + j
"""
```

```python
tree = processor.parse_fortran_statement(source)
```

```python
from graphviz import Digraph


def tree_to_graph(node, graph=None, parent=None):
    if graph is None:
        graph = Digraph()
        graph.attr("node",
                   color="blue",        # box border
                   fontcolor="blue", 
                   shape="ellipse")
        graph.attr(
            "edge",
            color="blue"
        )
    # Create node
    node_id = str(id(node))

    # Show the class name
    label = type(node).__name__

    graph.node(node_id, label)

    if parent is not None:
        graph.edge(parent, node_id)

    # Recursively inspect children
    if hasattr(node, "content"):
        children = node.content
    elif hasattr(node, "items"):
        children = node.items
    else:
        children = []

    if children is None:
        return graph

    # fparser often stores children in tuples/lists
    if not isinstance(children, (list, tuple)):
        children = [children]

    for child in children:
        if child is not None:
            tree_to_graph(child, graph, node_id)

    return graph


graph = tree_to_graph(tree.children[0])

graph.render(
    "fparser_tree",
    format="png",
    cleanup=True
)

```
