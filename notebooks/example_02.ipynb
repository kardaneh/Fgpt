# Isolating All Depencencies of Procedure

This notebook demonstrates how to isolate all dependencies of procedure (subroutine or function) from their original module and generate the self-contained versions that can be compiled and executed independently.

```python
# Notebook Setup

%load_ext autoreload
%autoreload 2

import os
import sys

# Change to the Fgpt directory
%cd /home/kardaneh/Fgpt

# Fortran Parser
from fparser.two import Fortran2003 as F23
import argparse

# Fgpt Imports
from fgpt.core.frontend import Processor
from fgpt.core.frontend import Extractor
from fgpt import Isolator
from fgpt.isolator import parse_args
from fgpt.core.common import Logger

# Initialize Fgpt
logger = Logger()
processor = Processor(logger=logger)
```

```python
# Initiate the isolator
isolator = Isolator(
    rest_of_path="Fgpt/examples",
    target_module="example_01",
    work="/home/kardaneh",
    config_path="/home/kardaneh/Fgpt/template.yaml",
    openacc=False,
    tapenade=False,
    f2py=False,
    py2jx=False,
)
```

```python
# Define the options 
sys.argv = [
    "my_program",

    "--rest_of_path", "Fgpt/examples",
    "--target_module", "example_01",
    "--work", "/home/kardaneh",

    "--parent_subroutine", "test_procedures",

    "--target_subroutines",
    "set_data", 
    "display_data",
    "process_data", 
    "scale_data", 
    "dump_to_file", 
    "reset_data",
    "compute_average",

    "--openacc", "False",
    "--f2py", "False",
    "--py2jx", "False",
    "--tapenade", "False",

    "--mode", "jax",

    "--benchmark_dir", "/tmp/benchmark",

    "--config_path", "/home/kardaneh/Fgpt/template.yaml",

    "--vectorize",
    "kjpindex",
    "nvm",
    "npts",
]

args = parse_args()
isolator.run(
        benchmark_dir=args.benchmark_dir,
        vectorize=args.vectorize,
        mode=args.mode,
        parent_subroutine=args.parent_subroutine,
        target_subroutines=args.target_subroutines,
    )
```
